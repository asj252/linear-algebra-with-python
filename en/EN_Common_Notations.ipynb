## Logic

| Symbol | Meaning |
|------|------|
| $\{x \mid P(x)\}$| The set of all $x$ for which the statement $P(x)$ holds (the vertical bar $\mid$ is read "such that").|
| $\forall$,$\exists$|"For all" and "there exists". For example, $\forall x \in A,\ \exists y \in B$ is read "for every $x$ in $A$ there exists some $y$ in $B$".|
|$\Rightarrow$| "Implies" (if ... then ...)|
|$\Leftrightarrow$| "If and only if" (implication in both directions)|
| $\mid S\mid$ |The **cardinality** of the set $S$, that is, its number of elements (for a finite set, simply a count).  |




## Sets and Spaces

| Symbol | Meaning | English term |
|------|------|----------|
| $\mathbb{K}$ | Abstract field | Field (DE Körper) |
| $\mathbb{R}$ | The set of real numbers | Real numbers |
| $\mathbb{C}$ | The set of complex numbers | Complex numbers |
| $\mathbb{R}^n$ | $n$-dimensional real vector space; Cartesian coordinate system | n-dimensional real vector space, Cartesian coordinate system |
| $\mathbb{C}^n$ | $n$-dimensional complex vector space | n-dimensional complex vector space |
| $V, W$ | General vector spaces | General vector spaces |
| $\text{span}_{\mathbb{K}}\{\mathbf{v}_0, \mathbf{v}_1, \ldots, \mathbf{v}_{n-1}\}$ | The span of the vectors $\mathbf{v}_0, \mathbf{v}_1, \ldots, \mathbf{v}_{n-1}$ over the field $\mathbb{K}$ | Linear span over field K |
| $P(A)$ | The power set of the set $A$ | Power set of A |
| $\operatorname{card}(A)$ | The number of elements in the set $A$ | The number of elements in set A |
| $P_n(\mathbb{R})$ | The space of real polynomials of degree at most $n$ | Real polynomial space of degree at most n |
| $M_{m \times n}(\mathbb{R})$ | The space of $m \times n$ real matrices | m×n real matrix space |
| $S_n(\mathbb{R})$ | The space of $n \times n$ real symmetric matrices | n×n real symmetric matrix space |

## Vectors and Matrices

| Symbol | Meaning | English term |
|------|------|----------|
| $\mathbf{v}$ or $\mathbf{v}⟩$ or $\vec{v}$ | Vector (a column vector by default) | Vector |
| $\overrightarrow{\mathbf{uv}}$ | The line segment from the point $\mathbf{u}$ to the point $\mathbf{v}$ | Line segment from u to v |
| $\mathbf{v}^{\top}$ | The transpose of a vector | Transpose of vector |
| $⟨\mathbf{v}$ or $\mathbf{v}^{*}$ | The conjugate transpose of a vector | Conjugate transpose of vector |
| $\|\mathbf{v}\|$ or $\|\|\mathbf{v}\|\|$ | The $l_2$ length (modulus, norm) of a vector | Length / Norm of vector |
| $\|\|\mathbf{v}\|\|_{l_p}$ | The $l_p$ norm of a vector | l_p norm of vector |
| $\mathbf{0}$ | Zero vector; zero matrix | Zero vector, Zero matrix |
| $\mathbf{e}_i$ | Standard basis vector (its $i$-th component is 1 and all others are 0) | Standard basis vector |
| $\mathbf{A}$ | Matrix | Matrix |
|  $\|\|\mathbf{A}\|\|$ | The norm (modulus) of a matrix | Norm of matrix |
| $a_{ij}$ or $[\mathbf{A}]_{ij}$ or $(\mathbf{A})_{ij}$ | The entry of the matrix in row $i$ and column $j$ | Matrix element at row i, column j |
| $\mathbf{I}_n$ | The $n \times n$ identity matrix | n×n identity matrix |
| $\mathbf{A}^{\top}$ | The transpose of the matrix $\mathbf{A}$ | Transpose of matrix A |
| $\mathbf{A}^{*}$ | The conjugate transpose of the matrix $\mathbf{A}$ | Conjugate transpose of matrix A |
| $\mathbf{A}^{-1}$ | The inverse of the matrix $\mathbf{A}$ | Inverse of matrix A |
| $\mathbf{A}^{+}$ | The pseudoinverse of the matrix $\mathbf{A}$ | Pseudoinverse of matrix A |
| $\text{row}_i(\mathbf{A})$ | The $i$-th row of the matrix $\mathbf{A}$ | i-th row of matrix A |
| $\text{col}_j(\mathbf{A})$ | The $j$-th column of the matrix $\mathbf{A}$ | j-th column of matrix A |
| $\text{Row}(\mathbf{A})$ | The row space of the matrix $\mathbf{A}$ | Row space of matrix A |
| $\text{Col}(\mathbf{A})$ or $\text{Im}(\mathbf{A})$ | The column space (image) of the matrix $\mathbf{A}$ | Column space / Image of matrix A |
| $\det(\mathbf{A})$ or $\|\mathbf{A}\|$ | The determinant of the matrix $\mathbf{A}$ | Determinant of matrix A |
| $\text{tr}(\mathbf{A})$ | The trace of the matrix $\mathbf{A}$ | Trace of matrix A |

## Operations

| Symbol | Meaning | English term |
|------|------|----------|
|$:= , =:  $|Definitional equality (the colon marks the side being defined)|Definition|
| $\mathbf{u} \cdot \mathbf{v}$ or $\mathbf{u}^{\top} \mathbf{v}$ | Inner product (dot product) of real vectors | Inner product / Dot product of real vectors |
| $⟨\mathbf{u}\|\mathbf{v}⟩$ or $\mathbf{u}^{*} \mathbf{v}$ | Inner product of (complex) vectors | Inner product of (complex) vectors |
| $\mathbf{u} \times \mathbf{v}$ | Cross product of vectors, defined only in (two and) three dimensions | Cross product / Vector product |
| $\mathbf{A} \circ \mathbf{B}$ | Hadamard product (entrywise multiplication) | Hadamard product / Element-wise product |
| $\mathbf{A} \otimes \mathbf{B}$ | Kronecker product | Kronecker product |
| $\mathbf{u}\mathbf{v}^*$ or $\|\mathbf{u}\rangle\langle\mathbf{v}\|$ | Outer product (a rank-1 matrix) | Outer product |
| $\text{Proj}_{\mathbf{v}}(\mathbf{u})$ or $\frac{\|\mathbf{v}\rangle\langle\mathbf{v}\|\mathbf{u}\rangle}{\langle\mathbf{v}\|\mathbf{v}\rangle}$ | The projection of the vector $\mathbf{u}$ onto the vector $\mathbf{v}$ | Projection of u onto v |

## Mappings and Subspaces

| Symbol | Meaning | English term |
|------|------|----------|
| $f: X \rightarrow Y$ | A mapping from the set $X$ to the set $Y$ | Function / Mapping from X to Y |
| $T: V \rightarrow W$ | A linear transformation from the vector space $V$ to $W$ | Linear transformation from V to W |
| $T_{\mathbf{A}}: V \rightarrow W$ | The linear transformation from $V$ to $W$ represented by the matrix $\mathbf{A}$ | Linear transformation represented by matrix A |
| $\ker(T)$ | The kernel (null space) of the linear transformation $T$ | Kernel / Null space of T |
| $\text{Im}(T)$ | The image of the linear transformation $T$ | Image / Range of T |
| $f[A]$ | The image of the set $A$ under the mapping $f$ | Image of set A under f |
| $f^{-1}[B]$ | The preimage of the set $B$ under the mapping $f$ | Preimage of set B under f |
| $\text{rank}(\mathbf{A})$ or $\dim \text{Im}(T_{\mathbf{A}})$ or $\dim \text{Im}(\mathbf{A})$ | The rank of the matrix $\mathbf{A}$ (the dimension of its image) | Rank of matrix A |
| $\text{nullity}(\mathbf{A})$ or $\dim \ker(T_{\mathbf{A}})$ or $\dim \ker(\mathbf{A})$ | The nullity of the matrix $\mathbf{A}$ (the dimension of its kernel) | Nullity of matrix A |

| $\mathbf{A}\mathbf{v} = \lambda\mathbf{v}$ | The eigenvalue equation | Eigenvalue equation |
| $\det(\mathbf{A}-\lambda\mathbf{I})=0$ | The characteristic polynomial equation | Characteristic polynomial equation |
