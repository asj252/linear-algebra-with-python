## Logic

| Symbol | Meaning |
|------|------|
| $\{x \mid P(x)\}$| The set of all $x$ for which the statement $P(x)$ holds (the vertical bar $\mid$ is read "such that").|
| $\forall$,$\exists$|"For all" and "there exists". For example, $\forall x \in A,\ \exists y \in B$ is read "for every $x$ in $A$ there exists some $y$ in $B$".|
|$\Rightarrow$| "Implies" (if ... then ...)|
|$\Leftrightarrow$| "If and only if" (implication in both directions)|
| $\mid S\mid$ |The **cardinality** of the set $S$, that is, its number of elements (for a finite set, simply a count).  |




## Sets and Spaces

| Symbol | Meaning |
|------|------|
| $\mathbb{K}$ | Abstract field (the symbol comes from German *Körper*) |
| $\mathbb{R}$ | The set of real numbers |
| $\mathbb{C}$ | The set of complex numbers |
| $\mathbb{R}^n$ | $n$-dimensional real vector space; Cartesian coordinate system |
| $\mathbb{C}^n$ | $n$-dimensional complex vector space |
| $V, W$ | General vector spaces |
| $\text{span}_{\mathbb{K}}\{\mathbf{v}_0, \mathbf{v}_1, \ldots, \mathbf{v}_{n-1}\}$ | The span of the vectors $\mathbf{v}_0, \mathbf{v}_1, \ldots, \mathbf{v}_{n-1}$ over the field $\mathbb{K}$ |
| $P(A)$ | The power set of the set $A$ |
| $\operatorname{card}(A)$ | The number of elements in the set $A$ |
| $P_n(\mathbb{R})$ | The space of real polynomials of degree at most $n$ |
| $M_{m \times n}(\mathbb{R})$ | The space of $m \times n$ real matrices |
| $S_n(\mathbb{R})$ | The space of $n \times n$ real symmetric matrices |

## Vectors and Matrices

| Symbol | Meaning |
|------|------|
| $\mathbf{v}$ or $\mathbf{v}⟩$ or $\vec{v}$ | Vector (a column vector by default) |
| $\overrightarrow{\mathbf{uv}}$ | The line segment from the point $\mathbf{u}$ to the point $\mathbf{v}$ |
| $\mathbf{v}^{\top}$ | The transpose of a vector |
| $⟨\mathbf{v}$ or $\mathbf{v}^{*}$ | The conjugate transpose of a vector |
| $\|\mathbf{v}\|$ or $\|\|\mathbf{v}\|\|$ | The $l_2$ length (modulus, norm) of a vector |
| $\|\|\mathbf{v}\|\|_{l_p}$ | The $l_p$ norm of a vector |
| $\mathbf{0}$ | Zero vector; zero matrix |
| $\mathbf{e}_i$ | Standard basis vector (its $i$-th component is 1 and all others are 0) |
| $\mathbf{A}$ | Matrix |
|  $\|\|\mathbf{A}\|\|$ | The norm (modulus) of a matrix |
| $a_{ij}$ or $[\mathbf{A}]_{ij}$ or $(\mathbf{A})_{ij}$ | The entry of the matrix in row $i$ and column $j$ |
| $\mathbf{I}_n$ | The $n \times n$ identity matrix |
| $\mathbf{A}^{\top}$ | The transpose of the matrix $\mathbf{A}$ |
| $\mathbf{A}^{*}$ | The conjugate transpose of the matrix $\mathbf{A}$ |
| $\mathbf{A}^{-1}$ | The inverse of the matrix $\mathbf{A}$ |
| $\mathbf{A}^{+}$ | The pseudoinverse of the matrix $\mathbf{A}$ |
| $\text{row}_i(\mathbf{A})$ | The $i$-th row of the matrix $\mathbf{A}$ |
| $\text{col}_j(\mathbf{A})$ | The $j$-th column of the matrix $\mathbf{A}$ |
| $\text{Row}(\mathbf{A})$ | The row space of the matrix $\mathbf{A}$ |
| $\text{Col}(\mathbf{A})$ or $\text{Im}(\mathbf{A})$ | The column space (image) of the matrix $\mathbf{A}$ |
| $\det(\mathbf{A})$ or $\|\mathbf{A}\|$ | The determinant of the matrix $\mathbf{A}$ |
| $\text{tr}(\mathbf{A})$ | The trace of the matrix $\mathbf{A}$ |

## Operations

| Symbol | Meaning |
|------|------|
|$:= , =:  $|Definitional equality (the colon marks the side being defined)|
| $\mathbf{u} \cdot \mathbf{v}$ or $\mathbf{u}^{\top} \mathbf{v}$ | Inner product (dot product) of real vectors |
| $⟨\mathbf{u}\|\mathbf{v}⟩$ or $\mathbf{u}^{*} \mathbf{v}$ | Inner product of (complex) vectors |
| $\mathbf{u} \times \mathbf{v}$ | Cross product of vectors, defined only in (two and) three dimensions |
| $\mathbf{A} \circ \mathbf{B}$ | Hadamard product (entrywise multiplication) |
| $\mathbf{A} \otimes \mathbf{B}$ | Kronecker product |
| $\mathbf{u}\mathbf{v}^*$ or $\|\mathbf{u}\rangle\langle\mathbf{v}\|$ | Outer product (a rank-1 matrix) |
| $\text{Proj}_{\mathbf{v}}(\mathbf{u})$ or $\frac{\|\mathbf{v}\rangle\langle\mathbf{v}\|\mathbf{u}\rangle}{\langle\mathbf{v}\|\mathbf{v}\rangle}$ | The projection of the vector $\mathbf{u}$ onto the vector $\mathbf{v}$ |

## Mappings and Subspaces

| Symbol | Meaning |
|------|------|
| $f: X \rightarrow Y$ | A mapping from the set $X$ to the set $Y$ |
| $T: V \rightarrow W$ | A linear transformation from the vector space $V$ to $W$ |
| $T_{\mathbf{A}}: V \rightarrow W$ | The linear transformation from $V$ to $W$ represented by the matrix $\mathbf{A}$ |
| $\ker(T)$ | The kernel (null space) of the linear transformation $T$ |
| $\text{Im}(T)$ | The image of the linear transformation $T$ |
| $f[A]$ | The image of the set $A$ under the mapping $f$ |
| $f^{-1}[B]$ | The preimage of the set $B$ under the mapping $f$ |
| $\text{rank}(\mathbf{A})$ or $\dim \text{Im}(T_{\mathbf{A}})$ or $\dim \text{Im}(\mathbf{A})$ | The rank of the matrix $\mathbf{A}$ (the dimension of its image) |
| $\text{nullity}(\mathbf{A})$ or $\dim \ker(T_{\mathbf{A}})$ or $\dim \ker(\mathbf{A})$ | The nullity of the matrix $\mathbf{A}$ (the dimension of its kernel) |

## Eigenvalues and Eigenvectors

| Symbol | Meaning |
|------|------|
| $\lambda$ | Eigenvalue |
| $\mathbf{v}_\lambda$ | The eigenvector corresponding to the eigenvalue $\lambda$ |
| $\mathbf{A}\mathbf{v} = \lambda\mathbf{v}$ | The eigenvalue equation |
| $\det(\mathbf{A}-\lambda\mathbf{I})=0$ | The characteristic polynomial equation |
