```python
import numpy as np
from itertools import combinations
np.set_printoptions(precision=4, suppress=True, linewidth=100)

# ── Visual hierarchy helpers ─────────────────────────────────────────
def print_header(title):
    print("=" * 60)
    print(f"  {title}")
    print("=" * 60)

def print_step(step, desc):
    print(f"\n▶ Step {step}: {desc}")
    print("-" * 40)

def compare_print(label, actual, expected_desc):
    print(f"\n[{label}]")
    print(f"  Computed value: {actual}")
    print(f"  Expected value: {expected_desc}")

# ── Math utilities ─────────────────────────────────────────
def proj(u, v):
    """Projection vector of v onto the direction of u"""
    return np.dot(u, v) / np.dot(u, u) * u

def para_area(u, v):
    """Area of the parallelogram spanned by u, v"""
    return np.sqrt(np.dot(u,u)*np.dot(v,v) - np.dot(u,v)**2)

def triple(u, v, w):
    """Signed volume D(u,v,w) = u·(v×w)"""
    return np.dot(u, np.cross(v, w))
```

::::{exercise} Computing a Linear Combination
:label: exer-ch2-linear-comb-calc

In $\mathbb{R}^4$, let $\mathbf{a} = (2, 0, -1, 3)$, $\mathbf{b} = (1, 4, 2, -1)$, $\mathbf{c} = (-1, 2, 0, 1)$.

Compute $2\mathbf{a} - 3\mathbf{b} + \mathbf{c}$ and $\|2\mathbf{a} - 3\mathbf{b} + \mathbf{c}\|$.

**Hint.** Compute component by component, then use $\|\mathbf{v}\| = \sqrt{v_0^2 + v_1^2 + v_2^2 + v_3^2}$.

:::{solution} exer-ch2-linear-comb-calc
:class: dropdown

**Step 1.** Expand component by component:

$$
2\mathbf{a} = (4, 0, -2, 6), \quad
3\mathbf{b} = (3, 12, 6, -3), \quad
\mathbf{c} = (-1, 2, 0, 1)
$$

**Step 2.** Add:

$$
2\mathbf{a} - 3\mathbf{b} + \mathbf{c}
= (4-3-1,\ 0-12+2,\ -2-6+0,\ 6+3+1)
= (0,\ -10,\ -8,\ 10)
$$

**Step 3.** Compute the norm:

$$
\|2\mathbf{a} - 3\mathbf{b} + \mathbf{c}\|
= \sqrt{0^2 + (-10)^2 + (-8)^2 + 10^2}
= \sqrt{0 + 100 + 64 + 100}
= \sqrt{264} = 2\sqrt{66}
$$
:::
::::

```python
# ============================================================
print_header("Exercise 1 | Computing a Linear Combination")
# ============================================================

print_step(1, "Define the vectors a, b, c ∈ ℝ⁴")
a = np.array([2, 0, -1, 3], float)
b = np.array([1, 4,  2, -1], float)
c = np.array([-1, 2, 0,  1], float)
print("a =", a)
print("b =", b)
print("c =", c)

print_step(2, "Compute 2a - 3b + c")
result = 2*a - 3*b + c
print("2a - 3b + c =", result)

compare_print("‖2a - 3b + c‖", f"{np.linalg.norm(result):.6f}", f"2√66 = {2*np.sqrt(66):.6f}")
```

::::{exercise} The Average of Cyclically Shifted Vectors
:label: exer-ch2-cyclic-avg

In $\mathbb{R}^n$, define the vectors

$$
\mathbf{a}_0 = (1,2,\ldots,n),\quad
\mathbf{a}_1 = (2,3,\ldots,n,1),\quad
\ldots,\quad
\mathbf{a}_{n-1} = (n,1,\ldots,n-2,n-1)
$$

($\mathbf{a}_i$ is obtained by cyclically shifting $\mathbf{a}_0$ to the left $i$ times.)

Compute the average vector $\mathbf{A} = \dfrac{1}{n}(\mathbf{a}_0 + \mathbf{a}_1 + \cdots + \mathbf{a}_{n-1})$, and find $\|\mathbf{A} - \mathbf{a}_i\|$ for $i = 0, 1, \ldots, n-1$.

**Hint.** Observe that in each component position, the corresponding entries of the $n$ vectors are exactly some arrangement of $1, 2, \ldots, n$, so all components have the same sum. Once you have found $\|\mathbf{A} - \mathbf{a}_i\|^2$, note that $i$ only affects where the summation starts, so the result does not depend on $i$.

:::{solution} exer-ch2-cyclic-avg
:class: dropdown

**Step 1. Compute the average vector.**

The $j$-th component of $\mathbf{a}_i$ (indexed from $0$, with $i,j \in \{0,1,\ldots,n-1\}$) is

$$
(\mathbf{a}_i)_j = ((i+j) \bmod n) + 1
$$

Fix $j$ and sum over $i = 0, \ldots, n-1$: $((i+j) \bmod n)$ runs through $0, 1, \ldots, n-1$, so

$$
\sum_{i=0}^{n-1}(\mathbf{a}_i)_j = (0+1+\cdots+(n-1)) + n = \frac{n(n+1)}{2}
$$

Therefore

$$
\mathbf{A} = \frac{1}{n}\sum_{i=0}^{n-1}\mathbf{a}_i = \frac{n+1}{2}(1,1,\ldots,1)
$$

**Step 2. Compute $\|\mathbf{A} - \mathbf{a}_i\|^2$.**

$$
(\mathbf{A} - \mathbf{a}_i)_j = \frac{n+1}{2} - \bigl(((i+j) \bmod n)+1\bigr) = \frac{n-1}{2} - ((i+j) \bmod n)
$$

As $j$ runs through $0,\ldots,n-1$, $((i+j) \bmod n)$ also runs through $0,\ldots,n-1$ (regardless of $i$), so

$$
\|\mathbf{A} - \mathbf{a}_i\|^2
= \sum_{k=0}^{n-1}\left(\frac{n-1}{2} - k\right)^2
= \sum_{m=-(n-1)/2}^{(n-1)/2} m^2
= \frac{n(n^2-1)}{12}
$$

(The last step expands and simplifies using $\sum_{k=0}^{n-1}k^2 = \frac{n(n-1)(2n-1)}{6}$.)

**Step 3. Conclusion.**

$$
\|\mathbf{A} - \mathbf{a}_i\| = \sqrt{\frac{n(n^2-1)}{12}}, \quad i = 0, 1, \ldots, n-1
$$

All the vectors are at the **same** distance from the average vector, reflecting the cyclic symmetry.
:::
::::

```python
# ============================================================
print_header("Exercise 2 | The Average of Cyclically Shifted Vectors (check with n=4)")
# ============================================================

print_step(1, "Generate the cyclically shifted vectors a_0, …, a_{n-1}")
n = 4
vecs = [np.roll(np.arange(1, n+1, dtype=float), -i) for i in range(n)]
for i, v in enumerate(vecs):
    print(f"  a_{i} =", v)

print_step(2, "Compute the average vector A")
A_avg = np.mean(vecs, axis=0)
compare_print("average vector A", A_avg, f"(n+1)/2 times the all-ones vector = {(n+1)/2} * [1,1,1,1]")

print_step(3, "Check that all the vectors are equidistant from A")
dists = [np.linalg.norm(A_avg - v) for v in vecs]
print("  ‖A - a_i‖ =", [f"{d:.4f}" for d in dists])
compare_print("theoretical value √(n(n²−1)/12)", f"{dists[0]:.6f}", f"{np.sqrt(n*(n**2-1)/12):.6f}")
```

::::{exercise} The Fourth Vertex of a Parallelogram
:label: exer-ch2-parallelogram-vertex

In $\mathbb{R}^3$, let $A(2,-1,3)$, $B(1,4,-2)$, $C(-2,6,1)$. If $A$, $B$, $C$, $D$ are the four vertices of a parallelogram, find all possible positions of $D$, and compute the side lengths of the parallelogram in each case.

**Hint.** The two diagonals of a parallelogram have the same midpoint. There are three ways to choose the pair of diagonals, $(AC,BD)$, $(AB,CD)$, $(BC,AD)$, each corresponding to one position of $D$.

:::{solution} exer-ch2-parallelogram-vertex
:class: dropdown

**Step 1.** Take the diagonals to be $(AC, BD)$. Equal midpoints, $\dfrac{\mathbf{A}+\mathbf{C}}{2}=\dfrac{\mathbf{B}+\mathbf{D}}{2}$, give

$$
\mathbf{D}_0 = \mathbf{A} + \mathbf{C} - \mathbf{B} = (2,-1,3)+(-2,6,1)-(1,4,-2) = (-1, 1, 6)
$$

Side lengths: $\|\mathbf{A}-\mathbf{B}\|=\sqrt{1+25+25}=\sqrt{51}$, $\|\mathbf{B}-\mathbf{C}\|=\sqrt{22}$.

**Step 2.** Take the diagonals to be $(AB, CD)$. From $\dfrac{\mathbf{A}+\mathbf{B}}{2}=\dfrac{\mathbf{C}+\mathbf{D}}{2}$ we get

$$
\mathbf{D}_1 = \mathbf{A} + \mathbf{B} - \mathbf{C} = (2,-1,3)+(1,4,-2)-(-2,6,1) = (5, -3, 0)
$$

Side lengths: $\|\mathbf{A}-\mathbf{C}\|=\sqrt{69}$, $\|\mathbf{B}-\mathbf{C}\|=\sqrt{9+4+9}=\sqrt{22}$.

**Step 3.** Take the diagonals to be $(BC, AD)$. From $\dfrac{\mathbf{B}+\mathbf{C}}{2}=\dfrac{\mathbf{A}+\mathbf{D}}{2}$ we get

$$
\mathbf{D}_2 = \mathbf{B} + \mathbf{C} - \mathbf{A} = (1,4,-2)+(-2,6,1)-(2,-1,3) = (-3, 11, -4)
$$

Side lengths: $\|\mathbf{A}-\mathbf{B}\|=\sqrt{51}$, $\|\mathbf{A}-\mathbf{C}\|=\sqrt{69}$.

**Conclusion.** There are three possibilities: $D_0=(-1,1,6)$, $D_1=(5,-3,0)$, $D_2=(-3,11,-4)$.
:::
::::

```python
# ============================================================
print_header("Exercise 3 | The Fourth Vertex of a Parallelogram")
# ============================================================

print_step(1, "Define the three vertices A, B, C ∈ ℝ³")
A = np.array([2, -1,  3], float)
B = np.array([1,  4, -2], float)
C = np.array([-2, 6,  1], float)

print_step(2, "Compute the three possible fourth vertices D (by choice of diagonals)")
D0 = A + C - B   # diagonal AC
D1 = A + B - C   # diagonal AB
D2 = B + C - A   # diagonal BC

for label, D in zip(["D0 (diagonal AC)", "D1 (diagonal AB)", "D2 (diagonal BC)"], [D0, D1, D2]):
    print(f"  {label}: D =", D)

print_step(3, "Check the side lengths in the case D0 (|AB| = |CD0|)")
print("  |AB| =", np.linalg.norm(B-A))
print("  |CD0|=", np.linalg.norm(D0-C))
print("  equal?", np.isclose(np.linalg.norm(B-A), np.linalg.norm(D0-C)))
```

::::{exercise} Existence of a Linear Combination of Vectors
:label: exer-ch2-lincomb-existence

Decide whether the vector $\mathbf{x}$ can be written as $\mathbf{x} = s\mathbf{u} + t\mathbf{v}$. If it can, find the coefficients $s, t$.

1. $\mathbf{u}=(1,2)$, $\mathbf{v}=(3,-1)$, $\mathbf{x}=(4,1)$.
2. $\mathbf{u}=(1,2,3)$, $\mathbf{v}=(2,4,6)$, $\mathbf{x}=(3,6,9)$.
3. $\mathbf{u}=(1,0,2,-1)$, $\mathbf{v}=(0,1,-1,3)$, $\mathbf{x}=(2,3,0,8)$.
4. $\mathbf{u}=(1,2,0,-1)$, $\mathbf{v}=(2,4,1,0)$, $\mathbf{x}=(1,2,2,0)$.

**Hint.** Write the condition as a system of linear equations and decide whether it has a solution. When there are more components than unknowns, the equations for all the components must be consistent.

:::{solution} exer-ch2-lincomb-existence
:class: dropdown

**Step 1 (part 1).** Solve $s+3t=4$, $2s-t=1$. The second equation gives $s=(1+t)/2$; substituting into the first: $(1+t)/2+3t=4 \Rightarrow t=1, s=1$. Check: $(1,2)+(3,-1)=(4,1)$ ✓.

**Step 2 (part 2).** Note that $\mathbf{v}=2\mathbf{u}$, so $\text{span}\{\mathbf{u},\mathbf{v}\}=\text{span}\{\mathbf{u}\}$. Since $\mathbf{x}=3\mathbf{u}$, there are infinitely many solutions: $s+2t=3$ (any $t$, $s=3-2t$). For example, $s=3, t=0$ or $s=1, t=1$.

**Step 3 (part 3).** Set $s=2$ (component 0) and $t=3$ (component 1). Check component 2: $2(2)+3(-1)=1\neq 0$. **No solution.**

**Step 4 (part 4).** Component 2 gives $t=2$; component 3 gives $-s=0\Rightarrow s=0$. Check component 0: $0+2(2)=4\neq 1$. **No solution.**
:::
::::

```python
# ============================================================
print_header("Exercise 4 | Existence of Linear Combinations")
# ============================================================

print_step(1, "(a) s(1,2)+t(3,−1)=(4,1) — unique solution")
M = np.array([[1,3],[2,-1]], float)
st = np.linalg.solve(M, [4,1])
print("  [s, t] =", st)
compare_print("check M @ [s,t]", M @ st, "[4, 1]")

print_step(2, "(b) x = 3u — infinitely many solutions")
u, x = np.array([1,2,3],float), np.array([3,6,9],float)
print("  x = 3u?", np.allclose(3*u, x), "(s+2t=3, infinitely many solutions)")

print_step(3, "(c) check that there is no solution")
u = np.array([1,0,2,-1],float); v = np.array([0,1,-1,3],float)
x = np.array([2,3,0,8],float)
print("  s=2,t=3 → su+tv =", 2*u+3*v)
print("  target x =", x, "→ no solution")

print_step(4, "(d) least-squares residual → no solution")
u = np.array([1,2,0,-1],float); v = np.array([2,4,1,0],float)
x = np.array([1,2,2,0],float)
coef, *_ = np.linalg.lstsq(np.column_stack([u,v]), x, rcond=None)
residual = np.linalg.norm(np.column_stack([u,v]) @ coef - x)
compare_print("residual ‖Ac−x‖", f"{residual:.6f}", "> 0 → no solution")
```

::::{exercise} Rewriting a System of Equations as a Linear Combination of Vectors
:label: exer-ch2-system-vector-form

Rewrite each of the following systems of linear equations as a linear combination of vectors, $x_0\mathbf{u}_0 + x_1\mathbf{u}_1 + \cdots = \mathbf{b}$ (indices start at $0$).

1. $\begin{cases} x_0 + 2x_1 = 3 \\ 2x_0 - x_1 = 1 \end{cases}$

2. $\begin{cases} x_0 + x_1 + x_2 = 4 \\ 2x_0 + 3x_2 = 5 \\ -x_0 + 2x_1 + x_2 = 1 \end{cases}$

3. $\begin{cases} 2x_1 - x_2 = 0 \\ 2x_0 + 3x_2 = 0 \\ -x_0 + 4x_1 = 0 \end{cases}$

4. $\begin{cases} x_0 + 2x_1 + x_3 = 3 \\ 2x_0 - x_1 = 1 \\ -x_0 + 3x_1 = 7 \\ 3x_0 = 4 \end{cases}$

**Hint.** Regard each column of the coefficient matrix as a vector; $x_i$ is the coefficient of the corresponding column.

:::{solution} exer-ch2-system-vector-form
:class: dropdown

**Part 1.**

$$
x_0\begin{pmatrix}1\\2\end{pmatrix} + x_1\begin{pmatrix}2\\-1\end{pmatrix} = \begin{pmatrix}3\\1\end{pmatrix}
$$

**Part 2.**

$$
x_0\begin{pmatrix}1\\2\\-1\end{pmatrix} + x_1\begin{pmatrix}1\\0\\2\end{pmatrix} + x_2\begin{pmatrix}1\\3\\1\end{pmatrix} = \begin{pmatrix}4\\5\\1\end{pmatrix}
$$

**Part 3.**

$$
x_0\begin{pmatrix}0\\2\\-1\end{pmatrix} + x_1\begin{pmatrix}2\\0\\4\end{pmatrix} + x_2\begin{pmatrix}-1\\3\\0\end{pmatrix} = \begin{pmatrix}0\\0\\0\end{pmatrix}
$$

**Part 4.**

$$
x_0\begin{pmatrix}1\\2\\-1\\3\end{pmatrix} + x_1\begin{pmatrix}2\\-1\\3\\0\end{pmatrix} + x_2\begin{pmatrix}0\\0\\0\\0\end{pmatrix} + x_3\begin{pmatrix}1\\0\\0\\0\end{pmatrix} = \begin{pmatrix}3\\1\\7\\4\end{pmatrix}
$$

(In part 4, $x_2$ does not appear, so its column vector is $\mathbf{0}$.)
:::
::::

```python
# ============================================================
print_header("Exercise 5 | Rewriting Systems of Equations as Linear Combinations of Vectors")
# ============================================================

print_step(1, "(a) x₀(1,2)+x₁(2,−1) = (3,1)")
A_mat = np.array([[1,2],[2,-1]], float)
x_sol = np.linalg.solve(A_mat, [3,1])
print("  [x₀, x₁] =", x_sol)
compare_print("check A·x", A_mat @ x_sol, "[3, 1]")

print_step(2, "(b) 3×3 system")
A_mat = np.array([[1,1,1],[2,0,3],[-1,2,1]], float)
x_sol = np.linalg.lstsq(A_mat, [4,5,1], rcond=None)[0]
print("  solution =", x_sol)
compare_print("check A·x", A_mat @ x_sol, "[4, 5, 1]")
```

::::{exercise} Division Points of a Segment
:label: exer-ch2-section-point

Find the point $P$ that divides the segment between two points $A$, $B$ (or its extension), and express it as a linear combination of vectors.

1. $A=(1,2)$, $B=(5,6)$, and $P$ divides $\overline{AB}$ internally with $\overline{AP}:\overline{PB}=2:1$.
2. $A=(2,-1)$, $B=(6,3)$, and $P$ divides $\overline{AB}$ externally with $\overline{AP}:\overline{PB}=3:1$.
3. $A=(1,0,2)$, $B=(4,3,-1)$:
   - (a) $P$ divides $\overline{AB}$ internally with $\overline{AP}:\overline{PB}=3:5$; find $P$.
   - (b) $A$ divides $\overline{QB}$ externally with $\overline{QA}:\overline{AB}=5:2$; find $Q$.

**Hint.**
- Internal division ($m:n$): $P = \dfrac{n\mathbf{A}+m\mathbf{B}}{m+n}$.
- External division ($m:n$, $m\neq n$): $P = \dfrac{-n\mathbf{A}+m\mathbf{B}}{m-n}$; pay attention to the direction of the signs.

:::{solution} exer-ch2-section-point
:class: dropdown

**Step 1 (part 1, internal division $2:1$).**

$$
P = \frac{1\cdot\mathbf{A}+2\cdot\mathbf{B}}{2+1} = \frac{(1,2)+2(5,6)}{3} = \frac{(11,14)}{3} = \left(\frac{11}{3},\frac{14}{3}\right)
$$

Vector form: $P = \dfrac{1}{3}\mathbf{A} + \dfrac{2}{3}\mathbf{B}$.

**Step 2 (part 2, external division $3:1$).**

$$
P = \frac{-1\cdot\mathbf{A}+3\cdot\mathbf{B}}{3-1} = \frac{-(2,-1)+3(6,3)}{2} = \frac{(16,10)}{2} = (8,5)
$$

Vector form: $P = -\dfrac{1}{2}\mathbf{A} + \dfrac{3}{2}\mathbf{B}$.

**Step 3 (part 3a, internal division $3:5$).**

$$
P = \frac{5\mathbf{A}+3\mathbf{B}}{8} = \frac{5(1,0,2)+3(4,3,-1)}{8} = \frac{(17,9,7)}{8} = \left(\frac{17}{8},\frac{9}{8},\frac{7}{8}\right)
$$

**Step 4 (part 3b, finding $Q$).**

We know that $Q$ lies on the ray from $A$ through $B$ and that $\overline{QA}:\overline{AB}=5:2$. Hence $\mathbf Q-\mathbf A=\dfrac52(\mathbf B-\mathbf A)$, which gives

$$
\mathbf{Q}=\frac{5\mathbf B-3\mathbf A}{2}=\frac{5(4,3,-1)-3(1,0,2)}{2}=\frac{(17,15,-11)}{2}
$$
:::
::::

```python
# ============================================================
print_header("Exercise 6 | Division Points of a Segment")
# ============================================================

print_step(1, "(1) ℝ²: A=(1,2), B=(5,6), internal division 2:1")
A, B = np.array([1,2],float), np.array([5,6],float)
P = (1*A + 2*B) / 3
print("  P =", P)

print_step(2, "(2) ℝ²: A=(2,−1), B=(6,3), external division 3:1")
A, B = np.array([2,-1],float), np.array([6,3],float)
P = (-1*A + 3*B) / 2
print("  P =", P)

print_step(3, "(3a) ℝ³: A=(1,0,2), B=(4,3,−1), internal division 3:5")
A, B = np.array([1,0,2],float), np.array([4,3,-1],float)
P = (5*A + 3*B) / 8
print("  P =", P)

print_step(4, "(3b) external division point Q: QA:AB = 5:2")
Q = (5*B - 3*A) / 2
print("  Q =", Q)
t_A = np.dot(A-Q, B-Q) / np.dot(B-Q, B-Q)
ratio = np.linalg.norm(Q-A) / np.linalg.norm(A-B)
assert np.allclose(A, Q + t_A*(B-Q))
assert np.isclose(ratio, 5/2) and not (0 <= t_A <= 1)
print("  QA/AB =", ratio, "; parameter of A on the line QB =", t_A, "(external division)")
```

::::{exercise} Division Points on a Parametrized Line
:label: exer-ch2-line-section

Let the line be given parametrically by $\mathbf{x}(t) = \mathbf{A} + t(\mathbf{B}-\mathbf{A})$, $t\in\mathbb{R}$.

1. $A=(0,-10,1,-3)$, $B=(5,0,6,7)$. Let $P$ lie on this line so that $P$ divides $\overline{AB}$ internally with $\overline{AP}:\overline{PB}=1:4$; find $\mathbf{P}$.
2. $A=(2,1,-1,0)$, $B=(6,3,2,4)$. Let $Q$ divide $\overline{AB}$ externally with $\overline{AQ}:\overline{QB}=4:3$; find $\mathbf{Q}$.
3. $A=(1,1,\ldots,1)\in\mathbb{R}^n$, $B=(2,2,\ldots,2)\in\mathbb{R}^n$:
   - (a) Find $P$ such that $P$ divides $\overline{AB}$ internally in the ratio $1:k$ ($k>0$).
   - (b) Which point does $P$ approach as $k\to\infty$? What about $k\to 0$?

**Hint.** Internal division in the ratio $1:k$ corresponds to $t=\dfrac{1}{1+k}$; for external division, $t>1$ or $t<0$.

:::{solution} exer-ch2-line-section
:class: dropdown

**Step 1 (part 1).** Internal division $1:4$, $t=\dfrac{1}{5}$:

$$
\mathbf{P} = \mathbf{A} + \frac{1}{5}(\mathbf{B}-\mathbf{A}) = \frac{4\mathbf{A}+\mathbf{B}}{5} = \frac{4(0,-10,1,-3)+(5,0,6,7)}{5} = \frac{(5,-40,10,-5)}{5} = (1,-8,2,-1)
$$

**Step 2 (part 2).** External division $4:3$, $t=\dfrac{4}{4-3}=4$ ($Q$ lies beyond $B$):

$$
\mathbf{Q} = \mathbf{A} + 4(\mathbf{B}-\mathbf{A}) = -3\mathbf{A}+4\mathbf{B} = -3(2,1,-1,0)+4(6,3,2,4) = (18,9,11,16)
$$

**Step 3 (part 3a).** $t=\dfrac{1}{1+k}$:

$$
P = \mathbf{A}+\frac{1}{1+k}(\mathbf{B}-\mathbf{A}) = \mathbf{A}+\frac{1}{1+k}(1,\ldots,1) = \left(1+\frac{1}{1+k}\right)(1,\ldots,1) = \frac{2+k}{1+k}(1,\ldots,1)
$$

**Step 4 (part 3b).** As $k\to\infty$: $P\to 1\cdot(1,\ldots,1)=\mathbf{A}$, so the division point approaches the starting point $A$. As $k\to 0$: $P\to 2(1,\ldots,1)=\mathbf{B}$, so the division point approaches $B$.
:::
::::

### The Centroid of a Triangle

::::{exercise} Discovering the Centroid in $\mathbb{R}^4$
:label: exer-ch2-centroid-discovery

In $\mathbb{R}^4$, consider the three vertices (position vectors) of a triangle:

$$
\mathbf{A}=(1,2,0,-1),\quad \mathbf{B}=(3,0,2,1),\quad \mathbf{C}=(-1,4,1,3)
$$

1. Verify that $\mathbf{M}_C=\dfrac{\mathbf{A}+\mathbf{B}}{2}$ is the midpoint of $AB$ ($\|\mathbf{A}-\mathbf{M}_C\|=\|\mathbf{B}-\mathbf{M}_C\|$), and find the midpoints $\mathbf{M}_A$, $\mathbf{M}_B$ of $BC$ and $AC$.
2. Write down the parametric equations of the medians $\mathbf{L}_A(t)=(1-t)\mathbf{A}+t\mathbf{M}_A$ and $\mathbf{L}_B(s)=(1-s)\mathbf{B}+s\mathbf{M}_B$.
3. Setting $\mathbf{L}_A(t)=\mathbf{L}_B(s)$, find the coordinates of the intersection point $\mathbf{G}$ of the two medians.
4. Verify that $\mathbf{C}$, $\mathbf{G}$, $\mathbf{M}_C$ are collinear.
5. Verify that $\mathbf{G}=\dfrac{\mathbf{A}+\mathbf{B}+\mathbf{C}}{3}$.
6. Verify that $\|\mathbf{A}-\mathbf{G}\|:\|\mathbf{M}_A-\mathbf{G}\|=2:1$.

**Hint.** In part 3, compare components to obtain $t$ and $s$, then substitute into either expression to find $\mathbf{G}$. In part 4, verify that $\mathbf{G}-\mathbf{C}=\lambda(\mathbf{M}_C-\mathbf{C})$.

:::{solution} exer-ch2-centroid-discovery
:class: dropdown

**Step 1.** Compute the midpoints:

$$
\mathbf{M}_C = \frac{\mathbf{A}+\mathbf{B}}{2} = (2,1,1,0),\quad
\mathbf{M}_A = \frac{\mathbf{B}+\mathbf{C}}{2} = (1,2,\tfrac{3}{2},2),\quad
\mathbf{M}_B = \frac{\mathbf{A}+\mathbf{C}}{2} = (0,3,\tfrac{1}{2},1)
$$

Check: $\mathbf{M}_C - \mathbf{A} = (1,-1,1,1)$ and $\mathbf{B}-\mathbf{M}_C=(1,-1,1,1)$, so $\|\mathbf{A}-\mathbf{M}_C\|=\|\mathbf{B}-\mathbf{M}_C\|=2$ ✓.

**Step 2.** The median $\mathbf{L}_A(t)$:

$$
\mathbf{L}_A(t) = (1-t)(1,2,0,-1)+t(1,2,\tfrac{3}{2},2) = \left(1,\, 2,\, \tfrac{3t}{2},\, -1+3t\right)
$$

The median $\mathbf{L}_B(s)$:

$$
\mathbf{L}_B(s) = (1-s)(3,0,2,1)+s(0,3,\tfrac{1}{2},1) = \left(3-3s,\, 3s,\, 2-\tfrac{3s}{2},\, 1\right)
$$

**Step 3.** Set $\mathbf{L}_A(t)=\mathbf{L}_B(s)$ and compare components:
- Component 0: $1=3-3s \Rightarrow s=2/3$
- Component 1: $2=3s \Rightarrow s=2/3$ ✓
- Component 3: $-1+3t=1 \Rightarrow t=2/3$

Substituting into $\mathbf{L}_A(2/3)$:

$$
\mathbf{G} = \left(1, 2, 1, 1\right)
$$

**Step 4.** $\mathbf{G}-\mathbf{C}=(2,-2,0,-2)$ and $\mathbf{M}_C-\mathbf{C}=(3,-3,0,-3)$. Hence $\mathbf{G}-\mathbf{C}=\dfrac{2}{3}(\mathbf{M}_C-\mathbf{C})$, so the points are collinear ✓.

**Step 5.** $\dfrac{\mathbf{A}+\mathbf{B}+\mathbf{C}}{3}=\dfrac{(3,6,3,3)}{3}=(1,2,1,1)=\mathbf{G}$ ✓.

**Step 6.** $\|\mathbf{A}-\mathbf{G}\|=\|(0,0,-1,-2)\|=\sqrt{5}$ and $\|\mathbf{M}_A-\mathbf{G}\|=\|(0,0,1/2,1)\|=\dfrac{\sqrt{5}}{2}$. The ratio is $= 2:1$ ✓.
:::
::::

```python
# ============================================================
print_header("Exercise 8 | Discovering the Centroid in ℝ⁴")
# ============================================================

print_step(1, "Define the three vertices and compute the midpoints of the sides")
A = np.array([1,2,0,-1],float)
B = np.array([3,0,2, 1],float)
C = np.array([-1,4,1, 3],float)
MC = (A+B)/2; MA = (B+C)/2; MB = (A+C)/2
print("  MA =", MA, "  MB =", MB, "  MC =", MC)

print_step(2, "Check that MC is the midpoint of AB (A and B are equidistant from MC)")
compare_print("|A−MC| vs |B−MC|",
    f"{np.linalg.norm(A-MC):.6f}",
    f"|B−MC| = {np.linalg.norm(B-MC):.6f} (should be equal)")

print_step(3, "Centroid G = (A+B+C)/3")
G = (A+B+C)/3
print("  G =", G)
print("  ratio of the distances to the centroid |A−G|/|MA−G| =",
      np.linalg.norm(A-G)/np.linalg.norm(MA-G))
compare_print("|A−G|/|MA−G|",
    f"{np.linalg.norm(A-G)/np.linalg.norm(MA-G):.6f}", "2")

print_step(4, "The three points C, G, MC are collinear: λ = 2/3")
diff = MC - C
lam = np.dot(G-C, diff) / np.dot(diff, diff)
print(f"  λ = {lam:.6f}  error = {np.linalg.norm((G-C) - lam*(MC-C)):.2e}")
```

::::{exercise} A General Proof for the Centroid of a Triangle in $\mathbb{R}^n$
:label: exer-ch2-centroid-proof

Let $\mathbf A,\mathbf B,\mathbf C$ be the position vectors of three non-collinear points in $\mathbb R^n$, and let

$$
\mathbf{G}=\frac{\mathbf{A}+\mathbf{B}+\mathbf{C}}{3},\quad
\mathbf{M}_A=\frac{\mathbf{B}+\mathbf{C}}{2},\quad
\mathbf{M}_B=\frac{\mathbf{A}+\mathbf{C}}{2},\quad
\mathbf{M}_C=\frac{\mathbf{A}+\mathbf{B}}{2}
$$

1. Prove that $\mathbf{A}$, $\mathbf{G}$, $\mathbf{M}_A$ are collinear, that is, there exists $\lambda$ such that $\mathbf{G}-\mathbf{A}=\lambda(\mathbf{M}_A-\mathbf{A})$; prove similarly that $\mathbf{B}$, $\mathbf{G}$, $\mathbf{M}_B$ are collinear and that $\mathbf{C}$, $\mathbf{G}$, $\mathbf{M}_C$ are collinear.
2. Prove that $\|\mathbf A-\mathbf G\|=2\|\mathbf M_A-\mathbf G\|$; only when $\mathbf M_A\neq\mathbf A$ is this written as the ratio $2:1$. The other two cases are similar.

**Hint.** Compute $\mathbf{G}-\mathbf{A}$ and $\mathbf{M}_A-\mathbf{A}$ directly and find the proportionality between them.

:::{solution} exer-ch2-centroid-proof
:class: dropdown

**1. Collinearity.**

$$
\mathbf{G}-\mathbf{A} = \frac{\mathbf{A}+\mathbf{B}+\mathbf{C}}{3}-\mathbf{A} = \frac{\mathbf{B}+\mathbf{C}-2\mathbf{A}}{3}
$$

$$
\mathbf{M}_A-\mathbf{A} = \frac{\mathbf{B}+\mathbf{C}}{2}-\mathbf{A} = \frac{\mathbf{B}+\mathbf{C}-2\mathbf{A}}{2}
$$

Hence $\mathbf{G}-\mathbf{A} = \dfrac{2}{3}(\mathbf{M}_A-\mathbf{A})$, that is, $\lambda=\dfrac{2}{3}$, and $\mathbf{A}$, $\mathbf{G}$, $\mathbf{M}_A$ are collinear. The same holds for $\mathbf{B}$ and $\mathbf{C}$ (by symmetry).

**2. The ratio.**

From the above, $\mathbf{G}-\mathbf{A} = \dfrac{2}{3}(\mathbf{M}_A-\mathbf{A})$, so

$$
\mathbf{G}-\mathbf{M}_A = \mathbf{G}-\mathbf{A}-(\mathbf{M}_A-\mathbf{A}) = \left(\frac{2}{3}-1\right)(\mathbf{M}_A-\mathbf{A}) = -\frac{1}{3}(\mathbf{M}_A-\mathbf{A})
$$

Therefore

$$
\|\mathbf{A}-\mathbf{G}\| = \frac{2}{3}\|\mathbf{M}_A-\mathbf{A}\|,\quad \|\mathbf{M}_A-\mathbf{G}\| = \frac{1}{3}\|\mathbf{M}_A-\mathbf{A}\|
$$

$$
\|\mathbf A-\mathbf G\|=2\|\mathbf M_A-\mathbf G\|
$$

The equality of lengths above includes the degenerate case; only when the corresponding median is nonzero can we divide to obtain $2:1$. The cases of $\mathbf B,\mathbf C$ are similar.
:::
::::

```python
# ============================================================
print_header("Exercise 9 | General Proof for the Centroid in ℝⁿ (numerical check)")
# ============================================================

print_step(1, "Random checks of collinearity and the ratio for n = 3, 5, 10")
rng = np.random.default_rng(42)
for n in [3, 5, 10]:
    A = rng.standard_normal(n); B = rng.standard_normal(n); C = rng.standard_normal(n)
    G  = (A+B+C)/3
    MA = (B+C)/2
    err = np.linalg.norm((G-A) - (2/3)*(MA-A))
    ratio = np.linalg.norm(A-G) / np.linalg.norm(MA-G)
    print(f"  n={n:2d}: collinearity error={err:.2e}  |A−G|/|MA−G| = {ratio:.6f} (should be 2)")
```

::::{exercise} The Law of Cosines and the Inner Product
:label: exer-ch2-cosine-inner-product

Let $\mathbf{A}=(a_0,\ldots,a_{n-1})$, $\mathbf{B}=(b_0,\ldots,b_{n-1})\in\mathbb{R}^n$, and $\mathbf{A}\cdot\mathbf{B}=\sum a_ib_i$.

1. Prove by computation that $\|\mathbf{A}-\mathbf{B}\|^2 = \|\mathbf{A}\|^2+\|\mathbf{B}\|^2-2\mathbf{A}\cdot\mathbf{B}$.
2. When $n=2$ or $n=3$ and $\mathbf A,\mathbf B$ are both nonzero, use the geometric law of cosines to prove that $\mathbf A\cdot\mathbf B=\|\mathbf A\|\|\mathbf B\|\cos\theta$.
3. When $n=3$, why does the law of cosines also hold for a triangle in space?
4. When $n>3$ and $\mathbf A,\mathbf B$ are both nonzero:
   - (a) For nonzero vectors $\mathbf A,\mathbf B$, define the angle $\theta$ by $\cos\theta=\dfrac{\mathbf{A}\cdot\mathbf{B}}{\|\mathbf{A}\|\|\mathbf{B}\|}$, $0\leq\theta\leq\pi$. Why is the right-hand side $\in[-1,1]$?
   - (b) Using this definition, prove that the law of cosines still holds when $n>3$.

**Hint.** Part 4(a) relies on the Cauchy-Schwarz inequality; 4(b) follows simply by substituting the definition.

:::{solution} exer-ch2-cosine-inner-product
:class: dropdown

**Part 1.**

$$
\|\mathbf{A}-\mathbf{B}\|^2 = (\mathbf{A}-\mathbf{B})\cdot(\mathbf{A}-\mathbf{B}) = \mathbf{A}\cdot\mathbf{A} - 2\mathbf{A}\cdot\mathbf{B}+\mathbf{B}\cdot\mathbf{B} = \|\mathbf{A}\|^2+\|\mathbf{B}\|^2-2\mathbf{A}\cdot\mathbf{B}
$$

**Part 2.** In the plane ($n=2$), for the triangle with vertices $O$, $A$, $B$, the geometric law of cosines gives $|OA|^2+|OB|^2-2|OA||OB|\cos\theta = |AB|^2$, that is, $\|\mathbf{A}\|^2+\|\mathbf{B}\|^2-2\|\mathbf{A}\|\|\mathbf{B}\|\cos\theta = \|\mathbf{A}-\mathbf{B}\|^2$. Comparing with part 1 gives $\mathbf{A}\cdot\mathbf{B}=\|\mathbf{A}\|\|\mathbf{B}\|\cos\theta$.

**Part 3.** Three points $O,A,B$ in space always lie in at least one common plane; applying the two-dimensional law of cosines within that plane suffices. The collinear case can be checked directly with $\theta=0$ or $\pi$.

**Part 4(a).** By the Cauchy-Schwarz inequality, $|\mathbf{A}\cdot\mathbf{B}|\leq\|\mathbf{A}\|\|\mathbf{B}\|$, so $\dfrac{\mathbf{A}\cdot\mathbf{B}}{\|\mathbf{A}\|\|\mathbf{B}\|}\in[-1,1]$ and $\theta$ is well-defined.

**Part 4(b).** By definition, $\mathbf{A}\cdot\mathbf{B}=\|\mathbf{A}\|\|\mathbf{B}\|\cos\theta$; substituting into part 1 gives $\|\mathbf{A}-\mathbf{B}\|^2 = \|\mathbf{A}\|^2+\|\mathbf{B}\|^2-2\|\mathbf{A}\|\|\mathbf{B}\|\cos\theta$, which is the law of cosines in higher dimensions.
:::
::::

```python
# ============================================================
print_header("Exercise 14 | The Law of Cosines and the Inner Product")
# ============================================================

print_step(1, "Define the vectors A, B ∈ ℝ⁴")
A = np.array([1,2,3,4],float); B = np.array([1,-1,0,2],float)

print_step(2, "Check ‖A−B‖² = ‖A‖² + ‖B‖² − 2A·B")
lhs = np.linalg.norm(A-B)**2
rhs = np.dot(A,A) + np.dot(B,B) - 2*np.dot(A,B)
compare_print("‖A−B‖²", f"{lhs:.4f}", f"‖A‖²+‖B‖²−2A·B = {rhs:.4f}  ✓ {np.isclose(lhs,rhs)}")

print_step(3, "Compute the angle cosθ (Cauchy–Schwarz guarantees ∈[−1,1])")
cos_t = np.dot(A,B)/(np.linalg.norm(A)*np.linalg.norm(B))
compare_print("cosθ", f"{cos_t:.6f}", "∈ [−1, 1]")
```

::::{exercise} General Properties of Projection
:label: exer-ch2-proj-properties

Let $\mathbf{u}\neq\mathbf{0}$.

1. If $\|\mathbf{u}\|=1$, prove that $\text{proj}_{\mathbf{u}}\mathbf{v}=(\mathbf{v}\cdot\mathbf{u})\mathbf{u}$.
2. Prove that the vector decomposition $\mathbf{v}=\text{proj}_{\mathbf{u}}\mathbf{v}+(\mathbf{v}-\text{proj}_{\mathbf{u}}\mathbf{v})$ satisfies $\text{proj}_{\mathbf{u}}\mathbf{v}\perp(\mathbf{v}-\text{proj}_{\mathbf{u}}\mathbf{v})$.
3. Using this decomposition, prove that the distance from $\mathbf{v}$ to the line through the origin with direction $\mathbf{u}$ equals

$$
\frac{\left(\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2\right)^{1/2}}{\|\mathbf{u}\|}
$$

**Hint.** For part 3: the distance equals $\|\mathbf{v}-\text{proj}_\mathbf{u}\mathbf{v}\|$; then expand and compute.

:::{solution} exer-ch2-proj-properties
:class: dropdown

**1.** If $\|\mathbf{u}\|=1$, then $\text{proj}_\mathbf{u}\mathbf{v}=\dfrac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|^2}\mathbf{u}=(\mathbf{v}\cdot\mathbf{u})\mathbf{u}$. ■

**2.** Let $\mathbf{p}=\text{proj}_\mathbf{u}\mathbf{v}=\dfrac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|^2}\mathbf{u}$ and $\mathbf{r}=\mathbf{v}-\mathbf{p}$. Compute:

$$
\mathbf{p}\cdot\mathbf{r}=\frac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|^2}\mathbf{u}\cdot(\mathbf{v}-\mathbf{p})
=\frac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|^2}\left(\mathbf{u}\cdot\mathbf{v}-\frac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|^2}\|\mathbf{u}\|^2\right)=0
$$

Hence $\mathbf{p}\perp\mathbf{r}$. ■

**3.** Since $\mathbf{v}=\mathbf{p}+\mathbf{r}$ and $\mathbf{p}\perp\mathbf{r}$ (Pythagorean theorem):

$$
\|\mathbf{v}\|^2=\|\mathbf{p}\|^2+\|\mathbf{r}\|^2
\Rightarrow \|\mathbf{r}\|^2 = \|\mathbf{v}\|^2-\|\mathbf{p}\|^2 = \|\mathbf{v}\|^2 - \frac{(\mathbf{u}\cdot\mathbf{v})^2}{\|\mathbf{u}\|^2}
=\frac{\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2}{\|\mathbf{u}\|^2}
$$

Taking the positive square root gives the result. Geometrically, the numerator, which the Cauchy-Schwarz inequality guarantees to be nonnegative, is the area of the parallelogram spanned by the two vectors; dividing by the length of the base gives the height. ■
:::
::::

```python
# ============================================================
print_header("Exercise 17 | General Properties of Projection")
# ============================================================

print_step(1, "Randomly generate u, v ∈ ℝ⁴ and compute the projection and the residual")
rng = np.random.default_rng(7)
u = rng.standard_normal(4); v = rng.standard_normal(4)
p = proj(u, v); r = v - p

print_step(2, "Check p ⊥ r")
compare_print("p·r (should be ≈0)", f"{np.dot(p,r):.2e}", "0")

print_step(3, "Check the distance formula ‖v−proj_u(v)‖")
dist1 = np.sqrt(np.dot(u,u)*np.dot(v,v) - np.dot(u,v)**2) / np.linalg.norm(u)
dist2 = np.linalg.norm(r)
compare_print("value of the distance formula", f"{dist1:.6f}", f"direct computation ‖r‖ = {dist2:.6f}  ✓ {np.isclose(dist1,dist2)}")
```

In $\mathbb{R}^3$, the plane $\Sigma$ through the origin has the parametric equation $\mathbf{p}(s,t)=s(1,-1,0)+t(0,1,-1)$. Given $\mathbf{v}=(0,0,4)$, find $\text{proj}_\Sigma\mathbf{v}$ by the following steps.

3. Verify that $\text{proj}_\Sigma\mathbf{v}\in\Sigma$ (that is, it is perpendicular to $\mathbf{n}$).

:::{solution} exer-ch2-proj-plane-r3
:class: dropdown

**Step 1.** Let $\mathbf{u}_0=(1,-1,0)$, $\mathbf{u}_1=(0,1,-1)$. The normal vector is

$$
\mathbf{n}=\mathbf{u}_0\times\mathbf{u}_1 = \begin{vmatrix}\mathbf{e}_0&\mathbf{e}_1&\mathbf{e}_2\\1&-1&0\\0&1&-1\end{vmatrix}=(1,1,1)

(Here $\begin{vmatrix}\cdots\end{vmatrix}$ is the symbolic notation for a determinant; in this setting it amounts to the cofactor expansion of the cross product along row 0 and is a convenient way to compute the cross product. The formal definition of the determinant is given in Chapter 7.)

**Step 2.**

$$
\text{proj}_\mathbf{n}\mathbf{v}=\frac{\mathbf{n}\cdot\mathbf{v}}{\|\mathbf{n}\|^2}\mathbf{n}=\frac{4}{3}(1,1,1)=\left(\frac{4}{3},\frac{4}{3},\frac{4}{3}\right)
$$

$$
\text{proj}_\Sigma\mathbf{v}=\mathbf{v}-\text{proj}_\mathbf{n}\mathbf{v}=\left(-\frac{4}{3},-\frac{4}{3},\frac{8}{3}\right)
$$

**Step 3.** $\text{proj}_\Sigma\mathbf{v}\cdot\mathbf{n}=-\tfrac{4}{3}-\tfrac{4}{3}+\tfrac{8}{3}=0$ ✓.
:::

### The Cross Product

::::{exercise} Basic Properties of the Cross Product
:label: exer-ch2-cross-product-basic

Let $\mathbf{u}=(u_0,u_1,u_2)$, $\mathbf{v}=(v_0,v_1,v_2)$, and define the cross product by

$$
\mathbf{u}\times\mathbf{v}=\left(
\begin{vmatrix}u_1&u_2\\v_1&v_2\end{vmatrix},\;
-\begin{vmatrix}u_0&u_2\\v_0&v_2\end{vmatrix},\;
\begin{vmatrix}u_0&u_1\\v_0&v_1\end{vmatrix}
\right)
$$

:::{note} Notation: 2×2 Determinants
:label: note-2x2-determinant-shorthand
In the formula above, $\begin{vmatrix}a&b\\c&d\end{vmatrix}$ is shorthand for a **2×2 determinant**, defined by

$$\begin{vmatrix}a&b\\c&d\end{vmatrix}=ad-bc$$

This chapter uses it only as a shorthand for writing the components of the cross product and does not yet involve any of its properties; the formal definition and full theory of determinants are given in Chapter 7.
:::

1. Compute and prove that $\mathbf{u}\times\mathbf{v}$ is perpendicular to $\text{span}\{\mathbf{u},\mathbf{v}\}$.
2. Prove the area formula: $\|\mathbf{u}\times\mathbf{v}\|^2=\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2$.
3. Assume $\mathbf u,\mathbf v$ are both nonzero. Using the angle definition from the inner product, prove that $\|\mathbf{u}\times\mathbf{v}\|=\|\mathbf{u}\|\|\mathbf{v}\||\sin\theta|$.
4. If $\mathbf{u}$ and $\mathbf{v}$ are linearly independent, explain why the area of the parallelogram spanned by $\mathbf{u}$ and $\mathbf{v}$ equals $\|\mathbf{u}\times\mathbf{v}\|$.

**Hint.** For part 1, compute $(\mathbf{u}\times\mathbf{v})\cdot\mathbf{u}$ directly and expand; for part 2, expand both sides and compare.

:::{solution} exer-ch2-cross-product-basic
:class: dropdown

**Step 1 (perpendicularity).** Let $\mathbf{w}=\mathbf{u}\times\mathbf{v}=(w_0,w_1,w_2)$ and compute directly:

$$
\mathbf{w}\cdot\mathbf{u}=u_0(u_1v_2-u_2v_1)-u_1(u_0v_2-u_2v_0)+u_2(u_0v_1-u_1v_0)=0
$$

(After expansion, the terms cancel exactly.) Similarly, $\mathbf{w}\cdot\mathbf{v}=0$.

**Step 2 (area formula).** Expand $\|\mathbf{u}\times\mathbf{v}\|^2=(u_1v_2-u_2v_1)^2+(u_0v_2-u_2v_0)^2+(u_0v_1-u_1v_0)^2$; after expanding and collecting terms, it equals $\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2$.

**Step 3 (sine).** From part 2 and $\cos\theta=\dfrac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|\|\mathbf{v}\|}$:

$$
\|\mathbf{u}\times\mathbf{v}\|^2=\|\mathbf{u}\|^2\|\mathbf{v}\|^2(1-\cos^2\theta)=\|\mathbf{u}\|^2\|\mathbf{v}\|^2\sin^2\theta
$$

**Step 4 (area of the parallelogram).** The area of the parallelogram is $=$ base$\times$height $=\|\mathbf{u}\|\cdot\|\mathbf{v}\||\sin\theta|=\|\mathbf{u}\times\mathbf{v}\|$.
:::
::::

```python
# ============================================================
print_header("Exercise 24 | Basic Properties of the Cross Product")
# ============================================================

print_step(1, "Compute u×v and check perpendicularity")
u = np.array([1.,2.,3.]); v = np.array([1.,0.,1.])
w = np.cross(u, v)
print("  u×v =", w)
compare_print("(u×v)·u (should be 0)", f"{np.dot(w,u):.2e}", "0")
compare_print("(u×v)·v (should be 0)", f"{np.dot(w,v):.2e}", "0")

print_step(2, "Check the area formula ‖u×v‖² = ‖u‖²‖v‖²−(u·v)²")
lhs = np.dot(w,w)
rhs = np.dot(u,u)*np.dot(v,v) - np.dot(u,v)**2
compare_print("‖u×v‖²", f"{lhs:.4f}", f"‖u‖²‖v‖²−(u·v)² = {rhs:.4f}  ✓ {np.isclose(lhs,rhs)}")

print_step(3, "The area agrees with para_area")
compare_print("‖u×v‖", f"{np.linalg.norm(w):.6f}", f"para_area(u,v) = {para_area(u,v):.6f}")
```

::::{exercise} Deciding Cross Product Identities
:label: exer-ch2-cross-identities

For each of the following cross product identities, prove it if it is true, and give a counterexample if it is false.

1. $\mathbf{u}\times\mathbf{u}=\mathbf{0}$
2. $\mathbf{u}\times\mathbf{v}=\mathbf{v}\times\mathbf{u}$
3. $\mathbf{u}\times(\mathbf{v}+\mathbf{w})=\mathbf{u}\times\mathbf{v}+\mathbf{u}\times\mathbf{w}$
4. $\mathbf{u}\times(\mathbf{v}\times\mathbf{w})=(\mathbf{u}\times\mathbf{v})\times\mathbf{w}$
5. $\mathbf{u}\cdot(\mathbf{v}\times\mathbf{w})=-\mathbf{w}\cdot(\mathbf{v}\times\mathbf{u})$

:::{solution} exer-ch2-cross-identities
:class: dropdown

**1. True.** Every component of $\mathbf{u}\times\mathbf{u}$ is $u_iv_j-u_jv_i|_{v=u}=0$.

**2. False.** Counterexample: $\mathbf{u}=\mathbf{e}_0=(1,0,0)$, $\mathbf{v}=\mathbf{e}_1=(0,1,0)$: $\mathbf{u}\times\mathbf{v}=(0,0,1)$, $\mathbf{v}\times\mathbf{u}=(0,0,-1)$. In fact $\mathbf{u}\times\mathbf{v}=-(\mathbf{v}\times\mathbf{u})$ (anticommutativity).

**3. True.** Expand each component directly: $(u_1(v_2+w_2)-u_2(v_1+w_1))$ and so on equals $(u_1v_2-u_2v_1)+(u_1w_2-u_2w_1)$. This holds for all three components, so distributivity over addition holds.

**4. False.** Take $\mathbf u=\mathbf v=\mathbf e_0=(1,0,0)$ and $\mathbf w=\mathbf e_1=(0,1,0)$. The left-hand side is $\mathbf u\times(\mathbf v\times\mathbf w)=\mathbf e_0\times\mathbf e_2=(0,-1,0)$; the right-hand side is $(\mathbf u\times\mathbf v)\times\mathbf w=\mathbf0$. The two sides are not equal, so the cross product is not associative.

**5. True.** $\mathbf{u}\cdot(\mathbf{v}\times\mathbf{w})$ is the signed volume, which has cyclic symmetry: $D(\mathbf{u},\mathbf{v},\mathbf{w})=D(\mathbf{v},\mathbf{w},\mathbf{u})=D(\mathbf{w},\mathbf{u},\mathbf{v})$, and exchanging two adjacent vectors changes the sign. $-\mathbf{w}\cdot(\mathbf{v}\times\mathbf{u})=\mathbf{w}\cdot(\mathbf{u}\times\mathbf{v})=D(\mathbf{w},\mathbf{u},\mathbf{v})=D(\mathbf{u},\mathbf{v},\mathbf{w})$ ✓.
:::
::::

```python
# ============================================================
print_header("Exercise 25 | Deciding Cross Product Identities")
# ============================================================

u=np.array([1.,0.,0.]); v=np.array([0.,1.,0.]); w=np.array([0.,0.,1.])

print_step(1, "(1) u×u = 0 (alternating property)")
compare_print("u×u", np.cross(u,u), "[0, 0, 0]")

print_step(2, "(2) u×v ≠ v×u (counterexample showing antisymmetry)")
print("  u×v =", np.cross(u,v), "  v×u =", np.cross(v,u))
compare_print("u×v = v×u? (should be False)", np.allclose(np.cross(u,v), np.cross(v,u)), "False")

print_step(3, "(3) distributivity u×(v+w) = u×v + u×w")
compare_print("distributivity holds?", np.allclose(np.cross(u,v+w), np.cross(u,v)+np.cross(u,w)), "True")

print_step(4, "(4) the cross product is not associative (counterexample)")
u4 = np.array([1.,0.,0.]); v4 = u4.copy(); w4 = np.array([0.,1.,0.])   # counterexample: u=v=e₀, w=e₁
print("  u×(v×w) =", np.cross(u4,np.cross(v4,w4)))
print("  (u×v)×w =", np.cross(np.cross(u4,v4),w4))
compare_print("associativity holds? (should be False)",
    np.allclose(np.cross(u4,np.cross(v4,w4)), np.cross(np.cross(u4,v4),w4)), "False")

print_step(5, "(5) cyclic symmetry: D(u,v,w) = −D(w,v,u)")
compare_print("D(u,v,w) vs −D(w,v,u)",
    f"{triple(u,v,w):.4f}", f"{-triple(w,v,u):.4f}")
```

::::{exercise} A Vector Proof of the Law of Sines
:label: exer-ch2-sine-rule

In $\mathbb{R}^3$, consider the triangle spanned by linearly independent vectors $\mathbf{a}$ and $\mathbf{b}$, and let $\mathbf{c}=\mathbf{a}-\mathbf{b}$. Starting from $\mathbf{c}\times\mathbf{c}=\mathbf{0}$, prove that the three sides of this triangle satisfy the law of sines.

**Hint.** Expanding $(\mathbf{a}-\mathbf{b})\times(\mathbf{a}-\mathbf{b})=\mathbf{0}$ gives $\mathbf{a}\times\mathbf{b}=-\mathbf{b}\times\mathbf{a}$... Then derive the result from the equation $\|\mathbf{a}\times\mathbf{b}\|=\|\mathbf{a}\|\|\mathbf{b}\|\sin C$.

:::{solution} exer-ch2-sine-rule
:class: dropdown

**Step 1.** Expanding, $\mathbf{c}\times\mathbf{c}=(\mathbf{a}-\mathbf{b})\times(\mathbf{a}-\mathbf{b})=\mathbf{a}\times\mathbf{a}-\mathbf{a}\times\mathbf{b}-\mathbf{b}\times\mathbf{a}+\mathbf{b}\times\mathbf{b}=\mathbf{0}$, so $-\mathbf{a}\times\mathbf{b}-\mathbf{b}\times\mathbf{a}=\mathbf{0}$, that is, $\mathbf{a}\times\mathbf{b}=-\mathbf{b}\times\mathbf{a}$ (anticommutativity, already known).

**Step 2.** The triangle has vertices $O,\mathbf a,\mathbf b$; let $\mathbf c=\mathbf a-\mathbf b$. Let $A,B,C$ denote the angles opposite the sides of lengths $\|\mathbf a\|,\|\mathbf b\|,\|\mathbf c\|$, respectively; that is, $A=\angle(-\mathbf b,\mathbf c)$, $B=\angle(-\mathbf a,-\mathbf c)$, $C=\angle(\mathbf a,\mathbf b)$.

**Step 3.** Using the common area $S>0$,

$$2S=\|\mathbf b\|\|\mathbf c\|\sin A=\|\mathbf a\|\|\mathbf c\|\sin B=\|\mathbf a\|\|\mathbf b\|\sin C.$$

**Step 4.** Dividing by the product of the three side lengths (all nonzero) and taking reciprocals, we obtain

$$\frac{\|\mathbf a\|}{\sin A}=\frac{\|\mathbf b\|}{\sin B}=\frac{\|\mathbf c\|}{\sin C}.$$

$\square$

::::{exercise} Properties of the Signed Volume
:label: exer-ch2-triple-product-props

*This exercise is in fact a three-dimensional preview of the determinant axioms of Chapter 7*: the following five properties (normalization, multilinearity, antisymmetry, cyclic symmetry, and the degeneracy criterion) are exactly the axiomatic framework that Chapter 7 uses to define the determinant. Once you understand them thoroughly in familiar three-dimensional geometry, the abstract definition in Chapter 7 will come with a ready-made geometric anchor.

Let $D(\mathbf{u},\mathbf{v},\mathbf{w})=\mathbf{u}\cdot(\mathbf{v}\times\mathbf{w})$ for $\mathbf{u},\mathbf{v},\mathbf{w}\in\mathbb{R}^3$. Prove:

1. $D(\mathbf{e}_0,\mathbf{e}_1,\mathbf{e}_2)=1$
2. $D(\alpha\mathbf{u}_0+\beta\mathbf{u}_1,\mathbf{v},\mathbf{w})=\alpha D(\mathbf{u}_0,\mathbf{v},\mathbf{w})+\beta D(\mathbf{u}_1,\mathbf{v},\mathbf{w})$ (linearity in the first argument)
3. $D(\mathbf{u},\mathbf{v},\mathbf{w})=-D(\mathbf{v},\mathbf{u},\mathbf{w})$
4. $D(\mathbf{u},\mathbf{v},\mathbf{w})=D(\mathbf{v},\mathbf{w},\mathbf{u})=D(\mathbf{w},\mathbf{u},\mathbf{v})$ (cyclic symmetry)
5. $\mathbf{u},\mathbf{v},\mathbf{w}$ are linearly dependent if and only if $D(\mathbf{u},\mathbf{v},\mathbf{w})=0$

**Hint.** For part 5: linear dependence is equivalent to one vector being expressible as a linear combination of the other two; substitute this and use properties 1–3.

:::{solution} exer-ch2-triple-product-props
:class: dropdown

**1.** $\mathbf{e}_1\times\mathbf{e}_2=\mathbf{e}_0$, so $D(\mathbf{e}_0,\mathbf{e}_1,\mathbf{e}_2)=\mathbf{e}_0\cdot\mathbf{e}_0=1$.

**2.** $D(\alpha\mathbf{u}_0+\beta\mathbf{u}_1,\mathbf{v},\mathbf{w})=(\alpha\mathbf{u}_0+\beta\mathbf{u}_1)\cdot(\mathbf{v}\times\mathbf{w})=\alpha(\mathbf{u}_0\cdot(\mathbf{v}\times\mathbf{w}))+\beta(\mathbf{u}_1\cdot(\mathbf{v}\times\mathbf{w}))$ (by the linearity of the inner product).

**3.** Expand directly: $D=u_0v_1w_2-u_0v_2w_1+u_1v_2w_0-u_1v_0w_2+u_2v_0w_1-u_2v_1w_0$. After exchanging the first two arguments, the six terms cancel one by one against the original expression, so $D(\mathbf u,\mathbf v,\mathbf w)=-D(\mathbf v,\mathbf u,\mathbf w)$; the sign change under exchanging the last two arguments comes directly from the anticommutativity of the cross product.

**4.** By repeated exchanges using property 3: $D(\mathbf{u},\mathbf{v},\mathbf{w})=-D(\mathbf{v},\mathbf{u},\mathbf{w})=D(\mathbf{v},\mathbf{w},\mathbf{u})=-D(\mathbf{w},\mathbf{v},\mathbf{u})=D(\mathbf{w},\mathbf{u},\mathbf{v})$.

**5.** ($\Rightarrow$) If the vectors are linearly dependent, take a set of relation coefficients that are not all zero, solve for a vector whose coefficient is nonzero, and use the fact that the scalar triple product changes only by a sign under permutations; after reordering we may assume $\mathbf{w}=\alpha\mathbf{u}+\beta\mathbf{v}$, and then $D(\mathbf{u},\mathbf{v},\mathbf{w})=\alpha D(\mathbf{u},\mathbf{v},\mathbf{u})+\beta D(\mathbf{u},\mathbf{v},\mathbf{v})=0$ ($D=0$ whenever a vector is repeated). ($\Leftarrow$) If the vectors are linearly independent, then geometrically the parallelepiped has nonzero volume, so $D\neq0$.
:::
::::

::::{exercise} The Perpendicular-Diagonals Condition for a Rhombus

In $\mathbb{R}^n$, consider the parallelogram spanned by $\mathbf{u}$ and $\mathbf{v}$ (with vertices $\mathbf{0}$, $\mathbf{u}$, $\mathbf{v}$, $\mathbf{u}+\mathbf{v}$). Prove that the two diagonals are perpendicular if and only if this parallelogram is a rhombus ($\|\mathbf{u}\|=\|\mathbf{v}\|$).

**Hint.** The two diagonal vectors are $\mathbf{u}+\mathbf{v}$ and $\mathbf{u}-\mathbf{v}$; compute their inner product.

:::{solution} exer-ch2-rhombus-diagonals

::::{exercise} The Area of a Parallelogram in $\mathbb{R}^4$
:label: exer-ch2-r4-parallelogram-area

There is no cross product in $\mathbb{R}^4$; derive the area formula for a parallelogram by following the guided steps below.

1. The vectors $\overrightarrow{OP}=(1,0,1,0)$, $\overrightarrow{OQ}=(0,1,0,1)$:
   - (a) Compute $\overline{OH}=\|\text{proj}_{\overrightarrow{OQ}}\overrightarrow{OP}\|$.
   - (b) Compute the height $\overline{PH}=\|\overrightarrow{OP}-\text{proj}_{\overrightarrow{OQ}}\overrightarrow{OP}\|$.
   - (c) Compute the area of the parallelogram.

2. For linearly independent vectors $\mathbf{u},\mathbf{v}\in\mathbb{R}^4$, prove in the same way that

$$
\text{area}=\sqrt{\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2}
$$

3. Starting from area $=\|\mathbf{u}\|\|\mathbf{v}\|\sin\theta$, derive the formula above again using the angle definition from the inner product.
4. If $\mathbf{v}=\lambda\mathbf{u}$, check that the area formula gives $0$.
5. If $\mathbf{u}\perp\mathbf{v}$, find the area and explain its geometric meaning.
6. If $\|\mathbf{u}\|=\|\mathbf{v}\|$ (a rhombus), check that the area of the rhombus is $=\dfrac{pq}{2}$ ($p,q$ are the lengths of the diagonals).

**Hint.** In part 2, the base is $=\|\mathbf{v}\|$ and the height is $=\|\mathbf{u}-\text{proj}_\mathbf{v}\mathbf{u}\|$; then expand.

:::{solution} exer-ch2-r4-parallelogram-area
:class: dropdown

**Part 1.** $\overrightarrow{OP}\cdot\overrightarrow{OQ}=0$, so $\text{proj}_{\overrightarrow{OQ}}\overrightarrow{OP}=\mathbf{0}$. Then $\overline{OH}=0$, the height is $\overline{PH}=\|\overrightarrow{OP}\|=\sqrt{2}$, and the area is $=\|\overrightarrow{OQ}\|\cdot\overline{PH}=\sqrt{2}\cdot\sqrt{2}=2$.

**Part 2.**

$$
\text{area}=\|\mathbf{v}\|\cdot\left\|\mathbf{u}-\frac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{v}\|^2}\mathbf{v}\right\|
$$

$$
= \|\mathbf{v}\|\cdot\sqrt{\|\mathbf{u}\|^2-\frac{(\mathbf{u}\cdot\mathbf{v})^2}{\|\mathbf{v}\|^2}}
= \sqrt{\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2}
$$

**Part 3.** $\text{area}=\|\mathbf{u}\|\|\mathbf{v}\|\sin\theta=\|\mathbf{u}\|\|\mathbf{v}\|\sqrt{1-\cos^2\theta}=\sqrt{\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2}$ (using $\cos\theta=\dfrac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|\|\mathbf{v}\|}$).

**Part 4.** $\mathbf{u}\cdot\mathbf{v}=\lambda\|\mathbf{u}\|^2$ and $\|\mathbf{v}\|^2=\lambda^2\|\mathbf{u}\|^2$. Substituting: $\|\mathbf{u}\|^2\cdot\lambda^2\|\mathbf{u}\|^2-\lambda^2\|\mathbf{u}\|^4=0$ ✓.

**Part 5.** $\mathbf{u}\cdot\mathbf{v}=0$, so the area is $=\|\mathbf{u}\|\|\mathbf{v}\|$, the product of the two side lengths—exactly the area formula for a rectangle; if the two sides have equal length, it is a square.

**Part 6.** Let $\|\mathbf{u}\|=\|\mathbf{v}\|=a$, with diagonals $p=\|\mathbf{u}+\mathbf{v}\|$, $q=\|\mathbf{u}-\mathbf{v}\|$:

$$
p^2=2a^2+2\mathbf{u}\cdot\mathbf{v},\quad q^2=2a^2-2\mathbf{u}\cdot\mathbf{v}
$$

$$
\frac{pq}{2}=\frac{1}{2}\sqrt{p^2q^2}=\frac{1}{2}\sqrt{(2a^2)^2-4(\mathbf{u}\cdot\mathbf{v})^2}=\sqrt{a^4-(\mathbf{u}\cdot\mathbf{v})^2}
$$

The area is $=\sqrt{a^4-(\mathbf{u}\cdot\mathbf{v})^2}=\dfrac{pq}{2}$ ✓.
:::
::::

```python
# ============================================================
print_header("Exercise 34 | The Area of a Parallelogram in ℝ⁴")
# ============================================================

print_step(1, "A concrete example: when OP⊥OQ, area = ‖OP‖‖OQ‖")
OP=np.array([1.,0.,1.,0.]); OQ=np.array([0.,1.,0.,1.])
compare_print("area", f"{para_area(OP,OQ):.6f}", "‖OP‖‖OQ‖ = 2 (OP⊥OQ)")

print_step(2, "General vectors: area formula vs ‖u‖‖v‖sinθ")
rng=np.random.default_rng(3)
u=rng.standard_normal(4); v=rng.standard_normal(4)
A1 = para_area(u,v)
cos_t = np.clip(np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v)),-1,1)
A2 = np.linalg.norm(u)*np.linalg.norm(v)*abs(np.sin(np.arccos(cos_t)))
compare_print("area formula", f"{A1:.6f}", f"‖u‖‖v‖sinθ = {A2:.6f}  ✓ {np.isclose(A1,A2)}")

print_step(3, "Area for parallel vectors = 0")
compare_print("area when v=3u", f"{para_area(u,3*u):.2e}", "≈ 0")

print_step(4, "Area of a rhombus = pq/2 (product of the diagonals / 2)")
theta=np.pi/4; a=3.0
u_=a*np.array([1.,0.,0.,0.]); v_=a*np.array([np.cos(theta),np.sin(theta),0.,0.])
p=np.linalg.norm(u_+v_); q=np.linalg.norm(u_-v_)
compare_print("rhombus area", f"{para_area(u_,v_):.4f}", f"pq/2 = {p*q/2:.4f}  ✓ {np.isclose(para_area(u_,v_),p*q/2)}")
```

### Generalized Area Formulas

::::{exercise} Projections of a Parallelogram onto the Coordinate Planes ($\mathbb{R}^3$)
:label: exer-ch2-area-projection-r3

Let $\mathbf{u}=(u_0,u_1,u_2)$, $\mathbf{v}=(v_0,v_1,v_2)$. The area $A$ of the parallelogram satisfies

$$
A^2=\|\mathbf{u}\times\mathbf{v}\|^2=\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2=
\begin{vmatrix}u_0&u_1\\v_0&v_1\end{vmatrix}^2+
\begin{vmatrix}u_0&u_2\\v_0&v_2\end{vmatrix}^2+
\begin{vmatrix}u_1&u_2\\v_1&v_2\end{vmatrix}^2
$$

1. Let $\mathbf{u}_{01}=(u_0,u_1,0)$, $\mathbf{v}_{01}=(v_0,v_1,0)$ be the projections of $\mathbf{u},\mathbf{v}$ onto the $x_0x_1$-plane; find the area $A_{01}$ of the projected parallelogram.
2. Similarly, find $A_{02}$ (the $x_0x_2$-plane) and $A_{12}$ (the $x_1x_2$-plane).
3. Verify that $A^2=A_{01}^2+A_{02}^2+A_{12}^2$.

**Hint.** After projection, we still have a parallelogram spanned by two vectors, and its area is the absolute value of the corresponding $2\times2$ determinant in the original formula.

:::{solution} exer-ch2-area-projection-r3
:class: dropdown

**Step 1.** $A_{01}=\left|\begin{vmatrix}u_0&u_1\\v_0&v_1\end{vmatrix}\right|=|u_0v_1-u_1v_0|$.

**Step 2.** $A_{02}=|u_0v_2-u_2v_0|$, $A_{12}=|u_1v_2-u_2v_1|$.

**Step 3.** By the cross product area formula:

$$
A^2=(u_1v_2-u_2v_1)^2+(u_0v_2-u_2v_0)^2+(u_0v_1-u_1v_0)^2=A_{12}^2+A_{02}^2+A_{01}^2
$$

This is the "generalized Pythagorean theorem": the square of the area of a parallelogram equals the sum of the squares of the areas of its projections onto the coordinate planes.
:::
::::

```python
# ============================================================
print_header("Exercise 38–39 | The Generalized Pythagorean Theorem (Cauchy–Binet)")
# ============================================================

rng=np.random.default_rng(99)

print_step(1, "Check in ℝ³: A² = Σ_{i<j} A_{ij}²")
u=rng.standard_normal(3); v=rng.standard_normal(3)
A2=para_area(u,v)**2
Aij2=sum((u[i]*v[j]-u[j]*v[i])**2 for i,j in combinations(range(3),2))
compare_print("A² (ℝ³)", f"{A2:.6f}", f"ΣA_ij² = {Aij2:.6f}  ✓ {np.isclose(A2,Aij2)}")

print_step(2, "Check in ℝ⁴: A² = Σ_{i<j} A_{ij}²")
u4=rng.standard_normal(4); v4=rng.standard_normal(4)
A2_4=para_area(u4,v4)**2
Aij2_4=sum((u4[i]*v4[j]-u4[j]*v4[i])**2 for i,j in combinations(range(4),2))
compare_print("A² (ℝ⁴)", f"{A2_4:.6f}", f"ΣA_ij² = {Aij2_4:.6f}  ✓ {np.isclose(A2_4,Aij2_4)}")
```

::::{exercise} Projections of a Parallelogram onto the Coordinate Planes ($\mathbb{R}^4$)
:label: exer-ch2-area-projection-r4

Let $\mathbf{u}=(u_0,u_1,u_2,u_3)$, $\mathbf{v}=(v_0,v_1,v_2,v_3)\in\mathbb{R}^4$, and let the area of the parallelogram be $A=\sqrt{\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2}$.

1. Check that the projection of $\mathbf{u}$ onto the $x_0x_1$-plane is $(u_0,u_1,0,0)$.
2. Define $A_{ij}$ to be the area of the parallelogram projected onto the $x_ix_j$-plane. Check that

$$
A_{01}^2=(u_0v_1-u_1v_0)^2,\quad A_{02}^2=(u_0v_2-u_2v_0)^2,\quad\ldots,\quad A_{23}^2=(u_2v_3-u_3v_2)^2
$$

3. Check the generalized Pythagorean theorem $A^2=A_{01}^2+A_{02}^2+A_{03}^2+A_{12}^2+A_{13}^2+A_{23}^2$ ($\binom{4}{2}=6$ terms in all).

**Hint.** Expand $\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2$ directly and compare with the sum of the squares of the six determinants.

:::{solution} exer-ch2-area-projection-r4
:class: dropdown

**Step 1.** The orthogonal projection operator onto the $x_0x_1$-plane keeps components 0 and 1 and sets the rest to $0$, so the projection is $(u_0,u_1,0,0)$ ✓.

**Step 2.** After projecting onto the $x_ix_j$-plane, the square of the area of the parallelogram spanned by the two projected vectors is $=\left|\begin{smallmatrix}u_i&u_j\\v_i&v_j\end{smallmatrix}\right|^2=(u_iv_j-u_jv_i)^2$; this holds for all six terms.

**Step 3.** Expand directly:

$$
\|\mathbf{u}\|^2\|\mathbf{v}\|^2-(\mathbf{u}\cdot\mathbf{v})^2
=\sum_{i<j}(u_iv_j-u_jv_i)^2
=\sum_{i<j}A_{ij}^2
$$

(This is the four-dimensional version of Lagrange's identity.) Hence the generalized Pythagorean theorem holds in $\mathbb{R}^4$, and likewise in $\mathbb{R}^n$.
