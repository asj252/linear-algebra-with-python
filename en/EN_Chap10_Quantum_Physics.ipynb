If classical physics is a "science of calculus," then quantum physics is, at its core, a "geometry of linear algebra."

In the models of classical mechanics, the state of a particle can be described by a point $(x,p)$ in phase space. Quantum theory adopts a different framework: in the finite-dimensional setting that this chapter mainly discusses, **a pure state is represented by a unit vector in a complex inner product space, but vectors that differ by a global phase represent the same physical state; observables are represented by Hermitian matrices.** The work of Heisenberg, Schrödinger, Dirac, and others developed this theory. For infinite-dimensional operators such as position and momentum, one must also deal with domains and self-adjointness, and not every finite-dimensional conclusion can be carried over directly.

The algebraic theorems of the previous two chapters provide a precise mathematical language for quantum theory. The physical correspondences below rest on the postulates of quantum theory for states, measurement, and evolution; they do not follow from theorems of linear algebra alone:

* **The eigenvalues of a Hermitian matrix are all real**: the possible outcomes of an ideal projective measurement are represented by these real eigenvalues.
* **Eigenvectors belonging to different eigenvalues are orthogonal**: the corresponding orthogonal pure states can be perfectly distinguished by a suitable ideal measurement; arbitrary distinct pure states need not be.
* **The projection form of the spectral theorem, $\hat A=\sum_j\lambda_j\mathbf{P}_j$**: the Born rule gives the probabilities $p_j=\|\mathbf{P}_j\psi\|^2$; in an ideal projective measurement, when $p_j>0$, the conditional state is $\mathbf{P}_j\psi/\sqrt{p_j}$. This is the measurement model we adopt, and a measurement need not always change the original state.
* **Unitary operators (unitary matrices), which preserve inner products and lengths**: these correspond to the smooth time evolution of a closed quantum system (the Schrödinger equation) and guarantee that the total probability of the system is always strictly conserved at 1.

When two Hermitian operators do not commute, they have no common complete orthonormal eigenbasis. This does not rule out individual common eigenstates, however, and one cannot conclude from it that no state can have definite values of both. The lower bound in the Robertson relation $\Delta A\Delta B\geq\tfrac12|\langle\psi|[\mathbf A,\mathbf B]|\psi\rangle|$ depends on the chosen state and may be zero; it describes the statistical spread of the state, not instrument error.

For readers interested only in pure mathematics and algorithms, this chapter can be treated as optional; skipping it does not affect the path to the singular value decomposition (SVD) in Chapter 11. But if you are willing to pause here for a while, you will see with your own eyes how eigenvalues, orthogonal projections, and unitary transformations—which looked abstract, perhaps even a little dry, on the blackboard—shed their purely symbolic garb and become a magnificent poem describing the deepest structure of nature.

## 10.1 A Review of Basic Probability

The finite-dimensional part of this chapter assumes nondegenerate observables, or rank-one ideal projective measurements that can distinguish every basis-vector label; apart from the infinite-dimensional exception of position and momentum, which is discussed separately, everything is treated within this scope.

Before delving into quantum mechanics, we need to review some basic notions of probability. The outcomes of quantum measurements are **discrete** (quantized), so we focus mainly on discrete probability.

#### Discrete Random Variables and Probability Distributions

A **discrete random variable** can take finitely many or countably infinitely many values; in this section we first restrict to finitely many values, which ensures that the expectation and variance exist. For example:
- The number shown by a die: $\{1,2,3,4,5,6\}$
- The outcome of a quantum measurement: the set of eigenvalues of an observable $\{a_0, a_1, \ldots, a_{n-1}\}$

The **probability mass function** (PMF) $P(\mathsf{x}=x)$ describes the probability of each possible value of a discrete random variable and satisfies:

**Normalization condition.**
$$\sum_{\text{all } x} P(\mathsf{x}=x) = 1$$

This ensures that the probabilities of all possible outcomes sum to 1.

:::{prf:example} Probability Distribution of a Fair Die
:label: ex-fair-die-pmf

The PMF of a fair six-sided die is:
$$P(\mathsf{x}=k) = \frac{1}{6}, \quad k \in \{1,2,3,4,5,6\}$$

Check the normalization: $\sum_{k=1}^{6} \frac{1}{6} = 1$ ✓
:::

#### Expectation Value and Variance

The **expectation value**, also called the mean, is the "average value" of a random variable, written $\mathbb{E}[\mathsf{x}]$ or $\mu$:

$$\mathbb{E}[\mathsf{x}] := \sum_{\text{all } x} x \cdot P(\mathsf{x}=x)$$

The expectation value gives the "center" of the values taken by the random variable.

:::{prf:example} Expectation Value of a Die Roll
:label: ex-die-expectation

The expectation value of the number shown by a fair die:
$$\mathbb{E}[\mathsf{x}] = \sum_{k=1}^{6} k \cdot \frac{1}{6} = \frac{1}{6}(1+2+3+4+5+6) = \frac{21}{6} = 3.5$$

Although a die can never show 3.5, this value represents the long-run average.
:::

The **variance** measures how spread out a random variable is, written $\text{Var}(\mathsf{x})$ or $\sigma^2$:

$$\text{Var}(\mathsf{x}) := \mathbb{E}[(\mathsf{x}-\mu)^2] = \mathbb{E}[\mathsf{x}^2] - (\mathbb{E}[\mathsf{x}])^2$$

where $\mu = \mathbb{E}[\mathsf{x}]$ is the expectation value. The larger the variance, the more spread out the values of the random variable.

The **standard deviation** $\sigma = \sqrt{\text{Var}(\mathsf{x})}$ measures the spread in the same units as the original variable.

:::{prf:example} Variance of a Die Roll
:label: ex-die-variance

Compute $\mathbb{E}[\mathsf{x}^2]$:
$$\mathbb{E}[\mathsf{x}^2] = \sum_{k=1}^{6} k^2 \cdot \frac{1}{6} = \frac{1}{6}(1+4+9+16+25+36) = \frac{91}{6}$$

Variance:
$$\text{Var}(\mathsf{x}) = \frac{91}{6} - \left(\frac{7}{2}\right)^2 = \frac{91}{6} - \frac{49}{4} = \frac{35}{12} \approx 2.92$$

Standard deviation: $\sigma = \sqrt{35/12} \approx 1.71$
:::

#### Independence

When we analyze several random variables, we need to consider how they relate to one another.

The **joint probability distribution** $P(\mathsf{x}=x, \mathsf{y}=y)$ describes the probability that two discrete variables simultaneously take particular values.

Two discrete random variables $\mathsf{x}$ and $\mathsf{y}$ are **independent** if and only if:
$$P(\mathsf{x}=x, \mathsf{y}=y) = P(\mathsf{x}=x) \cdot P(\mathsf{y}=y) \quad \text{for all } x,y$$

This means that knowing the value of $\mathsf{x}$ does not change the probability distribution of $\mathsf{y}$, and vice versa.

:::{prf:example} Independence of Two Die Rolls
:label: ex-two-dice-independence

The first roll of a die gives $\mathsf{x}$ and the second gives $\mathsf{y}$. The two rolls are independent:
$$P(\mathsf{x}=3, \mathsf{y}=5) = P(\mathsf{x}=3) \cdot P(\mathsf{y}=5) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36}$$

But if $\mathsf{y} = 7 - \mathsf{x}$ (the second die is set so that the two numbers add up to 7), they are not independent:
$$P(\mathsf{x}=3, \mathsf{y}=4) = P(\mathsf{x}=3) = \frac{1}{6} \neq \frac{1}{36} = P(\mathsf{x}=3) \cdot P(\mathsf{y}=4)$$
:::

---

#### Connections to Quantum Mechanics

These discrete probability notions appear concretely in quantum mechanics (§10.2-10.5) as follows:

**1. Discreteness of measurement outcomes** (§10.4):
- When an observable $\hat{A}$ is measured, the outcome can only be one of its eigenvalues $\{a_0, a_1, \ldots, a_{n-1}\}$
- One never obtains an arbitrary value between eigenvalues
- This is precisely where the word "quantum" (quantum = discretized) comes from

**2. The Born rule**:
- When the pure state $|\psi\rangle = \sum_j c_j |a_j\rangle$ is measured with $\hat{A}$
- the **probability** of obtaining the eigenvalue $a_j$ is $P(a_j) = |c_j|^2 = |\langle a_j|\psi\rangle|^2$
- This is the fundamental source of probability in quantum mechanics

**3. Expectation value of an observable**:
- $\mathbb{E}_\psi[\hat{A}] = \sum_j a_j P(a_j) = \langle\psi|\hat{A}|\psi\rangle$
- This is exactly the discrete expectation value defined in this section

**4. The uncertainty principle** (§10.5):
- **Uncertainty** is defined as the standard deviation: $\Delta A = \sqrt{\mathbb{E}[A^2] - (\mathbb{E}[A])^2}$
- The uncertainty relation: $\Delta A \cdot \Delta B \geq \frac{1}{2}|\mathbb{E}[[\hat{A},\hat{B}]]|$
- This uses the variance and expectation value of this section directly

```{note}
:label: note-discrete-vs-continuous
**Why discuss only discrete probability?**

Quantum mechanics also has operators with continuous spectra (such as position and momentum), whose "eigenvalues" form a continuum. However:
1. In this chapter we deal mainly with **finite-dimensional quantum systems** (such as qubits and spin-1/2)
2. The observables of these systems all have **discrete spectra**
3. Operators with continuous spectra require infinite-dimensional Hilbert spaces and more advanced mathematics (such as $L^2(\mathbb{R})$ and generalized eigenfunctions)
4. Discrete probability is enough to understand the core ideas of this chapter

When we discuss the position-momentum uncertainty relation in §10.5, we will briefly mention the continuous case, but no deep theory of continuous probability is needed.
```

:::{admonition} §10.1 Summary
:class: tip
Starting from the most familiar ground, this section reviewed the probability mass function of a discrete random variable, the expectation value $\mathbb{E}[\mathsf{x}]$, and the variance $\text{Var}(\mathsf{x})$, preparing a precise mathematical language for the theory of quantum measurement that follows. These tools have direct counterparts in quantum mechanics: the normalization condition $\sum P(x)=1$ corresponds to the conservation of probability for the wave function, the expectation-value formula extends to the quantum average of an observable $\langle\psi|\hat{A}|\psi\rangle$, and the standard deviation becomes, in §10.5, the central measure $\Delta A$ of the uncertainty principle.

This section focuses only on the discrete case, because the observables of the finite-dimensional quantum systems treated in this chapter (such as qubits and spin-1/2 particles) all have discrete spectra; the infinite-dimensional tools needed for continuous cases such as position and momentum are briefly explained at the end of §10.5. Elementary as it looks, the language of probability is the complete foundation for understanding the theory of quantum measurement. With this language in hand, we can turn directly to the mathematical structure of quantum states themselves.

## 10.2 Fundamentals of Quantum States

The mathematical foundation of quantum mechanics rests on the theory of vector spaces, and quantum states are precisely the elements of such a vector space. This section introduces the basic notions of quantum states and lays the groundwork for the later discussion of density matrices and quantum measurement.

#### Pure States

In quantum mechanics, a **pure state** is the most basic way of representing the state of a quantum system. A pure state is represented by a state vector or a wave function, which completely characterizes the quantum state of the system.

:::{prf:definition} Pure State
:label: def-pure-state

**Pure state**: the quantum system is in a definite quantum state that can be completely described by a single state vector $|\psi\rangle$. This state vector belongs to a Hilbert space (which can be understood as a complete vector space equipped with an inner product) and must satisfy the **normalization condition**:

$$\langle\psi|\psi\rangle = 1$$

This condition guarantees that the probabilities of all possible measurement outcomes sum to 1, reflecting the completeness of probability.
:::

```{note}
:label: note-normalization-physical
**Physical meaning of the normalization condition.**

If a pure state is written as $|\psi\rangle = \sum_i c_i |i\rangle$, then the normalization condition is equivalent to:
$$\sum_i |c_i|^2 = \sum_i |\langle i|\psi\rangle|^2 = \langle\psi|\psi\rangle = 1$$

This ensures that, when we measure in any basis $\{|i\rangle\}$, the probabilities $P(i) = |c_i|^2$ of the individual outcomes add up to 1.
```

In **Dirac notation**, a pure state is written $|\psi\rangle$ (read "ket psi"). Its conjugate transpose is $\langle\psi|$ (read "bra psi"). This notation corresponds exactly to the column vectors, and the conjugate transposes of column vectors, that we used in earlier chapters.

A pure state can be written as a linear combination of basis vectors:

$$|\psi\rangle = \sum_{i=0}^{n-1} c_i |i\rangle$$

where $c_i \in \mathbb{C}$ are complex amplitudes and $\{|i\rangle\}$ is an orthonormal basis (note that $|i\rangle$ here is exactly the same as the familiar standard basis vector $\mathbf{e}_i$; only the notation differs).

:::{prf:example} A Detailed Description of the Qubit
:label: ex-qubit-detailed

The **qubit** is the simplest quantum system; its state space is the two-dimensional complex vector space $\mathbb{C}^2$.

**1. The computational basis**:

The most commonly used basis is the computational basis:
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

These correspond, for example, to spin up/down, or to the horizontal/vertical polarization states of a photon.

**2. A general pure state**:

A general pure state of a qubit can be written as:
$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$$

where $\alpha, \beta \in \mathbb{C}$ satisfy the normalization condition $|\alpha|^2 + |\beta|^2 = 1$.

**3. Concrete examples**:

- **Equal superposition**: $|\psi_+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$

- **Superposition with a phase**: $|\psi_-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$

- **With a complex phase**: $|\psi_i\rangle = \frac{1}{\sqrt{2}}(|0\rangle + i|1\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ i \end{pmatrix}$

**4. Geometric representation—the Bloch sphere**:

Every pure state of a qubit can be represented by a point $(\theta, \phi)$ on the Bloch sphere:
$$|\psi\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle$$

where $\theta \in [0, \pi]$ is the polar angle and $\phi \in [0, 2\pi)$ is the azimuthal angle. This representation connects the abstract space of quantum states with three-dimensional geometric intuition. (See Experiment 10.)
:::

#### Superposition States and Eigenstates

In quantum mechanics we often need to distinguish **superposition states** from **eigenstates**, but this distinction is made **relative to a measurement basis**.

:::{prf:definition} Superposition State (a Relative Notion)
:label: def-superposition-state

Given an orthogonal basis $\{|i\rangle\}_{i=0}^{n-1}$ (for example, the eigenstates of some observable), if the expansion of a pure state $|\psi\rangle$ in this basis has more than one nonzero coefficient $c_i$:

$$|\psi\rangle = \sum_{i=0}^{n-1} c_i |i\rangle, \quad \text{where at least two } c_i \neq 0$$

then $|\psi\rangle$ is called **a superposition state in that basis**.

If only one $c_k \neq 0$, that is, $|\psi\rangle = |k\rangle$, then it is called **an eigenstate of that basis**.
:::

```{important}
:label: imp-superposition-basis-dependent
**Superposition is a basis-dependent notion.**

The same pure state can look completely different in different bases:
- In basis $\mathcal{B}_0$ it may be an eigenstate
- In basis $\mathcal{B}_1$ it may be a superposition state

**Key distinction**:
- ◇**Pure state vs. mixed state**: an absolute distinction, independent of the choice of basis (whether the state can be described by a single state vector)
- **Superposition state vs. eigenstate**: a relative notion that depends on the chosen measurement basis
```

:::{prf:example} Basis Dependence of Superposition
:label: ex-superposition-basis

Consider the "up" state of a spin-1/2 particle along the $z$ direction:
$$|↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}_z$$

**In the $z$ basis**:
- This is an eigenstate of $\hat{S}_z$
- It is not a superposition state, because it is the basis vector $|0\rangle_z$ itself

**In the $x$ basis**:

First, the basis vectors of the $x$ basis are:
$$|↑_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}_z, \quad |↓_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}_z$$

Expanding $|↑_z\rangle$ in the $x$ basis:
$$|↑_z\rangle = \frac{1}{\sqrt{2}}|↑_x\rangle + \frac{1}{\sqrt{2}}|↓_x\rangle$$

**In the $x$ basis, this is a superposition state!**

This example clearly shows that superposition is a notion **relative to the measurement basis**. The same pure state can be at once:
- an eigenstate of $\hat{S}_z$ (a definite spin along $z$)
- a superposition state with respect to a measurement of $\hat{S}_x$ (an indefinite spin along $x$)
:::

#### Dimension of a Quantum System

The **dimension** $d$ of a quantum system is the dimension of its state space (Hilbert space); it reflects the complexity and the information capacity of the system.

:::{prf:property} Physical Meaning of the Dimension
:label: pro-dimension

For a $d$-dimensional quantum system:
1. **State description**: a pure state requires $2(d-1)$ real parameters ($d$ complex amplitudes $c_i$, minus the normalization condition and the global phase)
2. **Measurement outcomes**: in any orthogonal basis there are at most $d$ distinguishable measurement outcomes
3. **Information capacity**: a $d$-dimensional system can encode the information of $\log_2 d$ qubits
:::

Physical systems of common dimensions:

| Dimension $d$ | Physical system | Typical examples | Remarks |
|---------|----------|----------|------|
| 2 | Qubit | Electron spin, photon polarization | Basic unit of quantum information |
| 3 | Qutrit | Three-level atom, orbital angular momentum of photons | Particles with nuclear spin $I=1$ |
| 4 | Two qubits | Two entangled qubits | $2 \times 2 = 4$ |
| 8 | Three qubits | Basic unit of quantum error-correcting codes | $2^3 = 8$ |
| $2^n$ | $n$ qubits | Quantum computer | Dimension grows exponentially |
| $\infty$ | Infinite-dimensional system | Harmonic oscillator, position states of a free particle | Requires tools of functional analysis |

```{note}
:label: note-tensor-product
**Dimension of a composite system.**

If system A has dimension $d_A$ and system B has dimension $d_B$, then the composite system A⊗B has dimension:
$$d_{AB} = d_A \times d_B$$

For example, the composite system of $n$ qubits has dimension $2^n$, which is the mathematical root of the exponential growth of quantum computing power. This tensor product structure is examined in detail in Chapter 11 (§11.5) in the discussion of the Schmidt decomposition.
```

#### The Probabilistic Nature of Quantum States

One of the central features of quantum mechanics is its **intrinsically probabilistic nature**. Even though a pure state completely describes the system, measurement outcomes are still probabilistic.

:::{prf:property} Measurement Probabilities (the Born Rule)
:label: pro-born-rule

Let the pure state be $|\psi\rangle = \sum_{i=0}^{n-1} c_i |i\rangle$, where $\{|i\rangle\}$ are the eigenstates of some observable.

When we measure in this basis, the probability of obtaining the eigenstate $|i\rangle$ is:
$$P(i) = |c_i|^2 = |\langle i|\psi\rangle|^2$$

This rule is called the **Born rule**, and it is one of the fundamental postulates of quantum mechanics.
:::

```{important}
:label: imp-quantum-randomness
**The nature of quantum randomness.**

The probabilistic character of quantum mechanics is fundamentally different from that of classical statistical mechanics:

**Classical statistical mechanics**:
- Probability comes from our **incomplete knowledge** of the microscopic state of the system
- In principle, if we knew the positions and momenta of all particles, the evolution of the system would be deterministic
- Probability is **epistemic**: it reflects our ignorance

**Quantum mechanics**:
- Probability is **ontological**: even when the quantum state is completely described (a pure state), measurement outcomes are still random
- This randomness is not due to insufficient knowledge on our part; it is a fundamental feature of nature
- Einstein's famous objection that "God does not play dice" was aimed precisely at this feature

For example, the pure state $|\psi\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$ describes the system **completely**, with no hidden information. Yet when we measure, all we can say is: with probability 50% we obtain $|0\rangle$, and with probability 50% we obtain $|1\rangle$. This is not because the description is incomplete; it is a fundamental feature of the quantum world.
```

:::{prf:example} The Probabilistic Nature of Measuring a Pure State
:label: ex-measurement-probability

Consider a qubit in a superposition state:
$$|\psi\rangle = \frac{\sqrt{3}}{2}|0\rangle + \frac{1}{2}|1\rangle$$

**Check the normalization**:
$$|\alpha|^2 + |\beta|^2 = \left(\frac{\sqrt{3}}{2}\right)^2 + \left(\frac{1}{2}\right)^2 = \frac{3}{4} + \frac{1}{4} = 1 \quad \checkmark$$

**Measure in the computational basis $\{|0\rangle, |1\rangle\}$**:
- Probability of obtaining $|0\rangle$: $P(0) = \left|\frac{\sqrt{3}}{2}\right|^2 = \frac{3}{4} = 75\%$
- Probability of obtaining $|1\rangle$: $P(1) = \left|\frac{1}{2}\right|^2 = \frac{1}{4} = 25\%$

**State after the measurement**:
- If 0 is obtained, the system collapses to $|0\rangle$
- If 1 is obtained, the system collapses to $|1\rangle$

**Key observation**: the information in the superposition before the measurement (including the relative phase of the amplitudes) is **irreversibly lost** after the measurement. This is a fundamental feature of quantum measurement.
:::

:::{prf:remark} Global Phase Invariance
:label: rem-global-phase

If a pure state is multiplied as a whole by a phase factor $e^{i\gamma}$ ($\gamma \in \mathbb{R}$), giving $|\psi'\rangle = e^{i\gamma}|\psi\rangle$, then no measurement probability changes:

$$P'(i) = |\langle i|\psi'\rangle|^2 = |e^{i\gamma}|^2\,|\langle i|\psi\rangle|^2 = |\langle i|\psi\rangle|^2 = P(i)$$

The density matrix (§10.3) is also exactly the same: $|\psi'\rangle\langle\psi'| = e^{i\gamma}e^{-i\gamma}|\psi\rangle\langle\psi| = |\psi\rangle\langle\psi|$, that is, $\boldsymbol{\psi}'(\boldsymbol{\psi}')^* = \boldsymbol{\psi}\boldsymbol{\psi}^*$. Hence $|\psi\rangle$ and $e^{i\gamma}|\psi\rangle$ describe **the same physical state**—what is physically observable is only the **relative phase** between the coefficients. For a complete step-by-step verification in all four representations (probabilities, expectation values, the density matrix, and the Bloch sphere), see {prf:ref}`exer-ch10-global-phase` in the Chapter 10 exercises.
:::

#### The Relationship Between Measurement and Bases

The discussion above hints at a deep fact: **measurement is closely tied to the choice of basis**. Different measurements correspond to different orthogonal bases.

```{admonition} Choosing a Measurement Basis
:class: tip

In quantum mechanics:
- **Every observable corresponds to an orthogonal basis** (its eigenstates)
- **Every measurement chooses a basis** in which to "observe" the system:
  - If the system is an eigenstate of that basis → the measurement outcome is certain
  - If the system is a superposition state → the measurement outcome is random, with probabilities given by $|c_i|^2$ (the Born rule)
- When the same pure state is measured in different bases, the probability distributions of the outcomes may differ

For example, for the spin state $|↑_z\rangle$:
- Measuring with $\hat{S}_z$ (the $z$ basis): "up" with 100% probability
- Measuring with $\hat{S}_x$ (the $x$ basis): "up" with 50% probability and "down" with 50% probability

This is not a problem of measurement technique; it is a fundamental feature of quantum mechanics. Two noncommuting observables have no common complete eigenbasis, although they may still have individual common eigenstates; the lower bound on the uncertainty depends on the prepared state. This is precisely the root of the uncertainty principle (see §10.5).
```

The full theory of measurement, including projective measurement, the mechanism of state collapse, and noncommutativity, is developed systematically in §10.4.

:class: tip
The central question of this section is: what exactly is the "state" of a quantum system, and how can we describe it precisely in mathematical language? The answer is the pure state—a normalized vector $|\psi\rangle$ in a Hilbert space—which completely characterizes all measurable physical information about the system. Through the correspondence between Dirac notation and column vectors, the mystique of "superposition" dissolves: a superposition is nothing more than the natural result of expanding a vector in different bases. Note in particular that superposition is a notion relative to the measurement basis—the same pure state is an eigenstate of one operator but a superposition state with respect to another—and this observation already foreshadows the deep root of the uncertainty principle in §10.5.

The Born rule $P(i) = |\langle i|\psi\rangle|^2$ is the bridge between state vectors and measurement probabilities, and it is also the source of the intrinsic probabilistic character of quantum mechanics—a character that is ontological rather than a consequence of missing information. The Bloch sphere representation of the qubit provides concrete geometric intuition and gives the abstract complex vector space a picture we can visualize. The description by pure states has a fundamental limitation, however: when our knowledge of how a system was prepared is only statistical, a single state vector is no longer sufficient. This is exactly the motivation for introducing the density matrix in §10.3.

## 10.3 Mixed States and the Density Matrix

In §10.2 we discussed pure states—quantum states that can be completely described by a single state vector. In many practical situations, however, the systems we face cannot be described by a pure state, and we must introduce the notion of a **mixed state**.

#### Why Do We Need Mixed States?

In real physical systems we often encounter the following situations, in which a description by a pure state no longer applies:

```{admonition} Typical Situations That Call for Mixed States
:class: note

**1. Incomplete preparation of the system**
- The system produced by an experimental apparatus may be a statistical mixture of several possible states
- For example, the photons emitted by a light source are horizontally polarized with probability 60% and vertically polarized with probability 40%
- This is not a single pure state but a statistical mixture of two pure states

**2. Systems in thermal equilibrium**
- A system at temperature $T > 0$ is in a thermal statistical mixture of energy eigenstates
- According to the Boltzmann distribution, the system is in the energy eigenstate $|E_n\rangle$ with probability $p_n = \frac{e^{-E_n/k_BT}}{Z}$
- Even if the Hamiltonian is known (all of $\{ E_n \}$), a system at finite temperature cannot be described by a single pure state/eigenstate

**3. Subsystems of entangled systems** (a preview of Chapter 11)
- Consider two entangled qubits: $|\Psi\rangle_{AB} = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$
- If we observe only subsystem A, the state of A cannot be described by a pure state
- The reduced density matrix of A is a mixed state: $\rho_A = \frac{1}{2}|0\rangle\langle 0| + \frac{1}{2}|1\rangle\langle 1|$
- This is the mathematical signature of quantum entanglement, discussed in detail with the Schmidt decomposition in Chapter 11 (§11.5)

**4. Incomplete measurement information**
- We know only that the system is "possibly" in one of certain pure states, but not which one
- For example, we know that a particle's spin is definitely up or down, but not which
```

In these situations a single state vector $|\psi\rangle$ cannot completely describe the state of the system, and we need to introduce the **density matrix** to handle such a **statistical mixture**.

```{important}
:label: imp-classical-vs-quantum-uncertainty
**Two kinds of uncertainty.**

Quantum systems exhibit two fundamentally different kinds of uncertainty:

1. **Quantum uncertainty** (the intrinsic randomness of a pure state)
   - Source: the probabilistic nature of superposition states (the Born rule)
   - Example: the randomness in measuring the pure state $\frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$
   - Characteristic: even when the quantum state is known completely, the measurement outcome is still random

2. **Classical uncertainty** (statistical mixture)
   - Source: incomplete knowledge of how the system was prepared or of its history
   - Example: the system is in $|0\rangle$ with probability 50% and in $|1\rangle$ with probability 50%
   - Characteristic: if we knew which state the system is actually in, there would be no uncertainty

A mixed state contains both kinds of uncertainty.
```

#### Definition of a Mixed State

Since pure states cannot describe statistical mixtures, we need a new mathematical tool:

:::{prf:definition} Mixed State
:label: def-mixed-state

**Density-matrix representation of an ensemble**: the quantum system is in a statistical mixture in which it is in the pure state $|\psi_i\rangle$ with **classical probability** $p_i \in [0,1]$, where:
$$\sum_i p_i = 1$$

Here each $|\psi_i\rangle$ is normalized, and $\rho=\sum_i p_i|\psi_i\rangle\langle\psi_i|$. The state is called a mixed state only when $\operatorname{rank}\rho>1$; equivalently, the pure states with positive weight do not all lie on the same ray. A mixed state cannot be represented by a single state vector and must be described by a **density matrix** (also called a density operator).
:::

```{note}
:label: note-ensemble-interpretation
**Ensemble interpretation.**

One way to understand a mixed state is through the notion of an **ensemble**:
- Imagine many identically prepared quantum systems
- A fraction $p_0$ of them is in the pure state $|\psi_0\rangle$
- A fraction $p_1$ is in the pure state $|\psi_1\rangle$
- and so on

For a single system we do not know which pure state it is in; for the ensemble as a whole we know the statistical distribution.
```

#### The Density Matrix

The density matrix is the general mathematical tool for describing quantum systems (both pure states and mixed states).

:::{prf:definition} Density Matrix
:label: def-density-matrix

For a mixed state in which the system is in the pure state $|\psi_i\rangle$ with probability $p_i \in [0,1]$, the **density matrix** is defined as:

$$\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i|$$

**Special case—pure states**:

For a pure state $|\psi\rangle$, the density matrix reduces to:
$$\rho = |\psi\rangle\langle\psi|$$

In this case the rank of the density matrix is 1.
:::

```{note}
:label: note-density-matrix-representation
**The decomposition of a density matrix is not unique.**

The same density matrix $\rho$ can be decomposed in many different ways:
$$\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i| = \sum_j q_j |\phi_j\rangle\langle\phi_j|$$

where $\{|\psi_i\rangle\}$ and $\{|\phi_j\rangle\}$ may be completely different sets of states. This reflects the statistical interpretation of mixed states: different preparation procedures can produce the same statistical properties.

However, the **spectral decomposition** provides a standard form given by the eigenstates of the density matrix; the spectral decomposition into orthogonal projections, grouped by distinct eigenvalues, is unique, while the expansion into rank-one terms within a degenerate subspace is usually not unique.
```

#### Properties of the Density Matrix

As a Hermitian operator, the density matrix has all the important properties of Hermitian matrices discussed in Chapter 9 and earlier in this chapter:

:::{prf:property} Basic Properties of the Density Matrix
:label: pro-density-matrix

Let $\rho$ be a density matrix. Then:

1. **Hermiticity (self-adjointness)**: $\rho = \rho^*$
   - This guarantees that the eigenvalues are real

2. **Positive semidefiniteness**: $\langle\phi|\rho|\phi\rangle \geq 0$ for all $|\phi\rangle$
   - Equivalent to all eigenvalues satisfying $\lambda_i \geq 0$
   - This guarantees that probabilities are nonnegative

3. **Trace one**: $\text{Tr}(\rho) = 1$
   - This reflects the normalization condition for probabilities: $\sum_i p_i = 1$

4. **Test for purity**:
   - $\text{Tr}(\rho^2) = 1 \Leftrightarrow \rho$ is a pure state
   - $\text{Tr}(\rho^2) < 1 \Leftrightarrow \rho$ is a mixed state
   - The quantity $\text{Tr}(\rho^2)$ is called the **purity**

5. **Spectral decomposition (following the spectral theorem of §9.3)**: by properties 1–2, $\rho$ is a positive semidefinite Hermitian matrix,
   so it can be unitarily diagonalized (the diagonalizing matrix is usually not unique):
   $$\rho = \sum_j \lambda_j |e_j\rangle\langle e_j|$$
   where $\{|e_j\rangle\}$ are orthonormal eigenstates, $\lambda_j \geq 0$, and $\sum_j \lambda_j = 1$.
   - The eigenvalue $\lambda_j$ is the probability weight of the system being in the eigenstate $|e_j\rangle$
   - A pure state corresponds to $\rho$ having rank 1: the only nonzero eigenvalue is $\lambda_0 = 1$
   - The maximally mixed state corresponds to equal eigenvalues: $\lambda_j = 1/d$ ($d$ is the dimension of the Hilbert space)
:::

:::{prf:proof} Proof of the Test for Purity (Property 4)
:label: proof-purity

**Necessity** ($\rho$ is a pure state $\Rightarrow$ $\text{Tr}(\rho^2) = 1$):

If $\rho = |\psi\rangle\langle\psi|$ is a pure state, then:
$$\rho^2 = |\psi\rangle\langle\psi|\psi\rangle\langle\psi| = |\psi\rangle\langle\psi| = \rho$$

Therefore:
$$\text{Tr}(\rho^2) = \text{Tr}(\rho) = 1$$

**Sufficiency** ($\text{Tr}(\rho^2) = 1$ $\Rightarrow$ $\rho$ is a pure state):

By the spectral decomposition ({prf:ref}`pro-density-matrix`, property 5), $\rho = \sum_j \lambda_j |e_j\rangle\langle e_j|$, where $\lambda_j \geq 0$ and $\sum_j \lambda_j = 1$.

Compute:
$$\rho^2 = \sum_{j,k} \lambda_j \lambda_k |e_j\rangle\langle e_j|e_k\rangle\langle e_k| = \sum_j \lambda_j^2 |e_j\rangle\langle e_j|$$

Therefore:
$$\text{Tr}(\rho^2) = \sum_j \lambda_j^2$$

From $0 \leq \lambda_j \leq 1$ (positive semidefinite with trace one) we get $\lambda_j^2 \leq \lambda_j$, and adding term by term:
$$\sum_j \lambda_j^2 \leq \sum_j \lambda_j = 1$$

Equality holds if and only if exactly one $\lambda_k = 1$ and all other $\lambda_j = 0$. In that case:
$$\rho = |e_k\rangle\langle e_k|$$

which is exactly the form of a rank-one pure state.
:::

#### Examples of Mixed States

Let us understand mixed states and density matrices through concrete examples.

:::{prf:example} A Mixed State of a Spin-1/2 Particle
:label: ex-mixed-spin

**Setup**: consider a system of spin-1/2 particles whose density matrix is:
$$\rho = 0.7|{↑_z}\rangle\langle{↑_z}| + 0.3|{↓_z}\rangle\langle{↓_z}|$$

**Physical meaning**:
- This is **not** a single particle in a superposition state
- Instead, it describes an ensemble of particles:
  - 70% of the particles are in $|{↑_z}\rangle$ (spin up)
  - 30% of the particles are in $|{↓_z}\rangle$ (spin down)

**Matrix representation**:

In the $z$ basis $\{|0\rangle = |{↑_z}\rangle, |1\rangle = |{↓_z}\rangle\}$:
$$\rho = 0.7 \begin{pmatrix} 1 \\ 0 \end{pmatrix}\begin{pmatrix} 1 & 0 \end{pmatrix} + 0.3 \begin{pmatrix} 0 \\ 1 \end{pmatrix}\begin{pmatrix} 0 & 1 \end{pmatrix}$$

$$= 0.7 \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} + 0.3 \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 0.7 & 0 \\ 0 & 0.3 \end{pmatrix}$$

**Purity check**:
$$\text{Tr}(\rho^2) = \text{Tr}\begin{pmatrix} 0.49 & 0 \\ 0 & 0.09 \end{pmatrix} = 0.49 + 0.09 = 0.58 < 1$$

Hence this is a **mixed state**.

**Observations**:
- The density matrix is **diagonal** (in the $z$ basis)
- The diagonal entries are exactly the probabilities of the basis states ($\rho_{00} = 0.7$, $\rho_{11} = 0.3$)
- **Zero off-diagonal entries** indicate the absence of coherence
:::

:::{prf:example} A Closer Comparison of the Maximally Mixed State and a Superposition State
:label: ex-mixed-vs-superposition

This example shows the fundamental difference between a mixed state and a superposition state.

**State A: the maximally mixed state**

The system is in $|0\rangle$ with probability 50% and in $|1\rangle$ with probability 50%:
$$\rho_{\text{mixed}} = 0.5|0\rangle\langle 0| + 0.5|1\rangle\langle 1| = \begin{pmatrix} 0.5 & 0 \\ 0 & 0.5 \end{pmatrix} = \frac{\mathbf{I}}{2}$$

**Purity**: $\text{Tr}(\rho_{\text{mixed}}^2) = 0.5 < 1$ → mixed state

**State B: the equal superposition**

A single system is in a superposition state:
$$|\psi_+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$$

Its density matrix is:
$$\rho_{\text{pure}} = |\psi_+\rangle\langle\psi_+| = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$$

**Purity**: $\text{Tr}(\rho_{\text{pure}}^2) = 1$ → pure state

---

**Key differences**:

| Property | Maximally mixed state $\rho_{\text{mixed}}$ | Superposition state $\rho_{\text{pure}}$ |
|------|----------------------------------|---------------------------|
| Matrix form | $\begin{pmatrix} 0.5 & 0 \\ 0 & 0.5 \end{pmatrix}$ | $\begin{pmatrix} 0.5 & 0.5 \\ 0.5 & 0.5 \end{pmatrix}$ |
| Off-diagonal entries | **0** (no coherence) | **0.5** (coherence present) |
| ◇Measured in the $Z$ basis | 50%-50% | 50%-50% |
| ◇Measured in the $X$ basis | 50%-50% | **100%-0%** |
| Ability to interfere | **None** | **Present** |
| Purity $\text{Tr}(\rho^2)$ | 0.5 | 1.0 |

---

**Mathematical explanation**:

The maximally mixed state $\rho = \frac{\mathbf{I}}{2}$ is **rotation invariant**. For any unitary transformation $\mathbf{U}$ (representing a change of basis):
$$\mathbf{U}\rho \mathbf{U}^* = \mathbf{U}\frac{\mathbf{I}}{2}\mathbf{U}^* = \frac{\mathbf{UU}^*}{2} = \frac{\mathbf{I}}{2} = \rho$$

This means that **no matter which measurement basis is chosen**, the statistics are always 50%-50%.

The superposition state $|\psi_+\rangle$, by contrast, behaves differently in different bases:
- **In the $Z$ basis** $\{|0\rangle, |1\rangle\}$: $|\psi_+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$ → 50%-50%
- **In the $X$ basis** $\{|+\rangle, |-\rangle\}$: $|\psi_+\rangle = |+\rangle$ → 100%-0%

---

**Photon polarization experiments**:

This difference can be understood intuitively through optical experiments:

The **maximally mixed state** corresponds to **unpolarized light**:
- The light source emits a collection of photons with random polarization directions
- Measured with a polarizer at any angle: the transmittance is always 50%
- Rotating the polarizer does not change the statistics
- **No coherence term in the H/V polarization basis**; whether there is double-slit interference depends separately on the path density matrix

The **superposition state** corresponds to **linearly polarized light** (for example, polarized at 45°):
- $|\psi_+\rangle = \frac{1}{\sqrt{2}}(|H\rangle + |V\rangle)$ corresponds to 45° linear polarization
- With horizontal/vertical polarizers: 50%-50%
- With 45°/135° polarizers: 100% passes the 45° polarizer and 0% passes the 135° polarizer
- **A coherence term is present in the H/V polarization basis**; this is not the same as whether given paths are coherent with each other

---

**Physical essence**:

Although these two states give the same measurement statistics in the $Z$ basis (both 50%-50%), their **physical nature is completely different**:

1. **Coherence**:
   - A superposition state has quantum coherence and can produce interference effects
   - The maximally mixed state in this example has no coherence in the given basis; a general mixed state may still have nonzero off-diagonal entries

2. **Off-diagonal entries**:
   - The off-diagonal entries of $\rho_{\text{mixed}}$ are 0
   - The off-diagonal entries of $\rho_{\text{pure}}$ are not 0
   - The off-diagonal entries encode phase information and coherence

3. **Sensitivity to the basis**:
   - The measurement outcomes of a superposition state depend strongly on the choice of measurement basis
   - The maximally mixed state gives the same statistics in every basis
:::

````{important}
:label: imp-off-diagonal-coherence
**Off-diagonal entries, quantum coherence, and decoherence**

The off-diagonal entries $\rho_{ij}$ ($i \neq j$) of the density matrix are the mathematical carriers of quantum coherence. The clearest way into their physical meaning is the **double-slit experiment**—the most direct experimental manifestation of wave-particle duality.

```{note}
**Reading guide**: the derivation below borrows the language of wave optics (diffraction amplitudes, path differences, the far-field approximation). These physical inputs serve as narrative material—the reader need only accept that $\phi_s(x)$ is "the complex amplitude for the particle to start from slit $s$ and arrive at position $x$ on the screen." Readers unfamiliar with diffraction can skip directly to the boxed conclusion $P(x)$ and the visibility formula $\mathcal{V} = 2|\rho_{01}|$ below; this does not affect the reading of this chapter or later chapters.
```

---

**1. Quantum description of the double-slit experiment**

Suppose a quantum particle (an electron, a photon, or even a fullerene molecule) passes through a double slit, with slit 0 and slit 1 corresponding to two localized states $|0\rangle$ and $|1\rangle$. If the state of the particle is a coherent superposition:
$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle, \quad |\alpha|^2 + |\beta|^2 = 1$$

then the density matrix is:
$$\boldsymbol{\rho}_{\text{pure}} = |\psi\rangle\langle\psi| = \begin{pmatrix} |\alpha|^2 & \alpha\beta^* \\ \alpha^*\beta & |\beta|^2 \end{pmatrix}$$

The off-diagonal entry $\rho_{01} = \alpha\beta^*$ is nonzero.

---

**2. Deriving the interference fringes**

The wave function of the particle in the position eigenstates $|x\rangle$ is:
$$\psi(x) = \langle x|\psi\rangle = \alpha\,\phi_0(x) + \beta\,\phi_1(x)$$

where $\phi_s(x) = \langle x|s\rangle$ is the **diffraction amplitude** (complex-valued) for the particle to start from slit $s$ and arrive at position $x$ on the screen. In the far-field approximation (the Fraunhofer approximation), the diffraction amplitude takes the form of a spherical wave $$\phi_s(x) \propto \frac{e^{ikr_s(x)}}{r_s(x)}$$ where $r_s(x)$ is the distance from slit $s$ to position $x$ and $k = 2\pi/\lambda$ is the wavenumber. Because the observation distance $L$ is much larger than the slit separation $d$ (that is, $L \gg d$), the amplitude attenuation factors $1/r_s(x) \approx 1/L$ of the two paths are nearly equal and cancel when we compute the interference term; what really determines the positions of the fringes is only the phase difference $k(r_0(x) - r_1(x))$ in the exponent.


Extending the Born rule to the continuous spectrum (here $P(x)$ is a **probability density**, which must be integrated over $x$ to give a probability; a rigorous treatment of the continuous spectrum is beyond the scope of this book), the probability density of detecting the particle at $x$ is:
$$P(x) = |\psi(x)|^2 = |\alpha\,\phi_0(x) + \beta\,\phi_1(x)|^2$$

Expanding the squared modulus, the last two cross terms are complex conjugates of each other, and combining them gives:
$$P(x) = |\alpha|^2|\phi_0(x)|^2 + |\beta|^2|\phi_1(x)|^2 + \alpha\beta^*\phi_0(x)\phi_1^*(x) + \alpha^*\beta\,\phi_0^*(x)\phi_1(x)$$
$$= |\alpha|^2 P_0(x) + |\beta|^2 P_1(x) + 2\,\text{Re}\!\left[\alpha\beta^*\,\phi_0(x)\phi_1^*(x)\right]$$

where $P_s(x) = |\phi_s(x)|^2$ is the **single-slit diffraction intensity** when only slit $s$ is open and the other slit is closed.

**Physical meaning of the three terms:**

- **First and second terms**: the diffraction envelopes of the individual slits, weighted by the probabilities $|\alpha|^2$ and $|\beta|^2$; they are independent of each other, with **no interference**
- **Third term**: the **interference term** produced by the superposition of the amplitudes of the two paths, equal to $2\operatorname{Re}[\rho_{01}\phi_0(x)\phi_1^*(x)]$ (for a pure state $\rho_{01}=\alpha\beta^*$)

Writing $\phi_s(x) = |\phi_s(x)|\,e^{i\varphi_s(x)}$, the interference term expands to:
$$2|\alpha||\beta|\,|\phi_0(x)||\phi_1(x)|\cdot \cos\!\bigl[(\varphi_0(x) - \varphi_1(x)) + (\theta_\alpha - \theta_\beta)\bigr]$$

where $\theta_\alpha, \theta_\beta$ are the phases of $\alpha, \beta$. The phase difference $\varphi_0(x) - \varphi_1(x) = k(r_0(x) - r_1(x))$ oscillates periodically in $x$, which produces exactly the **alternating bright and dark interference fringes** on the screen.

Now note that $\alpha\beta^* = \rho_{01}$, so the interference term can be expressed directly in terms of the density matrix:
$$P(x) = \rho_{00}\,P_0(x) + \rho_{11}\,P_1(x) + 2\,\text{Re}\!\left[\rho_{01}\,\phi_0(x)\phi_1^*(x)\right]$$

**The interference fringes are determined entirely by the off-diagonal entry $\rho_{01}$:**
- $\rho_{01} \neq 0$: the interference term is present and fringes appear on the screen; **wave behavior shows**
- $\rho_{01} = 0$: the interference term vanishes and only the sum of the two diffraction envelopes remains on the screen; **particle behavior shows**

For equal slit intensities ($|\phi_0| = |\phi_1|$), the **visibility** of the fringes is:
$$\mathcal{V} = \frac{P_{\max} - P_{\min}}{P_{\max} + P_{\min}} = 2|\rho_{01}|$$

The larger $|\rho_{01}|$, the sharper the fringes; when $\rho_{01} = 0$, $\mathcal{V} = 0$ and the fringes disappear completely. This is the clearest example showing that **the modulus of an off-diagonal entry can be measured directly in experiment**.

---

**3. "Which-slit" information and its complementarity with coherence**

If a path detector is placed next to the slits in an attempt to determine which slit the particle goes through, the interference fringes disappear—this is the most famous "which-path" phenomenon of the double-slit experiment. The density matrix gives a precise quantitative description.

After the detector interacts with the particle, the two form an entangled state; after taking the partial trace over the detector (discussed in detail in §11.5), the off-diagonal entry of the particle is reduced:
$$\rho_{01} \;\rightarrow\; \rho_{01}\cdot\langle d_1|d_0\rangle$$

where $|d_s\rangle$ is the state of the detector that records "the particle went through slit $s$":
- **Fully successful detection** ($|d_0\rangle \perp |d_1\rangle$, complete path information): $\langle d_0|d_1\rangle = 0$, the off-diagonal entry is cleared, $\mathcal{V} = 0$, and the fringes disappear completely
- **Incomplete detection** ($|\langle d_0|d_1\rangle| = \varepsilon \in (0,1)$): $\mathcal{V}$ is reduced to $\varepsilon$ times its original value
- **No detection** ($|d_0\rangle = |d_1\rangle$, no path information): $\varepsilon = 1$, and the fringes are fully preserved

**Acquiring path information and the sharpness of the interference fringes are in a precise complementary relationship**; this is exactly Bohr's **complementarity principle** expressed in the language of density matrices.

---

**4. Coherence in nature: magnetoreception in migratory birds and decoherence**

The role of quantum coherence in nature is an important research topic in quantum biology. The radical-pair model of geomagnetic navigation in migratory birds proposes the following: after cryptochrome in the retina absorbs light, electron transfer within the protein forms spin-correlated radical pairs, and the dynamics between singlet and triplet states may convert the direction of the magnetic field into differences in chemical yield. Cryptochrome 4 of the European robin has shown magnetically sensitive photochemistry in in vitro experiments; this result has not yet established the complete in vivo sensing and neural transmission mechanism, however, and it does not by itself justify the claim that birds can "see" a light-and-dark geomagnetic image. The model provides a concrete direction for studying biological function in terms of quantum states, coherence, and measurement statistics.

Source of the research: [Xu et al., *Nature* 594, 535–540 (2021)](https://www.nature.com/articles/s41586-021-03618-9).

Over longer time scales, however, interactions with the environment can also act as a destroyer: if environmental degrees of freedom such as molecular thermal vibrations form distinguishable records, the phase information leaks into the environment and the off-diagonal entries of the reduced state are suppressed. In commonly used approximate models of open systems (such as the Markov approximation), this decay is often written in exponential form:

$$\rho_{01}(t) \approx \rho_{01}(0)\,e^{-\Gamma t}$$

where $T_2 = 1/\Gamma$ is the **coherence time**. In the radical-pair model, the absorption of light triggers the formation of spin-correlated states, while coupling to the environment suppresses the coherence of the reduced state; the competition between the two time scales determines how long coherence can last. The formula above is the result of approximate models of open systems such as the **Lindblad master equation**, not the necessary form of all dynamics; the relevant theory is beyond the scope of this book.

This is exactly the central engineering challenge facing quantum computers: the $T_2$ of superconducting qubits is typically on the order of microseconds to milliseconds, and all quantum gate operations must be completed within $T_2$; otherwise the computational information will be irreversibly dissipated into the environment through decoherence.
````

:::{prf:example} Spectral Decomposition of a Mixed State
:label: ex-spectral-decomposition-mixed

Consider a non-diagonal mixed state:
$$\rho = \begin{pmatrix} 0.7 & 0.2 \\ 0.2 & 0.3 \end{pmatrix}$$

This density matrix has **nonzero off-diagonal entries**, which means the system has some degree of coherence. Let us find its spectral decomposition.

**Step 1. Compute the eigenvalues**

The characteristic polynomial:
$$\det(\rho - \lambda \mathbf{I}) = \det\begin{pmatrix} 0.7-\lambda & 0.2 \\ 0.2 & 0.3-\lambda \end{pmatrix} = (0.7-\lambda)(0.3-\lambda) - 0.04$$

$$= \lambda^2 - \lambda + 0.21 - 0.04 = \lambda^2 - \lambda + 0.17 = 0$$

Solving gives:
$$\lambda_0 = \frac{1 + \sqrt{1-0.68}}{2} = \frac{1 + \sqrt{0.32}}{2} \approx 0.783$$
$$\lambda_1 = \frac{1 - \sqrt{0.32}}{2} \approx 0.217$$

Check: $\lambda_0 + \lambda_1 = 1$ ✓, and $\lambda_0, \lambda_1 > 0$ ✓

**Step 2. Find the eigenvectors**

For $\lambda_0 \approx 0.783$:
$$(\rho - \lambda_0 \mathbf{I})|e_0\rangle = 0$$
$$\begin{pmatrix} 0.7-0.783 & 0.2 \\ 0.2 & 0.3-0.783 \end{pmatrix}\begin{pmatrix} x_0 \\ x_1 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

From row 0: $-0.083 x_0 + 0.2 x_1 = 0$ → $x_1 = 0.415 x_0$

Normalize: $x_0^2 + x_1^2 = 1$ → $x_0^2(1 + 0.415^2) = 1$

Solving gives:
$$|e_0\rangle \approx \begin{pmatrix} 0.924 \\ 0.383 \end{pmatrix}$$

Similarly, for $\lambda_1 \approx 0.217$:
$$|e_1\rangle \approx \begin{pmatrix} -0.383 \\ 0.924 \end{pmatrix}$$

Check orthogonality: $\langle e_0|e_1\rangle = 0.924 \times (-0.383) + 0.383 \times 0.924 = 0$ ✓

**Step 3. The spectral decomposition**

$$\rho = 0.783 |e_0\rangle\langle e_0| + 0.217 |e_1\rangle\langle e_1|$$

**Physical interpretation**:
- This mixed state is equivalent to: the system is in the state $|e_0\rangle$ with probability 78.3% and in the state $|e_1\rangle$ with probability 21.7%
- These two states $|e_0\rangle$ and $|e_1\rangle$ are **orthogonal** and form the eigenbasis of the system

**Purity check**:
$$\text{Tr}(\rho^2) = \lambda_0^2 + \lambda_1^2 = 0.783^2 + 0.217^2 \approx 0.613 + 0.047 = 0.660 < 1$$

This confirms that $\rho$ is a mixed state.
:::

```{note}
:label: note-spectral-uniqueness
**Uniqueness of the spectral decomposition.**

For a given density matrix $\rho$:
- The spectral decomposition $\rho = \sum_j \lambda_j |e_j\rangle\langle e_j|$ is **unique** (assuming the eigenvalues are all distinct)
- But a general decomposition $\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i|$ is **not unique**

In the nondegenerate case the rank-one eigenprojections are unique, although the eigenvectors can still be multiplied by a phase; in the degenerate case, uniqueness holds only after the projections belonging to the same eigenvalue are combined. This is also why the spectral decomposition is especially important in quantum information theory.
```

#### Two Levels of Probability in a Mixed State

A subtle point about mixed states is that they contain **two levels of probability**:

:::{prf:property} Levels of Probability in a Mixed State
:label: pro-two-level-probability

In the mixed state $\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i|$:

**First level: classical probability** $p_i$
- The probability that the system is in a particular pure state $|\psi_i\rangle$
- Comes from incomplete knowledge of how the system was prepared or of its history
- This is **epistemic** uncertainty

**Second level: quantum probability** $|c_{ij}|^2$
- In a particular pure state $|\psi_i\rangle = \sum_j c_{ij} |j\rangle$, the probability that a measurement yields the basis state $|j\rangle$
- Comes from the intrinsic randomness of quantum superposition (the Born rule)
- This is **ontological** uncertainty
:::

:::{prf:property} ◇Measurement Probabilities for a Mixed State
:label: pro-mixed-measurement-probability

In the mixed state $\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i|$, when we measure in the basis $\{|j\rangle\}$, the probability of obtaining $|j\rangle$ is:

$$P(j) = \langle j|\rho|j\rangle = \rho_{jj}$$

**Important conclusion**: the **diagonal entries** $\rho_{jj}$ of the density matrix directly give the probabilities of obtaining $|j\rangle$ when measuring in that basis.

For the detailed theory of measurement, see §10.4.
:::

:::{admonition} §10.3 Summary
:class: tip
This section answered the question: when our knowledge of how a quantum system was prepared is incomplete, how do we describe its state and its statistical properties? The answer is the density matrix $\boldsymbol{\rho} = \sum_i p_i|\psi_i\rangle\langle\psi_i|$—a Hermitian, positive semidefinite matrix with trace one that handles pure states and mixed states in a unified way. Its standard form is guaranteed directly by the spectral theorem of §9.3: $\boldsymbol{\rho} = \sum_j \lambda_j |e_j\rangle\langle e_j|$, where the eigenvalue $\lambda_j$ is the probability weight of each eigenstate, and the purity $\text{Tr}(\boldsymbol{\rho}^2) = \sum_j \lambda_j^2$ distinguishes exactly between pure states (equal to 1) and mixed states (strictly less than 1).

Pure states and mixed states are distinguished by purity: $\operatorname{Tr}(\rho^2)=1$ for a pure state and $<1$ for a mixed state; coherence, on the other hand, depends on the chosen basis. In the path basis, the double-slit interference term is $2\operatorname{Re}[\rho_{01}\phi_0(x)\phi_1^*(x)]$, which also depends on the propagation amplitudes and phases. A path detector or certain couplings to the environment can suppress the off-diagonal entries of the reduced state, but this does not mean that the overall correlations necessarily disappear for good.

With the density matrix as a tool, §10.4 can give a complete description of how the state changes before and after a measurement: the act of measurement makes the off-diagonal entries vanish (decoherence), and after the outcome is read out, the system collapses to a definite eigenstate (the state becomes pure).
:::

## 10.4 Quantum Measurement

In §10.3 we developed the complete theory of the density matrix. We now turn to another central topic of quantum mechanics: **quantum measurement**. Measurement is the bridge between the abstract mathematical formalism and experimental observation, and it is also the most mysterious and most controversial part of quantum mechanics.

#### The Nature of Quantum Measurement

Quantum measurement differs fundamentally from classical measurement:

```{important}
:label: imp-quantum-vs-classical-measurement
**Quantum measurement vs. classical measurement.**

**Classical measurement**:
- Passive observation that does not disturb the system being measured
- The state of the system is the same before and after the measurement
- The measurement outcome is completely determined by the state of the system (no randomness)
- All physical quantities can be measured simultaneously with unlimited precision

**Quantum measurement**:
- A **physical action** that may in general disturb the system; eigenstates can remain unchanged
- The state is updated conditionally on the outcome (state collapse); when the outcome is ignored, states compatible with the block structure of the measurement projections can remain unchanged
- Measurement outcomes have **intrinsic randomness** (the Born rule)
- **Incompatible observables** have no common complete eigenbasis; the uncertainty within a state is constrained by the corresponding inequality (the uncertainty principle, §10.5)
```

#### Projective Measurement

The mathematical description of quantum measurement is built on the notion of an **observable**:

:::{prf:definition} Projective Measurement
:label: def-projection-measurement

Let the observable $\hat{A}$ be a nondegenerate Hermitian matrix (or let the measurement distinguish every basis-vector label), with spectral decomposition:
$$\hat{A} = \sum_j a_j |\mathbf{a}_j\rangle\langle \mathbf{a}_j|$$

where $a_j \in \mathbb{R}$ are the eigenvalues and $\{|\mathbf{a}_j\rangle\}$ are the corresponding orthogonal eigenstates.

A **projective measurement** of $\hat{A}$ on the pure state $|\psi\rangle$:

1. **Measurement outcome**: the result can only be some eigenvalue $a_j$
   
2. **Measurement probability**: the probability of obtaining $a_j$ is
   $$P(a_j) = |\langle \mathbf{a}_j|\psi\rangle|^2$$

3. **State collapse**: after the measurement, the system collapses to the corresponding eigenstate
   $$|\psi\rangle \rightarrow |\mathbf{a}_j\rangle$$

4. **Expectation value**: the expectation value of the observable is
   $$\mathbb{E}_{\psi}[\hat{A}] = \sum_j a_j P(a_j) = \langle\psi|\hat{A}|\psi\rangle$$
:::

```{note}
:label: note-measurement-quantization
**Quantization of measurement outcomes.**

The outcome of a quantum measurement **must be an eigenvalue of the observable**; this is a central feature of quantum mechanics:
- A measurement cannot yield an arbitrary value between eigenvalues
- For example, if the eigenvalues of $\hat{A}$ are $\{0, 1\}$, the measurement outcome can only be 0 or 1, never 0.5
- This **discreteness** is where the word "quantum" comes from
```

#### Distinguishing the Act of Measurement from Reading Out the Outcome

The Schrödinger's cat thought experiment reminds us to distinguish among the overall state, the reduced state, and the conditional state. We first use two apparatus branches $|a_0\rangle,|a_1\rangle$ to represent idealized macroscopic records; before the apparatus is coupled, one cannot simply call the superposition of the atom a cat that has already split into two branches.

Once the coupling of the apparatus has established correlations, the overall pure state of the system and the environment can be written as
$$|\Psi\rangle=\alpha|a_0\rangle|E_0\rangle+\beta|a_1\rangle|E_1\rangle,\qquad |\alpha|^2+|\beta|^2=1.$$
Ignoring the environment and taking the partial trace over it gives the reduced state of the apparatus
$$\boldsymbol{\rho}_{\mathrm{red}}=
\begin{pmatrix}|\alpha|^2&\alpha\overline\beta\langle E_1|E_0\rangle\\
\overline\alpha\beta\langle E_0|E_1\rangle&|\beta|^2\end{pmatrix}.$$
When the environment records are nearly orthogonal, the off-diagonal entries are suppressed and the reduced state is approximately a diagonal mixed state. This is the decoherence approximation; the overall state can still be pure, and the information in the correlations has not disappeared permanently in the mathematical sense.

After an outcome of positive probability is read out, conditioning according to this chapter's postulate of ideal projective measurement gives the conditional state of the corresponding branch. Decoherence explains why locally visible interference is suppressed, but it does not by itself derive, from unitary evolution alone, a unique measurement outcome in each run of the experiment.

| Level of description | State and information |
|---|---|
| System and environment as a whole | Can remain a pure state containing the correlations between branches |
| Reduced state, ignoring the environment | Approximately a diagonal mixed state when the environment records are nearly orthogonal |
| Conditional state after reading out the outcome | Updated according to the measurement postulate given the known outcome; differs from the averaged state when the outcome is not read out |




To give each row of the table above a precise mathematical counterpart, we need to establish three things in order: the trace formula provides a unified tool for computing expectation values, projective measurement of mixed states gives a precise expression for the probabilities, and only then come the formal definitions of the conditional state and the averaged state.

#### The Trace Formula for Mixed States

:::{prf:lemma} The Trace Formula for Mixed States
:label: lem-trace-formula-mixed

For a mixed state $\boldsymbol{\rho} = \sum_i p_i |\psi_i\rangle\langle\psi_i|$ and an observable $\hat{A}$:

$$\text{Tr}(\boldsymbol{\rho}\hat{A}) = \sum_i p_i \langle\psi_i|\hat{A}|\psi_i\rangle$$

**Proof**: compute the trace using a complete basis $\{|j\rangle\}$:

$$\text{Tr}(\boldsymbol{\rho}\hat{A}) = \sum_j \langle j|\boldsymbol{\rho}\hat{A}|j\rangle = \sum_j \sum_i p_i \langle j|\psi_i\rangle\langle\psi_i|\hat{A}|j\rangle$$

$$= \sum_i p_i \sum_j \langle\psi_i|\hat{A}|j\rangle\langle j|\psi_i\rangle = \underline{\sum_i p_i \langle\psi_i|\hat{A}|\psi_i\rangle}$$
:::

The trace formula shows that $\text{Tr}(\boldsymbol{\rho}\hat{A})$ is the classical weighted average of the expectation values in the individual pure states; this is the central computational power of the density matrix as a statistical tool. In particular, taking $\hat{A} = \boldsymbol{\Pi}_j := |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$ gives a unified formula for measurement probabilities.

#### Projective Measurement of Mixed States

:::{prf:definition} Projective Measurement of a Mixed State
:label: def-mixed-state-projection

Perform a projective measurement of the observable $\hat{A} = \sum_j a_j |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$ on the mixed state $\boldsymbol{\rho} = \sum_i p_i |\psi_i\rangle\langle\psi_i|$, where the projection operators are $\boldsymbol{\Pi}_j := |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$:

1. **Measurement probability**: the probability of obtaining $a_j$ is
   $$P(a_j) = \sum_i p_i |\langle\mathbf{a}_j|\psi_i\rangle|^2 = \text{Tr}(\boldsymbol{\Pi}_j \boldsymbol{\rho})$$

2. **Expectation value**: the expectation value of the observable is
   $$\mathbb{E}_{\boldsymbol{\rho}}[\hat{A}] = \sum_j a_j P(a_j) = \text{Tr}(\boldsymbol{\rho}\hat{A})$$
:::

```{note}
**A note on notation**: $P(a_j) \in \mathbb{R}$ is a scalar probability; $\boldsymbol{\Pi}_j \in M_n(\mathbb{C})$ is a matrix operator. This book uses $\boldsymbol{\Pi}_j$ (a bold Greek letter) exclusively for projection operators and $P(\cdot)$ (written as a function with parentheses) exclusively for probabilities.
```

With the probability formula $P(a_j) = \text{Tr}(\boldsymbol{\Pi}_j \boldsymbol{\rho})$ in hand, we can define precisely the two kinds of post-measurement states.

#### The Conditional State and the Averaged State

:::{prf:definition} Conditional State and Averaged State
:label: def-conditional-averaged-state

Perform a projective measurement of the observable $\hat{A} = \sum_j a_j |\mathbf{a}_j\rangle\langle \mathbf{a}_j|$ on the state $\boldsymbol{\rho}$, where the projection operators are $\boldsymbol{\Pi}_j := |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$:

**Conditional state** (the measurement outcome is known to be $a_j$, and $P(a_j)>0$):

$$\boldsymbol{\rho}_{\text{cond}}^{(j)} := \frac{\boldsymbol{\Pi}_j \boldsymbol{\rho}\, \boldsymbol{\Pi}_j}{\text{Tr}(\boldsymbol{\Pi}_j \boldsymbol{\rho})}$$

This is a **pure state**; it describes the state of the system "given that the outcome $a_j$ was obtained."

**Averaged state** (the outcome is not read out, or we average over all possible outcomes):

$$\boldsymbol{\rho}_{\text{avg}} := \sum_j P(a_j) \cdot |\mathbf{a}_j\rangle\langle\mathbf{a}_j| = \sum_j \text{Tr}(\boldsymbol{\Pi}_j \boldsymbol{\rho}) \cdot |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$$

where $P(a_j) = \text{Tr}(\boldsymbol{\Pi}_j \boldsymbol{\rho}) \in [0,1]$ is the probability of obtaining the outcome $a_j$. This is **usually a mixed state** (unless the initial state is already an eigenstate), but it is necessarily **diagonal** (its off-diagonal entries in the measurement basis are 0).
:::



**Summary of the key differences**:

| | Conditional state $\rho_{\text{cond}}^{(j)}$ | Averaged state $\rho_{\text{avg}}$ |
|---|---|---|
| **Premise** | The measurement outcome $a_j$ is known | The outcome is unknown or ignored |
| **Type of state** | Pure state | Usually a mixed state |
| **Purity** | $\text{Tr}(\rho^2) = 1$ | $\text{Tr}(\rho^2) \leq 1$ |
| **Off-diagonal entries** | 0 (projection onto a pure state) | 0 (decoherence) |
| **Diagonal entries** | A single nonzero entry (= 1) | Several nonzero entries (a probability distribution) |
| **Physical meaning** | A single run with a known outcome | Statistics over many runs, or a single run with an unknown outcome |
| **Mathematical form** | $\|a_j\rangle\langle a_j\|$ | $\sum_j P(a_j)\|a_j\rangle\langle a_j\|$ |

:::{prf:example} Conditional State vs. Averaged State
:label: ex-conditional-vs-averaged

Consider a $\sigma_z$ measurement on the pure state $|\psi\rangle = \frac{1}{\sqrt{2}}(|↑_z\rangle + |↓_z\rangle)$.

**Before the measurement**:
$$\rho_{\text{init}} = |\psi\rangle\langle\psi| = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$$
- Purity: $\text{Tr}(\rho_{\text{init}}^2) = 1$ (a pure state)
- Off-diagonal entries are present (quantum coherence)

**After the measurement (conditional states)**:

If $+1$ is obtained (probability 50%):
$$\rho_{\text{cond}}^{(+)} = |↑_z\rangle\langle↑_z| = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$$
- Purity: $\text{Tr}((\rho_{\text{cond}}^{(+)})^2) = 1$ ✓ (a pure state)

If $-1$ is obtained (probability 50%):
$$\rho_{\text{cond}}^{(-)} = |↓_z\rangle\langle↓_z| = \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}$$
- Purity: $\text{Tr}((\rho_{\text{cond}}^{(-)})^2) = 1$ ✓ (a pure state)

Both conditional states are **pure states**.

**After the measurement (averaged state)**:
$$\rho_{\text{avg}} = 0.5 \cdot \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} + 0.5 \cdot \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = \frac{1}{2}\begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}$$
- Purity: $\text{Tr}(\rho_{\text{avg}}^2) = 0.5 < 1$ (a mixed state)
- No off-diagonal entries (decoherence)
- This is the maximally mixed state

**Comparison of the evolutions**:
$$\text{Before measurement: } \begin{pmatrix} 0.5 & 0.5 \\ 0.5 & 0.5 \end{pmatrix} \quad \xrightarrow{\text{measure + read out}} \quad \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \text{ or } \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}$$

$$\text{Before measurement: } \begin{pmatrix} 0.5 & 0.5 \\ 0.5 & 0.5 \end{pmatrix} \quad \xrightarrow{\text{measure (not read out)}} \quad \begin{pmatrix} 0.5 & 0 \\ 0 & 0.5 \end{pmatrix}$$

The act of measurement **removes the off-diagonal entries** (decoherence), but if the outcome is not read out, the system is still in a mixed state.
:::

```{note}
:label: note-when-to-use-which
**Which description should be used when?**

- **Conditional state**: describes the state of the system in a single run after the measurement outcome is known
  - Example: "After spin up is obtained, the system is in $|↑_z\rangle$"
  - Experimental situation: the display of the measuring instrument has been read
  
- **Averaged state**: describes the statistical ensemble over many runs, or a single run whose outcome is not read out
  - Example: "Measuring $\sigma_z$ on a large number of identical systems, 50% are in $|↑_z\rangle$ and 50% are in $|↓_z\rangle$"
  - Experimental situation: statistics over many repeated runs, or discarding the instrument reading after the measurement

**Convention of this book**: unless stated otherwise, **"the post-measurement state" means the conditional state** (the outcome has been read out). When we discuss the averaged state, we will label it explicitly as the "averaged state" or say that the "outcome is not read out."
```

#### Pauli Matrices: The Mathematical Framework of Spin Measurement

With the complete conceptual framework of measurement in place, we now introduce the most important concrete example in quantum mechanics: **the spin-1/2 system**.

:::{prf:definition} Pauli Matrices and Spin Operators
:label: def-pauli-matrices

The **Pauli matrices** are the basic $2 \times 2$ Hermitian matrices that describe the spin state of a spin-1/2 particle (such as an electron or a proton):

$$\sigma_x = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}, \quad
\sigma_y = \begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix}, \quad
\sigma_z = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$$

The corresponding **spin operators** are:
$$\hat{S}_x = \frac{ℏ}{2}\sigma_x, \quad 
\hat{S}_y = \frac{ℏ}{2}\sigma_y, \quad 
\hat{S}_z = \frac{ℏ}{2}\sigma_z$$

where $ℏ = \frac{h}{2\pi} \approx 1.055 \times 10^{-34}$ J·s is the reduced Planck constant.
:::

**Physical background**:
- Spin is the **intrinsic angular momentum** of a quantum particle, distinct from orbital angular momentum
- For a spin-1/2 particle, a measurement of spin along any direction has only two possible outcomes: $+\frac{ℏ}{2}$ ("up") or $-\frac{ℏ}{2}$ ("down")
- The spin state is completely described by the two-dimensional complex vector space $\mathbb{C}^2$

:::{prf:property} Basic Properties of the Pauli Matrices
:label: pro-pauli-properties

1. **Hermiticity**: $\sigma_j^* = \sigma_j$ for all $j \in \{x, y, z\}$

2. **Eigenvalues**: the eigenvalues of every Pauli matrix are $\pm 1$ (corresponding to the spin measurement outcomes $\pmℏ/2$)

3. **Eigenstates**:
   
   **Eigenstates of $\sigma_z$** (the computational basis):
   $$\sigma_z|↑_z\rangle = |↑_z\rangle, \quad \sigma_z|↓_z\rangle = -|↓_z\rangle$$
   where:
   $$|↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |↓_z\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

   **Eigenstates of $\sigma_x$**:
   $$|↑_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}, \quad 
     |↓_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$$
   
   Check:
   $$\sigma_x|↑_x\rangle = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = |↑_x\rangle$$ ✓

   **Eigenstates of $\sigma_y$**:
   $$|↑_y\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ i \end{pmatrix}, \quad 
     |↓_y\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -i \end{pmatrix}$$

4. **Squares**: $\sigma_j^2 = \mathbf{I}$ for all $j$

5. **Trace**: $\text{Tr}(\sigma_j) = 0$ for all $j$
:::

**Commutation relations and a preview of the uncertainty principle**:

```{important}
:label: imp-pauli-commutation
**Noncommutativity of the Pauli matrices** (see §10.5).

The Pauli matrices **do not commute** with one another; they satisfy:
$$[\sigma_x, \sigma_y] = 2i\sigma_z, \quad [\sigma_y, \sigma_z] = 2i\sigma_x, \quad [\sigma_z, \sigma_x] = 2i\sigma_y$$

**Check** (taking the first as an example):
$$[\sigma_x, \sigma_y] = \sigma_x\sigma_y - \sigma_y\sigma_x$$
$$= \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix} - \begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$$
$$= \begin{bmatrix} i & 0 \\ 0 & -i \end{bmatrix} - \begin{bmatrix} -i & 0 \\ 0 & i \end{bmatrix} = \begin{bmatrix} 2i & 0 \\ 0 & -2i \end{bmatrix} = 2i\sigma_z$$ ✓

This **noncommutativity** $[\sigma_j, \sigma_k] \neq 0$ is the mathematical root of the uncertainty principle: spin components along different directions have no common complete eigenbasis (for spin-1/2 they do not even have a common eigenstate), so they cannot have definite values simultaneously in the same state. In §10.5 we will see that the lower bound on the product of the standard deviations is determined by the expectation value of the commutator and a covariance term in that state, and it may be zero.
```

```{note}
:label: note-pauli-cyclic-relation
**Cyclic commutation relations.**

The commutation relations of the Pauli matrices can be written compactly as:
$$[\sigma_j, \sigma_k] = 2i\epsilon_{jkl}\sigma_l$$

where $\epsilon_{jkl}$ is the Levi-Civita symbol ($+1$ for cyclic permutations of $xyz$, $-1$ for cyclic permutations of $zyx$, and $0$ when an index is repeated). This relation has the same form as the Lie algebra structure $[L_j, L_k] = iℏ\epsilon_{jkl}L_l$ of quantum orbital angular momentum, revealing the mathematical nature of spin as an intrinsic angular momentum.
```

#### Spin Measurement and State Collapse

With an explicit definition of the Pauli matrices and a clear conceptual framework for measurement, we can now analyze the process of spin measurement systematically.

:::{prf:property} Measurement Collapse of a Pure State
:label: pro-pure-state-measurement-collapse

Let the pure state be $|\psi\rangle = \sum_j c_j |a_j\rangle$ (expanded in the eigenbasis of $\hat{A}$).

**Case 1: $|\psi\rangle$ is an eigenstate of $\hat{A}$**
- If $|\psi\rangle = |a_k\rangle$ for some $k$
- The measurement outcome is certain: $P(a_k) = 1$
- **The state is unchanged after the measurement** (conditional state): $|\psi\rangle \rightarrow |a_k\rangle = |\psi\rangle$

**Case 2: $|\psi\rangle$ is a superposition state** (several $c_j \neq 0$)
- The measurement outcome is random, with probabilities $|c_j|^2$
- **The state changes after the measurement** (conditional state): $|\psi\rangle \rightarrow |a_j\rangle \neq |\psi\rangle$
- The **phase information of the superposition is permanently lost**
:::

:::{prf:example} Spin Measurement and State Collapse
:label: ex-spin-measurement-collapse

Consider the pure spin state (the "up" state along $z$):
$$|\psi\rangle = |↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

**Measuring spin along $z$ ($\hat{S}_z = \frac{ℏ}{2}\sigma_z$)**:

The spin operator:
$$\hat{S}_z = \frac{ℏ}{2}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$$

Eigenvalues and eigenstates:
- $\lambda_0 = +\frac{ℏ}{2}$, eigenstate $|↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $\lambda_1 = -\frac{ℏ}{2}$, eigenstate $|↓_z\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

Since $|\psi\rangle = |↑_z\rangle$ is itself an eigenstate of $\hat{S}_z$:
- Measurement outcome: $+ℏ/2$ with probability 100%
- State after the measurement (conditional state): $|↑_z\rangle$ (**unchanged**)

---

**Measuring spin along $x$ ($\hat{S}_x = \frac{ℏ}{2}\sigma_x$)**:

First expand $|↑_z\rangle$ in the $x$ basis. Since:
$$|↑_x\rangle = \frac{1}{\sqrt{2}}(|↑_z\rangle + |↓_z\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$$
$$|↓_x\rangle = \frac{1}{\sqrt{2}}(|↑_z\rangle - |↓_z\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$$

solving back gives:
$$|↑_z\rangle = \frac{1}{\sqrt{2}}|↑_x\rangle + \frac{1}{\sqrt{2}}|↓_x\rangle$$

Check:
$$\frac{1}{\sqrt{2}} \cdot \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix} + \frac{1}{\sqrt{2}} \cdot \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix} = \frac{1}{2}\begin{pmatrix} 2 \\ 0 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$ ✓

Measurement outcomes:
- With probability 50%, $+ℏ/2$ is obtained → the state collapses to the conditional state $|↑_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$
- With probability 50%, $-ℏ/2$ is obtained → the state collapses to the conditional state $|↓_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$

**Key observation**: the state has changed fundamentally!
$$\text{Before measurement: } |↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad \neq \quad \text{conditional state (if +ℏ/2 is obtained): } |↑_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$$

**Averaged state** (if the outcome is not read out):
$$\rho_{\text{avg}} = 0.5|↑_x\rangle\langle↑_x| + 0.5|↓_x\rangle\langle↓_x| = \frac{1}{2}\mathbf{I}$$
This is a mixed state ($\text{Tr}(\rho_{\text{avg}}^2) = 0.5 < 1$), but the conditional states are all pure states.

---

**Computing expectation values**:

Measuring in the $z$ basis:
$$\mathbb{E}_{|↑_z\rangle}[\hat{S}_z] = \langle↑_z|\hat{S}_z|↑_z\rangle = \frac{ℏ}{2}\begin{pmatrix} 1 & 0 \end{pmatrix}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = \frac{ℏ}{2}$$

Measuring in the $x$ basis:
$$\mathbb{E}_{|↑_z\rangle}[\hat{S}_x] = \langle↑_z|\hat{S}_x|↑_z\rangle = \frac{ℏ}{2}\begin{pmatrix} 1 & 0 \end{pmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = 0$$

This agrees with the probabilistic prediction: 50% × $(+ℏ/2)$ + 50% × $(-ℏ/2)$ = 0 ✓
:::

#### Basis Dependence of Measurement and Noncommutativity

```{important}
:label: imp-measurement-basis-dependence
**The importance of the measurement basis.**

The same pure state, measured in different bases, collapses to **different conditional states**:
- Measuring in the basis $\mathcal{B}_A = \{|a_j\rangle\}$ (the eigenstates of the observable $\hat{A}$) → collapses to some $|a_j\rangle$
- Measuring in the basis $\mathcal{B}_B = \{|b_k\rangle\}$ (the eigenstates of the observable $\hat{B}$) → collapses to some $|b_k\rangle$

If two Hermitian observables have no common orthonormal eigenbasis (equivalently, they do not commute), order effects can arise in general. Below we distinguish the conditional states with read-out outcomes, the joint distribution of outcomes, and the averaged state without read-out:
- $|a_j\rangle \neq |b_k\rangle$ (in general)
- **The order of measurements matters**: measuring $\hat{A}$ first and then $\hat{B}$, versus measuring $\hat{B}$ first and then $\hat{A}$, may give different conditional states or joint distributions of outcomes; in particular examples, the averaged states may also turn out the same

For the Pauli matrices, since $[\sigma_x, \sigma_z] = -2i\sigma_y \neq 0$, spin measurements along $x$ and along $z$ are **incompatible**. This is precisely the root of the **uncertainty principle** (§10.5).
```

:::{prf:example} Noncommutativity of Sequential Measurements
:label: ex-sequential-measurement

Choose a pure state that is **neither an eigenstate of $\sigma_x$ nor an eigenstate of $\sigma_z$**:
$$|\psi\rangle = \frac{1}{\sqrt{5}}\begin{pmatrix} 1 \\ 2 \end{pmatrix} = \frac{1}{\sqrt{5}}|↑_z\rangle + \frac{2}{\sqrt{5}}|↓_z\rangle$$

Normalization check: $|\frac{1}{\sqrt{5}}|^2 + |\frac{2}{\sqrt{5}}|^2 = \frac{1}{5} + \frac{4}{5} = 1$ ✓

**Properties of the initial state**:
- Expectation values: $\langle\sigma_z\rangle = \frac{1}{5}(1 - 4) = -\frac{3}{5}$, $\langle\sigma_x\rangle = \frac{4}{5}$
- This state "leans down" along $z$ and "leans up" along $x$

---

**Order A: measure $\sigma_z$ first, then $\sigma_x$ (reading out both outcomes)**

**Step 1. Measure $\sigma_z$ and read out the outcome**

Probabilities:
- $P(+1) = |\frac{1}{\sqrt{5}}|^2 = \frac{1}{5} = 20\%$ → conditional state $|↑_z\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$
- $P(-1) = |\frac{2}{\sqrt{5}}|^2 = \frac{4}{5} = 80\%$ → conditional state $|↓_z\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

**Averaged state after step 1** (taking all possible outcomes into account):
$$\rho_A^{(1)} = 0.2|↑_z\rangle\langle↑_z| + 0.8|↓_z\rangle\langle↓_z| = \begin{pmatrix} 0.2 & 0 \\ 0 & 0.8 \end{pmatrix}$$

Key observations:
- Purity: $\text{Tr}((\rho_A^{(1)})^2) = 0.68 < 1$ (a mixed state)
- $\langle\sigma_z\rangle_A^{(1)} = 0.2 - 0.8 = -0.6$ (the information along $z$ is preserved)
- $\langle\sigma_x\rangle_A^{(1)} = 0$ (the information along $x$ is washed out)

---

**Step 2. Measure $\sigma_x$ and read out the outcome**

Expand: $|↑_z\rangle = \frac{1}{\sqrt{2}}(|↑_x\rangle + |↓_x\rangle)$, $|↓_z\rangle = \frac{1}{\sqrt{2}}(|↑_x\rangle - |↓_x\rangle)$

**(a) If step 1 gave $|↑_z\rangle$** (probability 20%):
- $+1$ with probability 50% → conditional state $|↑_x\rangle$
- $-1$ with probability 50% → conditional state $|↓_x\rangle$

**(b) If step 1 gave $|↓_z\rangle$** (probability 80%):
- $+1$ with probability 50% → conditional state $|↑_x\rangle$
- $-1$ with probability 50% → conditional state $|↓_x\rangle$

**Total probabilities**:
- $P(|↑_x\rangle) = 0.2 \times 0.5 + 0.8 \times 0.5 = 0.5$
- $P(|↓_x\rangle) = 0.2 \times 0.5 + 0.8 \times 0.5 = 0.5$

**Final averaged state**:
$$\rho_A = \frac{1}{2}\mathbf{I} = \begin{pmatrix} 0.5 & 0 \\ 0 & 0.5 \end{pmatrix}$$

This is the **maximally mixed state**.

---

**Order B: measure $\sigma_x$ first, then $\sigma_z$ (reading out both outcomes)**

**Step 1. Measure $\sigma_x$ and read out the outcome**

Expand $|\psi\rangle$ in the $x$ basis:
$$|\psi\rangle = \frac{1}{\sqrt{5}}|↑_z\rangle + \frac{2}{\sqrt{5}}|↓_z\rangle = \frac{3}{\sqrt{10}}|↑_x\rangle - \frac{1}{\sqrt{10}}|↓_x\rangle$$

Probabilities:
- $P(+1) = \frac{9}{10} = 90\%$ → conditional state $|↑_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$
- $P(-1) = \frac{1}{10} = 10\%$ → conditional state $|↓_x\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$

**Averaged state after step 1**:
$$\rho_B^{(1)} = 0.9|↑_x\rangle\langle↑_x| + 0.1|↓_x\rangle\langle↓_x| = \begin{pmatrix} 0.5 & 0.4 \\ 0.4 & 0.5 \end{pmatrix}$$

Key observations:
- Purity: $\text{Tr}((\rho_B^{(1)})^2) = 0.82 < 1$ (a mixed state)
- $\langle\sigma_x\rangle_B^{(1)} = 0.8$ (the information along $x$ is preserved)
- $\langle\sigma_z\rangle_B^{(1)} = 0$ (the information along $z$ is washed out)

---

**Step 2. Measure $\sigma_z$ and read out the outcome**

A similar computation shows that the final averaged state is again:
$$\rho_B = \frac{1}{2}\mathbf{I} = \begin{pmatrix} 0.5 & 0 \\ 0 & 0.5 \end{pmatrix}$$

---

**The key manifestation of noncommutativity**:

| | Initial state | After step 1 (averaged state) | Final (averaged state) |
|---|---|---|---|
| **Order A** | $\langle\sigma_z\rangle = -0.6$ | $\langle\sigma_z\rangle = -0.6$, $\langle\sigma_x\rangle = 0$ | $\langle\sigma_z\rangle = 0$, $\langle\sigma_x\rangle = 0$ |
| **Order B** | $\langle\sigma_z\rangle = -0.6$ | $\langle\sigma_z\rangle = 0$, $\langle\sigma_x\rangle = 0.8$ | $\langle\sigma_z\rangle = 0$, $\langle\sigma_x\rangle = 0$ |

**Core lessons**:

1. **Measurement preserves information selectively**: the first measurement preserves the information along its own direction and washes out the other directions

2. **The second measurement completely washes out the remaining information**: in this example, for mutually perpendicular spin directions, averaging over the outcomes of both measurements gives the maximally mixed state

3. **Different intermediate states, the same final state**:
   - $\rho_A^{(1)} = \text{diag}(0.2, 0.8)$ vs. $\rho_B^{(1)} = \begin{pmatrix} 0.5 & 0.4 \\ 0.4 & 0.5 \end{pmatrix}$
   - Diagonal vs. non-diagonal

4. **Physical meaning of noncommutativity**: $[\sigma_x, \sigma_z] \neq 0$ ⇒ measuring one operator destroys the information about the other
:::

```{note}
:label: note-maximal-mixing-inevitable
**Averaged measurements along perpendicular spin directions.**

For a spin-1/2 system, if ideal projective measurements are performed along two mutually perpendicular spin directions and we average over the outcomes of both measurements, the final state is $\rho=\mathbf{I}/2$. This conclusion is not guaranteed for arbitrary noncommuting directions; and the conditional states obtained by reading out each outcome also differ from this averaged state.
```

#### Projective Measurement (Mixed States)


Recall {prf:ref}`def-mixed-state-projection` and {prf:ref}`def-conditional-averaged-state`: when a projective measurement of the observable $\hat{A} = \sum_j a_j |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$ (with projection operators $\boldsymbol{\Pi}_j := |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$) is performed on a mixed state $\boldsymbol{\rho}$, the probability of obtaining $a_j$ is $P(a_j) = \langle\mathbf{a}_j|\boldsymbol{\rho}|\mathbf{a}_j\rangle = \text{Tr}(\boldsymbol{\Pi}_j\boldsymbol{\rho})$; if the **outcome is read out**, the conditional state $\boldsymbol{\rho}_{\text{cond}}^{(j)} = \boldsymbol{\Pi}_j\boldsymbol{\rho}\,\boldsymbol{\Pi}_j\,/\,\text{Tr}(\boldsymbol{\Pi}_j\boldsymbol{\rho}) = |\mathbf{a}_j\rangle\langle\mathbf{a}_j|$ is necessarily a pure state. We now apply this general theory to concrete mixed states described with the Pauli matrices.

```{important}
:label: imp-mixed-to-pure-collapse
**In a rank-one projective measurement, the conditional state for an outcome of positive probability is a pure state.**

Whatever mixed state the system starts in, if this chapter's rank-one projective measurement is performed **and an outcome of positive probability is read out**, the conditional state is the pure state $|\mathbf{a}_j\rangle\langle \mathbf{a}_j|$.

**Physical explanation**: reading out the outcome **removes the classical uncertainty**
- Before the measurement: we do not know which pure state $|\psi_i\rangle$ the system is in (probability $p_i$)
- After reading out the outcome: we know for certain that the system is in the pure state $|\mathbf{a}_j\rangle$

The quantum uncertainty is converted into the selection of a measurement outcome (which $a_j$), but the conditional state after the outcome is read out is itself a definite pure state.

**Contrast with the averaged state**: if the outcome is not read out, the system is usually in a mixed state (exceptions such as eigenstates can remain pure) (with diagonal entries distributed according to the probabilities), but the off-diagonal entries have been removed by the act of measurement.
```

:::{prf:example} A Concrete Example of Measuring a Mixed State
:label: ex-mixed-state-measurement-detailed

Consider a more general mixed spin state (with off-diagonal entries):
$$\rho = \frac{1}{4}\begin{pmatrix} 3 & 1 \\ 1 & 1 \end{pmatrix}$$

**Step 1. Verify that this is a valid density matrix**

(a) Hermiticity: $\rho^* = \frac{1}{4}\begin{pmatrix} 3 & 1 \\ 1 & 1 \end{pmatrix} = \rho$ ✓

(b) Trace one: $\text{Tr}(\rho) = \frac{1}{4}(3 + 1) = 1$ ✓

(c) Positive semidefiniteness: the eigenvalues are $\lambda_0 \approx 0.854$ and $\lambda_1 \approx 0.146$ (both $> 0$) ✓

(d) Verify that this is a mixed state:
$$\text{Tr}(\rho^2) = \frac{3}{4} < 1$$ ✓

**Step 2. Physical interpretation**

The nonzero off-diagonal entries $\rho_{01} = \rho_{10} = \frac{1}{4} \neq 0$ indicate that the system has **quantum coherence**.

---

**Measuring spin along $z$ ($\sigma_z$) and reading out the outcome**:

The measurement basis is $\{|↑_z\rangle, |↓_z\rangle\}$.

**Probability of obtaining "up" (eigenvalue $+1$, spin $+ℏ/2$)**:
$$P(↑) = \langle↑_z|\rho|↑_z\rangle = \rho_{00} = \frac{3}{4} = 75\%$$

**Conditional state** (outcome read out as "up"):
$$\rho_{\text{cond}}^{(↑)} = |↑_z\rangle\langle↑_z| = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$$

Check the purity: $\text{Tr}((\rho_{\text{cond}}^{(↑)})^2) = 1$ ✓ (a pure state)

**Probability of obtaining "down" (eigenvalue $-1$, spin $-ℏ/2$)**:
$$P(↓) = \langle↓_z|\rho|↓_z\rangle = \rho_{11} = \frac{1}{4} = 25\%$$

**Conditional state** (outcome read out as "down"):
$$\rho_{\text{cond}}^{(↓)} = |↓_z\rangle\langle↓_z| = \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}$$

Check the purity: $\text{Tr}((\rho_{\text{cond}}^{(↓)})^2) = 1$ ✓ (a pure state)

**Averaged state** (outcome not read out):
$$\rho_{\text{avg}} = 0.75 \cdot \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} + 0.25 \cdot \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 0.75 & 0 \\ 0 & 0.25 \end{pmatrix}$$

Check the purity: $\text{Tr}(\rho_{\text{avg}}^2) = 0.625 < 1$ (a mixed state, but without coherence)

**Comparison**:
- Before the measurement: $\rho = \begin{pmatrix} 0.75 & 0.25 \\ 0.25 & 0.25 \end{pmatrix}$ (coherence present)
- Averaged state: $\rho_{\text{avg}} = \begin{pmatrix} 0.75 & 0 \\ 0 & 0.25 \end{pmatrix}$ (no coherence, still mixed)
- Conditional state: $\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$ or $\begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}$ (a pure state)

**Key observations**:
- **The act of measurement**: removes the off-diagonal entries $0.25 \to 0$ (decoherence)
- **Reading out the outcome**: further purifies the state to a single eigenstate (the conditional state)

**Expectation value**:
$$\mathbb{E}_{\rho}[\sigma_z] = \text{Tr}(\rho\sigma_z) = \frac{1}{2}$$

Check: $75\% \times (+1) + 25\% \times (-1) = 0.5$ ✓

---

**Measuring spin along $x$ ($\sigma_x$) and reading out the outcome**:

The measurement basis is $\{|↑_x\rangle, |↓_x\rangle\}$.

A similar computation (omitted) gives:
- $P(↑_x) = 75\%$, $P(↓_x) = 25\%$
- The conditional state is $|↑_x\rangle\langle↑_x|$ or $|↓_x\rangle\langle↓_x|$ (a pure state)
- The averaged state is $0.75|↑_x\rangle\langle↑_x| + 0.25|↓_x\rangle\langle↓_x|$ (a mixed state, diagonal in the $x$ basis)
- $\mathbb{E}_{\rho}[\sigma_x] = 0.5$

**A coincidental symmetry**: this particular $\rho$ gives the same probability distributions and expectation values along both $x$ and $z$.
:::



::::{exercise} Projective Measurement of a Mixed State: Conditional State, Averaged State, and Expectation Value
:label: exer-ch10-measurement

Consider the spin-1/2 mixed state $\boldsymbol{\rho} = \dfrac{1}{3}\begin{pmatrix} 2 & 1 \\ 1 & 1 \end{pmatrix}$, and perform a projective measurement of $\sigma_z$ in the $z$ basis $\{|↑_z\rangle, |↓_z\rangle\}$.

1. Verify that $\boldsymbol{\rho}$ is a valid density matrix and compute its purity.
2. Find the probabilities $P(↑) = \text{Tr}(\boldsymbol{\Pi}_↑\boldsymbol{\rho})$ and $P(↓)$ of obtaining "up" and "down," and the conditional state **after each outcome is read out**.
3. Write the averaged state $\boldsymbol{\rho}_{\text{avg}}$ **when the outcome is not read out** and compute its purity; explain what the act of measurement does to the off-diagonal entries.
4. Compute the expectation value $\mathbb{E}_{\boldsymbol{\rho}}[\sigma_z] = \text{Tr}(\boldsymbol{\rho}\sigma_z)$.

#### Summary of the Effects of Measurement on States

```{admonition} A Unified Picture of Measurement Collapse
:class: tip

| Before the measurement | Measured, outcome read out | Measured, outcome not read out | Change in purity (conditional state / averaged state) |
|--------|----------------|----------------|--------------------------|
| Mixed state | **Pure state** (conditional state) | Mixed state (averaged state, no coherence) | $\text{Tr}(\rho^2)$ rises to 1 / may stay the same |
| Pure state (eigenstate) | The same pure state | The same pure state | Unchanged (= 1) / unchanged (= 1) |
| Pure state (superposition) | **A different pure state** (conditional state) | Mixed state (averaged state) | Unchanged (= 1) / becomes $< 1$ |

**Unifying principles**:
1. **Reading out the outcome**: the conditional state is always pure, which removes the classical uncertainty
2. **Not reading out the outcome**: the averaged state may be mixed, but the off-diagonal entries are removed
3. **The coherence of the system is erased**: the off-diagonal entries of both the conditional state and the averaged state in the measurement basis are 0; if the apparatus and the environment are included in an overall unitary description, the information in the correlations can be retained in the overall state (see the note below) and does not disappear permanently
4. **The measurement basis determines the direction of collapse**: noncommuting operators give different collapse outcomes
```

```{note}
**The dynamical competition between coherence and decoherence**

In a given basis, a suitable Hamiltonian and initial state can change the coherence; not every unitary evolution produces a superposition. Certain system–environment couplings establish distinguishable environment records, and after the partial trace over the environment is taken, the coherence of the reduced state of the system is suppressed as a result. The coupling itself does not guarantee that the environment states are orthogonal, nor does it guarantee a monotone or permanent decay.

The Lindblad equation and exponential decay are approximate models of open systems that adopt suitable assumptions, such as the Markov assumption. Overall unitary evolution can retain the information in the correlations; local decoherence and the disappearance of information from the whole should be understood separately.
```


:::{admonition} §10.4 Summary
:class: tip
The central question of this section is: what exactly does a quantum measurement do to a system, and how can the change of state before and after it be described precisely? Projective measurement gives the probability of each outcome through the Born rule $P(a_j) = \text{Tr}(\boldsymbol{\Pi}_j\boldsymbol{\rho})$, and the act of measurement itself has two levels: when the measurement is made but the outcome is not read out, the state evolves into the averaged state, whose off-diagonal entries vanish (decoherence); after the outcome is read out, the state collapses to the conditional pure state of the corresponding eigenstate (the state becomes pure). This distinction between "act" and "read-out" is a key concept of quantum information theory, and it marks the fundamental difference between quantum uncertainty and classical uncertainty.

The Pauli matrices $\sigma_x, \sigma_y, \sigma_z$, as the three observables of the spin-1/2 system, provide a concrete model that can be computed completely, and their noncommutation relations $[\sigma_j, \sigma_k] = 2i\epsilon_{jkl}\sigma_l$ are not merely algebraic identities but the direct reason why spin measurements along different directions interfere with one another. From the point of view of linear algebra, the spectral decomposition is used to construct the measurement projections; a measurement channel that ignores the outcome erases the coherence between branches, which is a change of state, unlike the unitary changes of coordinates in Chapters 8 and 9, which keep the same operator. The consequences of this noncommutativity are raised in §10.5 to the rigorous inequalities of the uncertainty principle, revealing the deepest mathematical root of quantum mechanics.

## 10.5 The Uncertainty Principle

The uncertainty principle is one of the fundamental principles of quantum mechanics. It describes the limits on the intrinsic standard deviations of the distributions of measurement outcomes on identically prepared states; it is not a relation about instrument error or measurement disturbance. This is not a problem of measurement technique; it stems from the noncommutativity of operators—a central feature of quantum mechanics.

#### Basic Definitions: The Commutator and the Anticommutator

:::{prf:definition} Commutator and Anticommutator
:label: def-commutator
The **commutator and anticommutator** are important notions for describing the noncommutativity between two operators $\mathbf{A}$ and $\mathbf{B}$:

1. **Commutator**: $[\mathbf{A},\mathbf{B}] := \mathbf{A}\mathbf{B} - \mathbf{B}\mathbf{A}$
   - When $[\mathbf{A},\mathbf{B}] = \mathbf{0}$, we say that $\mathbf{A}$ and $\mathbf{B}$ **commute**

2. **Anticommutator**: $\{\mathbf{A},\mathbf{B}\} := \mathbf{A}\mathbf{B} + \mathbf{B}\mathbf{A}$
   - When $\{\mathbf{A},\mathbf{B}\} = \mathbf{0}$, we say that $\mathbf{A}$ and $\mathbf{B}$ **anticommute**
:::
```{note}
**Physical meaning of the commutator.**
- $[\mathbf{A},\mathbf{B}] = \mathbf{0}$ ⇔ the two observables **can be measured simultaneously with arbitrary precision** (they have a common complete eigenbasis)
- $[\mathbf{A},\mathbf{B}] \neq \mathbf{0}$ ⇔ the two observables are **incompatible**, and an uncertainty relation holds
- The size of the commutator quantifies the "degree of incompatibility" of the two operators

The rigorous mathematical basis for "can be measured simultaneously with arbitrary precision" is given by {prf:ref}`thm-commute-simultaneous-diag` below: two Hermitian matrices commute if and only if they can be diagonalized simultaneously—that is, there is a common orthonormal eigenbasis on whose quantum states both physical quantities have definite values at the same time. Conversely, if they do not commute, no such complete basis exists.
```


:::{prf:theorem} ◆ Commuting Matrices and Simultaneous Diagonalization
:label: thm-commute-simultaneous-diag

Let $\mathbf{A}, \mathbf{B} \in M_n(\mathbb{C})$ both be Hermitian matrices. Then $[\mathbf{A},\mathbf{B}] = \mathbf{0}$ if and only if there is a unitary matrix $\mathbf{U}$ such that $\mathbf{U}^*\mathbf{A}\mathbf{U}$ and $\mathbf{U}^*\mathbf{B}\mathbf{U}$ are both diagonal matrices—that is, $\mathbf{A}$ and $\mathbf{B}$ have a **common orthonormal eigenbasis**.
:::

:::{prf:proof}
:label: proof-commute-simultaneous-diag

1. (Sufficiency) If $\mathbf{U}^*\mathbf{A}\mathbf{U} = \boldsymbol{\Lambda}_A$ and $\mathbf{U}^*\mathbf{B}\mathbf{U} = \boldsymbol{\Lambda}_B$ are both diagonal matrices, then since diagonal matrices commute with one another, $\mathbf{A}\mathbf{B} = \mathbf{U}\boldsymbol{\Lambda}_A\boldsymbol{\Lambda}_B\mathbf{U}^* = \mathbf{U}\boldsymbol{\Lambda}_B\boldsymbol{\Lambda}_A\mathbf{U}^* = \mathbf{B}\mathbf{A}$.
2. (Necessity) Suppose $[\mathbf{A},\mathbf{B}] = \mathbf{0}$. By the spectral theorem (§9.3), the eigenspaces $E_{\lambda_0}, \ldots, E_{\lambda_{k-1}}$ corresponding to the distinct eigenvalues $\lambda_0, \ldots, \lambda_{k-1}$ of $\mathbf{A}$ are pairwise orthogonal, and their dimensions add up to $n$.
3. $\mathbf{B}$ leaves each eigenspace invariant: if $\mathbf{v} \in E_{\lambda_i}$, then $\mathbf{A}(\mathbf{B}\mathbf{v}) = \mathbf{B}(\mathbf{A}\mathbf{v}) = \lambda_i(\mathbf{B}\mathbf{v})$, so $\mathbf{B}\mathbf{v} \in E_{\lambda_i}$.
4. The restriction of $\mathbf{B}$ to each $E_{\lambda_i}$ is still Hermitian (for any $\mathbf{u},\mathbf{v} \in E_{\lambda_i}$, $\langle\mathbf{u}|\mathbf{B}\mathbf{v}\rangle = \langle\mathbf{B}\mathbf{u}|\mathbf{v}\rangle$ still holds), so by the spectral theorem again, each $E_{\lambda_i}$ has an orthonormal basis consisting of eigenvectors of $\mathbf{B}$.
5. These basis vectors are at the same time eigenvectors of $\mathbf{A}$ (with eigenvalue $\lambda_i$). Taking the union of the bases of the $k$ subspaces gives an orthonormal basis of $\mathbb{C}^n$ in which every vector is an eigenvector of both $\mathbf{A}$ and $\mathbf{B}$; the unitary matrix $\mathbf{U}$ whose columns are these vectors is the required matrix. $\square$
:::

#### Algebraic Properties of the Commutator

The expectation value of the commutator, $\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]$, has several important properties in quantum mechanics, and these properties are essential for understanding the uncertainty principle and quantum dynamics:

:::{prf:property} Basic Properties of the Commutator
:label: pro-commutator-expectation

1. **Anti-Hermiticity**: for Hermitian operators $\mathbf{A}$ and $\mathbf{B}$, the commutator $[\mathbf{A},\mathbf{B}]$ is anti-Hermitian, that is:
   $$[\mathbf{A},\mathbf{B}]^* = -[\mathbf{A},\mathbf{B}]$$
   
   Hence its expectation value is necessarily purely imaginary:
   $$\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]^* = -\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]$$

2. **Decomposition of operator products**: the expectation value of the product of any two operators can be split into a commutator part and an anticommutator part:
   $$\mathbb{E}_{\psi}[\mathbf{A}\mathbf{B}] = \frac{1}{2}\mathbb{E}_{\psi}[\{\mathbf{A},\mathbf{B}\}] + \frac{1}{2}\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]$$
   
   Since the anticommutator $\{\mathbf{A},\mathbf{B}\}$ is Hermitian, its expectation value is real, while the expectation value of the commutator is purely imaginary. This decomposition splits the operator product into a real part and an imaginary part.

3. **Behavior in eigenstates**: if $|\psi\rangle$ is an eigenstate of $\mathbf{A}$, that is, $\mathbf{A}|\psi\rangle = a|\psi\rangle$ ($a$ real), then:
   $$\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]] = 0$$
   
   This means that in an eigenstate of $\mathbf{A}$, $\mathbf{A}$ "effectively commutes" with every operator (from the point of view of expectation values).
:::

#### The Robertson-Schrödinger Inequality

The following theorem gives a lower bound for the product of the uncertainties of two observables; it is the mathematical statement of the uncertainty principle:

:::{prf:theorem} Robertson-Schrödinger Inequality
:label: thm-robertson-schrodinger-inequality
For any two Hermitian operators $\mathbf{A}$ and $\mathbf{B}$ and any quantum state $|\psi\rangle$,

$$
(\Delta\mathbf{A})^2 \cdot (\Delta\mathbf{B})^2 \geq \frac{1}{4}\left|\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]\right|^2 + \frac{1}{4}\left|\mathbb{E}_{\psi}[\{\delta\mathbf{A},\delta\mathbf{B}\}]\right|^2
$$

where:
- $\Delta\mathbf{A} := \sqrt{\text{Var}_\psi(\mathbf{A})} = \sqrt{\mathbb{E}_{\psi}[\mathbf{A}^2] - \mathbb{E}_{\psi}[\mathbf{A}]^2}$ is the standard deviation
- $\delta\mathbf{A} := \mathbf{A} - \mathbb{E}_{\psi}[\mathbf{A}]\mathbf{I}$ is the deviation operator
- The second term is $\geq 0$, which makes this inequality **stronger** than the classical Robertson inequality (which contains only the first term)
:::

:::{prf:proof}
:label: prf-robertson-schrodinger-inequality
1. **Define the deviation operators**

Define the deviation of each operator from its expectation value:
$$\delta\mathbf{A} := \mathbf{A} - \mathbb{E}_{\psi}[\mathbf{A}] \mathbf{I}, \quad \delta\mathbf{B} := \mathbf{B} - \mathbb{E}_{\psi}[\mathbf{B}] \mathbf{I}$$

Clearly $\mathbb{E}_{\psi}[\delta\mathbf{A}] = 0$ and $\mathbb{E}_{\psi}[(\delta\mathbf{A})^2] = (\Delta\mathbf{A})^2$.

2. **Construct auxiliary state vectors**

Define the state vectors:
$$|\phi_A\rangle := \delta\mathbf{A}|\psi\rangle, \quad |\phi_B\rangle := \delta\mathbf{B}|\psi\rangle$$

3. **Apply the Cauchy-Schwarz inequality**

For any two vectors in an inner product space, the Cauchy-Schwarz inequality gives:
$$|\langle\phi_A|\phi_B\rangle|^2 \leq \langle\phi_A|\phi_A\rangle \cdot \langle\phi_B|\phi_B\rangle$$

Expanding the left-hand side:
$$|\langle\phi_A|\phi_B\rangle|^2 = |\langle\psi|\delta\mathbf{A}^*\delta\mathbf{B}|\psi\rangle|^2 = |\mathbb{E}_{\psi}[\delta\mathbf{A}\delta\mathbf{B}]|^2$$

(Note that $\delta\mathbf{A}$ is Hermitian, so $\delta\mathbf{A}^* = \delta\mathbf{A}$.)

Expanding the right-hand side:
$$\langle\phi_A|\phi_A\rangle \cdot \langle\phi_B|\phi_B\rangle = \mathbb{E}_{\psi}[(\delta\mathbf{A})^2] \cdot \mathbb{E}_{\psi}[(\delta\mathbf{B})^2] = (\Delta\mathbf{A})^2 \cdot (\Delta\mathbf{B})^2$$

Therefore:
$$|\mathbb{E}_{\psi}[\delta\mathbf{A}\delta\mathbf{B}]|^2 \leq (\Delta\mathbf{A})^2 \cdot (\Delta\mathbf{B})^2$$

4. **Decompose the operator product**

Note that $\delta\mathbf{A}\delta\mathbf{B}$ can be split into a commutator part and an anticommutator part:
$$\delta\mathbf{A}\delta\mathbf{B} = \frac{1}{2}[\delta\mathbf{A},\delta\mathbf{B}] + \frac{1}{2}\{\delta\mathbf{A},\delta\mathbf{B}\}$$

**Key observation**: the commutator is unaffected by constant shifts:
$$[\delta\mathbf{A},\delta\mathbf{B}] = [\mathbf{A} - c_A\mathbf{I}, \mathbf{B} - c_B\mathbf{I}] = [\mathbf{A},\mathbf{B}]$$

5. **Use Hermiticity**

For the expectation values:
- The commutator $[\mathbf{A},\mathbf{B}]$ is anti-Hermitian ⇒ $\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]$ is purely imaginary
- The anticommutator $\{\delta\mathbf{A},\delta\mathbf{B}\}$ is Hermitian ⇒ $\mathbb{E}_{\psi}[\{\delta\mathbf{A},\delta\mathbf{B}\}]$ is real

These two terms are orthogonal to each other (one lies on the real axis and the other on the imaginary axis), so:
$$\left|\mathbb{E}_{\psi}[\delta\mathbf{A}\delta\mathbf{B}]\right|^2 = \frac{1}{4}\left|\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]\right|^2 + \frac{1}{4}\left|\mathbb{E}_{\psi}[\{\delta\mathbf{A},\delta\mathbf{B}\}]\right|^2$$

Combining this with the result of step 3 gives:
$$(\Delta\mathbf{A})^2 \cdot (\Delta\mathbf{B})^2 \geq \frac{1}{4}\left|\mathbb{E}_{\psi}[[\mathbf{A},\mathbf{B}]]\right|^2 + \frac{1}{4}\left|\mathbb{E}_{\psi}[\{\delta\mathbf{A},\delta\mathbf{B}\}]\right|^2$$
:::

```{important}
:label: imp-convexity
**The uncertainty relation for mixed states.**

The theorem above holds for pure states $|\psi\rangle$. For a mixed state $\rho = \sum_i p_i |\psi_i\rangle\langle\psi_i|$, the uncertainty relation can be proved by **convexity**:

If every pure-state component $|\psi_i\rangle$ satisfies $\Delta_i \mathbf{A} \cdot \Delta_i \mathbf{B} \geq c_i$, then for the mixed state we have at least:
$$\Delta\mathbf{A} \cdot \Delta\mathbf{B} \geq \min_i c_i$$

In fact, Jensen's inequality (together with the Cauchy-Schwarz inequality) gives a stronger result: mixing increases uncertainty, $(\Delta\mathbf{A})^2 \geq \sum_i p_i (\Delta_i\mathbf{A})^2$ (and likewise for $\mathbf{B}$), so $\Delta\mathbf{A} \cdot \Delta\mathbf{B} \geq \sum_i p_i\, \Delta_i\mathbf{A} \cdot \Delta_i\mathbf{B} \geq \sum_i p_i c_i$; that is, the uncertainty product of a mixed state is no smaller than the weighted average of the lower bounds of its pure-state components, and this weighted average is itself no smaller than $\min_i c_i$.
```

#### The Spin Uncertainty Relation (a Concrete Example)

Before discussing the abstract position-momentum uncertainty relation, we first look at a concrete, computable example: **the uncertainty relation for a spin-1/2 system**. This example is based on the Pauli matrices introduced in §10.4 and shows the core mechanism of the uncertainty principle.

:::{prf:example} The Uncertainty Relation for the Pauli Matrices
:label: ex-pauli-uncertainty

Consider the uncertainty relation between $\sigma_x$ and $\sigma_y$. Let the quantum state be $|\psi\rangle \in \mathbb{C}^2$ with $\|\psi\| = 1$.

**Step 1. Compute the expectation value of the commutator**

From §10.4 we know that $[\sigma_x, \sigma_y] = 2i\sigma_z$, so:

$$
\mathbb{E}_{\psi}[[\sigma_x, \sigma_y]] = \langle\psi|[\sigma_x, \sigma_y]|\psi\rangle = \langle\psi|2i\sigma_z|\psi\rangle = 2i\langle\psi|\sigma_z|\psi\rangle
$$

**Step 2. Apply the Robertson-Schrödinger inequality**

Substituting the expectation value of the commutator into {prf:ref}`thm-robertson-schrodinger-inequality`:

$$
\Delta\sigma_x \cdot \Delta\sigma_y \geq \frac{1}{2}\left|\mathbb{E}_{\psi}[[\sigma_x,\sigma_y]]\right| = \frac{1}{2}|2i\langle\psi|\sigma_z|\psi\rangle| = |\langle\psi|\sigma_z|\psi\rangle|
$$

This gives the uncertainty relation for spin measurements:
$$
\underline{\Delta\sigma_x \cdot \Delta\sigma_y \geq |\langle\sigma_z\rangle_\psi|}
$$

Similarly, cyclic symmetry gives:
$$
\Delta\sigma_y \cdot \Delta\sigma_z \geq |\langle\sigma_x\rangle_\psi|, \quad \Delta\sigma_z \cdot \Delta\sigma_x \geq |\langle\sigma_y\rangle_\psi|
$$

**Step 3. Concrete examples**

**(a) An eigenstate of $\sigma_x$**: $|\psi\rangle = |↑_x\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$ 

This is the spin-up state along $x$ (§10.4):

Compute the expectation values:
$$\langle\sigma_x\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{2}(2) = 1$$

$$\langle\sigma_x^2\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\mathbf{I}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = 1$$

Hence $\Delta\sigma_x = \sqrt{1 - 1^2} = 0$ (**completely certain**).

Compute $\sigma_y$:
$$\langle\sigma_y\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{2}(i - i) = 0$$

$$\langle\sigma_y^2\rangle = 1$$

Hence $\Delta\sigma_y = 1$ (**completely uncertain**).

Compute $\sigma_z$:
$$\langle\sigma_z\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{2}(1 - 1) = 0$$

Check the inequality:
$$\Delta\sigma_x \cdot \Delta\sigma_y = 0 \cdot 1 = 0 \geq |\langle\sigma_z\rangle| = 0$$ ✓

**Physical interpretation**: when we are completely certain that the particle's spin is up along $x$ ($\Delta\sigma_x = 0$), we know nothing about its spin along $y$ ($\Delta\sigma_y = 1$). The inequality holds with equality because the right-hand side is zero.

---

**(b) An eigenstate of $\sigma_z$**: $|\psi\rangle = |↑_z\rangle = |0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$ 

This is the spin-up state along $z$:

Compute $\sigma_x$:
$$\langle\sigma_x\rangle = \begin{pmatrix} 1 & 0 \end{pmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = 0$$

$$\langle\sigma_x^2\rangle = 1 \Rightarrow \Delta\sigma_x = 1$$

Compute $\sigma_y$:
$$\langle\sigma_y\rangle = 0, \quad \langle\sigma_y^2\rangle = 1 \Rightarrow \Delta\sigma_y = 1$$

Compute $\sigma_z$:
$$\langle\sigma_z\rangle = \begin{pmatrix} 1 & 0 \end{pmatrix}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = 1$$

Check the inequality:
$$\Delta\sigma_x \cdot \Delta\sigma_y = 1 \cdot 1 = 1 \geq |\langle\sigma_z\rangle| = 1$$ ✓

Equality holds again.

**Physical interpretation**: when the particle's spin along $z$ is certain, the spins along $x$ and $y$ both become completely uncertain, and the product of their uncertainties attains the lower bound $|\langle\sigma_z\rangle| = 1$.

---

**(c) A tilted state** (a nontrivial general case): $|\psi\rangle = \cos\left(\frac{\pi}{8}\right)|0\rangle + \sin\left(\frac{\pi}{8}\right)|1\rangle$ 

This state is tilted by $45°$ on the Bloch sphere ($\theta = \pi/4$, so the coefficients use $\theta/2 = \pi/8$); it **lies neither on the $z$ axis nor in the $xy$ plane**, and it shows perfectly how the uncertainty principle constrains a general case.

**Why choose this state?**
- The earlier eigenstate examples showed the boundary cases ($0$ or $1$), but they were a bit like computing "$0 = 0$" or "$1 = 1$"
- This tilted state makes $\Delta\sigma_x$ and the lower bound take the intermediate value $1/\sqrt2$, while $\Delta\sigma_y$ still attains its maximum value 1
- It shows how uncertainty "slides" and is "distributed" in a general state

**Compute along $z$ (which determines the lower bound)**:

Using the trigonometric identity $\cos^2\alpha - \sin^2\alpha = \cos(2\alpha)$:

$$\langle\sigma_z\rangle = \cos^2\left(\frac{\pi}{8}\right) - \sin^2\left(\frac{\pi}{8}\right) = \cos\left(\frac{\pi}{4}\right) = \frac{1}{\sqrt{2}} \approx 0.707$$

This means that **the lower bound on the right-hand side of the inequality is no longer $0$ or $1$ but an intermediate value** $\frac{1}{\sqrt{2}}$.

**Compute along $x$**:

$$\langle\sigma_x\rangle = \begin{pmatrix} \cos\frac{\pi}{8} & \sin\frac{\pi}{8} \end{pmatrix}\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}\begin{pmatrix} \cos\frac{\pi}{8} \\ \sin\frac{\pi}{8} \end{pmatrix}$$
$$= 2\cos\left(\frac{\pi}{8}\right)\sin\left(\frac{\pi}{8}\right) = \sin\left(\frac{\pi}{4}\right) = \frac{1}{\sqrt{2}}$$

$$\langle\sigma_x^2\rangle = 1 \quad \Rightarrow \quad \Delta\sigma_x = \sqrt{1 - \frac{1}{2}} = \frac{1}{\sqrt{2}} \approx 0.707$$

**Compute along $y$**:

$$\langle\sigma_y\rangle = \begin{pmatrix} \cos\frac{\pi}{8} & \sin\frac{\pi}{8} \end{pmatrix}\begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix}\begin{pmatrix} \cos\frac{\pi}{8} \\ \sin\frac{\pi}{8} \end{pmatrix}$$
$$= i\sin\left(\frac{\pi}{8}\right)\cos\left(\frac{\pi}{8}\right) - i\cos\left(\frac{\pi}{8}\right)\sin\left(\frac{\pi}{8}\right) = 0$$

(Because the coefficients are all real, the expectation value along $y$ is $0$.)

$$\langle\sigma_y^2\rangle = 1 \quad \Rightarrow \quad \Delta\sigma_y = 1$$

**Check the inequality**:

Left-hand side (product of the measurement fluctuations):
$$\Delta\sigma_x \cdot \Delta\sigma_y = \frac{1}{\sqrt{2}} \cdot 1 = \frac{1}{\sqrt{2}} \approx 0.707$$

Right-hand side (theoretical lower bound):
$$|\langle\sigma_z\rangle| = \frac{1}{\sqrt{2}} \approx 0.707$$

Conclusion:
$$\frac{1}{\sqrt{2}} \geq \frac{1}{\sqrt{2}} \quad \text{(equality holds)}$$ ✓

**A closer look: why is this example more representative?**

1. **A nonzero lower bound**: because the particle leans toward the $z$ axis ($\langle\sigma_z\rangle \neq 0$), the measurement outcomes for $x$ and $y$ are forced to maintain a certain degree of "mutual exclusivity." In this state the $x$ component is partly uncertain, while the $y$ component is still maximally uncertain, with equal probabilities.

2. **The mechanism by which uncertainty is distributed**:
   - Because the state lies in the $xz$ plane, we have **partial knowledge** of $x$ ($\Delta\sigma_x \approx 0.707$), unlike the complete ignorance in the $\sigma_z$ eigenstate ($\Delta\sigma_x = 1$)
   - As the price, since $\sigma_y$ is completely orthogonal to this plane, our observation of $y$ remains **maximally uncertain** ($\Delta\sigma_y = 1$)
   - This shows the "distribution of uncertainty" in quantum mechanics: gaining partial information in one direction must be paid for with uncertainty in other directions

3. **Minimum-uncertainty states** (intelligent states): this particular tilted state still makes the inequality hold with equality, which shows that it is the "optimal solution" under this lower-bound constraint—for the given $\langle\sigma_z\rangle = \frac{1}{\sqrt{2}}$, this state attains the theoretical minimum of $\Delta\sigma_x \cdot \Delta\sigma_y$.

**Comparing the three examples**:

| Example | $\langle\sigma_z\rangle$ | $\Delta\sigma_x$ | $\Delta\sigma_y$ | $\Delta\sigma_x \cdot \Delta\sigma_y$ | Lower bound | Feature |
|------|-------------------------|------------------|------------------|--------------------------------------|------|------|
| (a) $\sigma_x$ eigenstate | $0$ | $0$ | $1$ | $0$ | $0$ | Boundary case: completely certain vs. completely uncertain |
| (b) $\sigma_z$ eigenstate | $1$ | $1$ | $1$ | $1$ | $1$ | Boundary case: maximal uncertainty attained |
| (c) Tilted state | $\frac{1}{\sqrt{2}}$ | $\frac{1}{\sqrt{2}}$ | $1$ | $\frac{1}{\sqrt{2}}$ | $\frac{1}{\sqrt{2}}$ | **General case**: shows the dynamic constraint |

Example (c) fills the gap between $0$ and $1$ and shows how the uncertainty principle constrains uncertainty in **non-extreme states**.

---

**Key conclusions**:

1. **The commutator determines the lower bound**: $[\sigma_x, \sigma_y] = 2i\sigma_z$ ⇒ the lower bound on the uncertainty product is $|\langle\sigma_z\rangle|$
   - The lower bound **depends on the state**: different $|\psi\rangle$ have different $\langle\sigma_z\rangle$

2. **The three examples give the complete picture**:
   - **Example (a)**: a $\sigma_x$ eigenstate → the lower bound is $0$ (one direction completely certain)
   - **Example (c)**: a tilted state → the lower bound is $\frac{1}{\sqrt{2}}$ (**the dynamic constraint of the general case**)
   - **Example (b)**: a $\sigma_z$ eigenstate → the lower bound is $1$ (maximal uncertainty attained)
   
   Going from $0 \to \frac{1}{\sqrt{2}} \to 1$, these three examples show completely how uncertainty "slides" as the state changes

3. **The special place of the tilted state**:
   - It is neither "all" nor "nothing," and it is closer to realistic physical situations
   - It shows the **mechanism by which uncertainty is distributed**: gaining partial information in one direction must be paid for with uncertainty in other directions
   - It is a **minimum-uncertainty state** (equality holds), optimal under this constraint

4. **Computability in finite dimensions**: the Pauli matrices are $2 \times 2$ matrices, all computations are elementary matrix operations, and students can verify every step directly

5. **Physical intuition**: spin components along different directions cannot be definite at the same time; here the eigenstates of spin correspond to the standard axes of the Bloch sphere, and clearly no point can be close to two of these axes at once. The tilted state lets us see how this "incompatibility" shows up in non-extreme situations.

:::

This example shows the core mechanism of the uncertainty principle: **through the Robertson-Schrödinger inequality, the noncommutativity of operators leads directly to lower bounds on the uncertainties of physical quantities**.

Next we generalize this idea to the infinite-dimensional position-momentum system, where we will see a similar structure, except that the lower bound becomes the universal constant $\frac{ℏ}{2}$.

#### The Heisenberg Uncertainty Relation

The most famous example of the uncertainty principle is the **position-momentum uncertainty relation**. Having shown the finite-dimensional example of the Pauli matrices, we now turn to continuous systems, which require infinite-dimensional spaces.

:::{prf:example} The Position-Momentum Uncertainty Relation
:label: ex-position-momentum-uncertainty

In quantum mechanics, the uncertainty relation between position and momentum can be described in two equivalent mathematical languages: **wave mechanics** (1926) and **matrix mechanics** (1925).

---

#### The Matrix Mechanics Formulation

In an abstract Hilbert space $\mathcal{H}$, a quantum state is described by a state vector $|\psi\rangle$.

1. **The position operator** $\hat{X}$: a self-adjoint operator (Hermitian operator) acting on $\mathcal{H}$, satisfying $\hat{X}^* = \hat{X}$.

2. **The momentum operator** $\hat{P}$: a self-adjoint operator acting on $\mathcal{H}$, satisfying $\hat{P}^* = \hat{P}$.

3. **The canonical commutation relation**: these two operators satisfy the fundamental commutation relation
   $$[\hat{X}, \hat{P}] = iℏ \mathbf{I}$$
   where $ℏ = \frac{h}{2\pi} \approx 1.055 \times 10^{-34}$ J·s is the **reduced Planck constant**.

**The infinite-dimensional representation of matrix mechanics, restated in modern language**:

In the language of modern quantum mechanics, the harmonic oscillator basis $\{|n\rangle\}_{n=0}^{\infty}$ is used to represent position and momentum as **infinite-dimensional matrices**. In a certain orthonormal basis, $\hat{X}$ and $\hat{P}$ can be represented as:

$$\hat{X} = \sqrt{\frac{ℏ}{2M\omega}} \begin{pmatrix} 0 & \sqrt{1} & 0 & 0 & \dots \\ \sqrt{1} & 0 & \sqrt{2} & 0 & \dots \\ 0 & \sqrt{2} & 0 & \sqrt{3} & \dots \\ 0 & 0 & \sqrt{3} & 0 & \dots \\ \vdots & \vdots & \vdots & \vdots & \ddots \end{pmatrix}$$

$$\hat{P} = i\sqrt{\frac{M\omegaℏ}{2}} \begin{pmatrix} 0 & -\sqrt{1} & 0 & 0 & \dots \\ \sqrt{1} & 0 & -\sqrt{2} & 0 & \dots \\ 0 & \sqrt{2} & 0 & -\sqrt{3} & \dots \\ 0 & 0 & \sqrt{3} & 0 & \dots \\ \vdots & \vdots & \vdots & \vdots & \ddots \end{pmatrix}$$

where $M>0$ is the mass of the particle and $\omega$ is the angular frequency of the harmonic oscillator. The matrix entries are:
$$X_{mn} = \sqrt{\frac{ℏ}{2M\omega}}(\sqrt{n}\delta_{m,n-1} + \sqrt{n+1}\delta_{m,n+1})$$
$$P_{mn} = i\sqrt{\frac{M\omegaℏ}{2}}(-\sqrt{n}\delta_{m,n-1} + \sqrt{n+1}\delta_{m,n+1})$$

These two infinite-dimensional matrices satisfy the canonical commutation relation $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$.

**Key observations**:
- The matrices are **tridiagonal** (nonzero entries appear only on the diagonal and next to it)
- The matrices are **unbounded** (the entries grow with $n$: $X_{n,n+1} \propto \sqrt{n+1}$)
- The matrices **must be infinite-dimensional** (we prove below that the canonical commutation relation cannot be realized in finite dimensions)

In matrix form the commutation relation reads:
$$XP - PX = iℏ I_\infty$$

where $I_\infty$ is the infinite-dimensional identity matrix.

---

#### The Wave Mechanics Formulation

In the **position representation**, the Hilbert space is realized as $\mathcal{H} = L^2(\mathbb{R})$, and a quantum state is described by the wave function $\psi(x) = \langle x|\psi\rangle$.

1. **The position operator** $\hat{X}$ (in the position representation):
   $$(\hat{X}\psi)(x) = x\psi(x)$$
   That is, the position operator multiplies the wave function by the position coordinate $x$.

2. **The momentum operator** $\hat{P}$ (in the position representation):
   $$(\hat{P}\psi)(x) = -iℏ \frac{d\psi}{dx}(x)$$
   The momentum operator corresponds to differentiation.

**Verifying the commutation relation** (position representation):

Below we take a normalized Schwartz function $\psi\in\mathcal{S}(\mathbb{R})$, so that differentiation, multiplication, Fourier integrals, and second moments all make sense; the CCR holds on this common dense domain, and the substitution into the uncertainty relation that follows stays within this scope. Compute the commutator:

$$
\begin{align*}
([\hat{X},\hat{P}]\psi)(x) &= (\hat{X}\hat{P} - \hat{P}\hat{X})\psi(x) \\
&= \hat{X}\left(-iℏ \frac{d\psi}{dx}\right) - \hat{P}(x\psi) \\
&= -iℏ x\frac{d\psi}{dx} - \left(-iℏ \frac{d(x\psi)}{dx}\right) \\
&= -iℏ x\frac{d\psi}{dx} + iℏ\left(\psi + x\frac{d\psi}{dx}\right) \\
&= iℏ \psi(x)
\end{align*}
$$

This confirms that $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$.

---

#### The Connection Between the Two Formulations

The position representation and the momentum representation are related by the **Fourier transform**. Let $\tilde{\psi}(p) = \langle p|\psi\rangle$ be the wave function in the momentum representation; then:
$$\tilde{\psi}(p) = \frac{1}{\sqrt{2\piℏ}}\int_{-\infty}^{\infty} e^{-ipx/ℏ}\psi(x)dx$$

In the momentum representation:
- The momentum operator: $(\hat{P}\tilde{\psi})(p) = p\tilde{\psi}(p)$
- The position operator: $(\hat{X}\tilde{\psi})(p) = iℏ \frac{d\tilde{\psi}}{dp}(p)$

The two representations give the same commutation relation. The infinite-dimensional matrix representation written above in modern language is the representation in the basis of energy eigenstates of the harmonic oscillator, and it is related to the position and momentum representations by a change of basis.

---

#### Deriving the Uncertainty Relation

Substituting into the Robertson-Schrödinger inequality ({prf:ref}`thm-robertson-schrodinger-inequality`), note the expectation value of the commutator:
$$\mathbb{E}_{\psi}[[\hat{X},\hat{P}]] = \langle \psi | iℏ \mathbf{I} | \psi \rangle = iℏ$$

Therefore:
$$\Delta X \cdot \Delta P \geq \frac{1}{2}\left|\mathbb{E}_{\psi}[[\hat{X},\hat{P}]]\right| = \frac{1}{2}|iℏ| = \frac{ℏ}{2}$$

This is the famous **Heisenberg uncertainty principle**:
$$\underline{\Delta X \cdot \Delta P \geq \frac{ℏ}{2}}$$

**Comparison with the Pauli matrix example**:
- Pauli matrices: $[\sigma_x, \sigma_y] = 2i\sigma_z$ → lower bound $|\langle\sigma_z\rangle|$ (depends on the state)
- Position-momentum: $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$ → lower bound $\frac{ℏ}{2}$ (a universal constant, independent of the state)

The key difference: a commutator equal to an operator vs. a commutator equal to a constant times the identity operator!

---

#### Physical Meaning

1. **A fundamental limit, not a technical problem**: this is not a limit on the precision of measuring instruments but a fundamental feature of quantum mechanics. $[\hat{X},\hat{P}] \neq \mathbf{0}$ means that there is no quantum state that is simultaneously an eigenstate of $\hat{X}$ and of $\hat{P}$ (the two operators have no common complete system of eigenvectors).

2. **The complementarity principle**: the more concentrated the position distribution of the prepared state ($\Delta X$ smaller), the more uncertain its momentum ($\Delta P$ larger), and vice versa. Position and momentum are **complementary observables**.

3. **Interpretation in matrix mechanics**: the noncommutativity $[\hat{X},\hat{P}] \neq \mathbf{0}$ means that the infinite-dimensional matrices $X$ and $P$ cannot be diagonalized simultaneously; this is the core of the uncertainty principle in matrix mechanics.

4. **Interpretation in wave mechanics**: in the position representation, a particle appears as a "wave packet." The narrower the wave packet (definite position), the broader its Fourier spectrum (uncertain momentum); this corresponds to the **time-frequency uncertainty principle** of signal processing.

5. **The importance of the quantum scale**: because $ℏ$ is extremely small, at macroscopic scales ($\Delta X \sim 1$ m) the bound $\Delta P \geq 5 \times 10^{-35}$ kg·m/s is negligible. At atomic scales ($\Delta X \sim 10^{-10}$ m), however, the momentum uncertainty becomes significant; for example, an electron bound in an atom necessarily has momentum fluctuations of about $\Delta P \sim 10^{-24}$ kg·m/s.

---

#### Historical Note

Matrix mechanics in 1925 and wave mechanics in 1926 provided two formulations of quantum theory. The tridiagonal matrices of the harmonic oscillator here use modern notation; they are a concrete link between the matrix and wave-function representations.
:::

---

#### Why Can the Position-Momentum Uncertainty Relation Not Be Proved in Finite Dimensions?

We have seen that the Pauli matrices (§10.4 and the spin example above) display the mathematical structure of the uncertainty principle perfectly. However, **the position-momentum uncertainty relation cannot be realized in a finite-dimensional Hilbert space**. This is the essential difference between continuous and discrete degrees of freedom in quantum mechanics.

:::{prf:theorem} The Infinite-Dimensional Nature of the Canonical Commutation Relation
:label: thm-ccr-infinite-dimensional

Let $\hat{X}$ and $\hat{P}$ be linear operators acting on a finite-dimensional Hilbert space $\mathbb{C}^n$. Then they **cannot** satisfy the canonical commutation relation:
$$
[\hat{X}, \hat{P}] = iℏ \mathbf{I}_n
$$

**Proof (trace argument)**:

Suppose there are $n \times n$ matrices $\hat{X}$ and $\hat{P}$ satisfying the relation above. Take the trace of both sides:

$$
\text{Tr}([\hat{X}, \hat{P}]) = \text{Tr}(\hat{X}\hat{P} - \hat{P}\hat{X}) = \text{Tr}(\hat{X}\hat{P}) - \text{Tr}(\hat{P}\hat{X})
$$

By the **cyclic property of the trace** $\text{Tr}(AB) = \text{Tr}(BA)$ (see Exercise {prf:ref}`exer-trace-cyclic` in the Chapter 3 exercises; it can also be verified directly by interchanging the order of summation in $\text{Tr}(AB) = \sum_i\sum_j a_{ij}b_{ji} = \text{Tr}(BA)$), we have:

$$
\text{Tr}(\hat{X}\hat{P}) = \text{Tr}(\hat{P}\hat{X})
$$

Hence the left-hand side is:
$$
\text{Tr}([\hat{X}, \hat{P}]) = 0
$$

But the right-hand side is:
$$
\text{Tr}(iℏ \mathbf{I}_n) = iℏ \sum_{j=0}^{n-1} (\mathbf{I}_n)_{jj} = iℏ \cdot n \neq 0 \quad (\text{for every } n \geq 1)
$$

This is a contradiction! Therefore the canonical commutation relation $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$ **can hold only in an infinite-dimensional Hilbert space**.

:::

**Key comparison**:

| Property | Pauli matrices (spin) | Position and momentum operators |
|------|----------------|-------------|
| **Dimension of the Hilbert space** | Finite ($\dim = 2$) | Infinite ($\dim = \infty$) |
| **Nature of the physical quantities** | Discrete spectrum ($\pm 1$) | Continuous spectrum ($\mathbb{R}$) |
| **Form of the commutator** | $[\sigma_x, \sigma_y] = 2i\sigma_z$ | $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$ |
| **Type of the commutator** | A bounded operator (diagonalizable) | On a common domain it equals $iℏ\mathbf{I}$, which extends to a bounded operator; $X,P$ themselves are unbounded |
| **Trace of the commutator** | $\text{Tr}(2i\sigma_z) = 0$ | Not trace class; the usual trace does not apply |
| **Applicable systems** | Spin-$\frac{1}{2}$ (e.g., the electron) | Particles in continuous space |
| **Matrix representation** | Explicit $2 \times 2$ matrices | Infinite-dimensional tridiagonal matrices |
| **Lower bound of uncertainty** | $|\langle\sigma_z\rangle|$ (depends on the state) | $\frac{ℏ}{2}$ (a universal constant) |

**Deeper physical reasons**:

1. **Pauli matrices**: spin is a **discrete** intrinsic degree of freedom. Whatever the direction of measurement, a spin-$\frac{1}{2}$ particle has only two possible values ($\pm\frac{ℏ}{2}$). This discreteness makes the state space naturally finite-dimensional ($\mathbb{C}^2$), and the commutator $[\sigma_x, \sigma_y] = 2i\sigma_z$ is itself a bounded $2 \times 2$ matrix with trace zero.

2. **Position and momentum operators**: the position $x$ and momentum $p$ of a particle are both **continuous** variables taking values in $(-\infty, \infty)$. Describing such a system requires the infinite-dimensional function space $L^2(\mathbb{R})$. The right-hand side of the commutator $[\hat{X}, \hat{P}] = iℏ \mathbf{I}$ is the identity operator, which in infinite dimensions is bounded but not trace class, so the usual finite trace cannot be used.

3. **Mathematical essence**: a commutator equal to a constant times the identity operator ($[\hat{A}, \hat{B}] = c\mathbf{I}$, $c \neq 0$) is a **property reserved for infinite-dimensional Hilbert spaces**; every finite-dimensional representation fails because of the trace contradiction. This is precisely a corollary of the cyclic property of the trace: the trace is invariant under cyclic permutations ({prf:ref}`exer-trace-cyclic`), so in finite dimensions (or under suitable trace-class conditions) the trace of a commutator, $\text{Tr}([\mathbf{A},\mathbf{B}]) = \text{Tr}(\mathbf{AB}) - \text{Tr}(\mathbf{BA}) = 0$, must be zero.

**Conclusion**:

The Pauli matrices give us a **finite-dimensional analogue** of the uncertainty principle and show clearly how the noncommutativity of operators leads to the uncertainty of physical quantities. This is a concrete example that students can master completely and compute by hand.

Because position and momentum are continuous variables, however, the true position-momentum uncertainty relation
$$
\Delta X \cdot \Delta P \geq \frac{ℏ}{2}
$$
must be proved in the infinite-dimensional Hilbert space $L^2(\mathbb{R})$ (or with the infinite-dimensional matrix representation of matrix mechanics) and **cannot be reduced to a finite-dimensional matrix model**. This reflects the essential difference between continuous and discrete degrees of freedom in quantum mechanics, and it is also why we need functional analysis and the theory of infinite-dimensional spaces to describe quantum mechanics completely.

```{admonition} Other Important Uncertainty Relations
:class: tip

Besides position-momentum and spin components, quantum mechanics has many other uncertainty relations:

1. **Energy-time uncertainty**: $\Delta E \cdot \Delta t \geq \frac{ℏ}{2}$
   - For example, for an observable $B$ with no explicit time dependence, define the Mandelstam–Tamm characteristic time $\Delta t=\Delta B/|d\langle B\rangle/dt|$ (with nonzero denominator); the Robertson inequality then gives this bound. This is not the lifetime of every quantum state
   - Deriving this relation requires more care, because time is not an operator in quantum mechanics

2. **Angular momentum components**: $\Delta L_x \cdot \Delta L_y \geq \frac{ℏ}{2}|\langle L_z \rangle|$
   - The three components of angular momentum do not commute with one another: $[L_i, L_j] = iℏ\epsilon_{ijk}L_k$
   - Similar to the Pauli matrices, but corresponding to spin-1 or higher-spin systems
   - Can be realized in finite-dimensional spaces (dimension = $2L + 1$)

3. **Number-phase uncertainty**: in quantum optics, photon number and phase are conjugate variables
   - A laser (definite phase) has an uncertain photon number
   - A Fock state (definite photon number) has a completely uncertain phase

All these uncertainty relations stem from the noncommutativity of the corresponding operators and embody a universal feature of quantum mechanics.
```

::::{exercise} Substitution Check of the Robertson-Schrödinger Inequality and the Trace of a Commutator
:label: exer-ch10-uncertainty

**Part 1**: consider the state $|\psi\rangle = \cos\frac{\pi}{8}|0\rangle + i\sin\frac{\pi}{8}|1\rangle$.

1. Compute $\langle\sigma_x\rangle$, $\langle\sigma_y\rangle$, and $\langle\sigma_z\rangle$.
2. Compute $\Delta\sigma_z$ and $\Delta\sigma_x$, and verify the uncertainty relation $\Delta\sigma_z \cdot \Delta\sigma_x \geq |\langle\sigma_y\rangle|$ for the pair $\sigma_z, \sigma_x$.

**Part 2**:

3. Use the cyclic property of the trace ({prf:ref}`exer-trace-cyclic`) to prove that for **any** two $n \times n$ matrices $\mathbf{A}, \mathbf{B}$ we always have $\text{Tr}([\mathbf{A},\mathbf{B}]) = 0$, and explain how this rules out the canonical commutation relation $[\hat{X},\hat{P}] = iℏ\mathbf{I}$ in finite-dimensional spaces.

:::{solution} exer-ch10-uncertainty
:class: dropdown

1. Since $|\psi\rangle = \begin{pmatrix} \cos\frac{\pi}{8} \\ i\sin\frac{\pi}{8} \end{pmatrix}$ (the coefficients contain the imaginary unit $i$):
   $$\langle\sigma_x\rangle = 2\,\text{Re}\!\left(\overline{\cos\tfrac{\pi}{8}} \cdot i\sin\tfrac{\pi}{8}\right) = 0, \quad \langle\sigma_y\rangle = 2\cos\tfrac{\pi}{8}\sin\tfrac{\pi}{8} = \sin\tfrac{\pi}{4} = \tfrac{1}{\sqrt{2}}$$
   $$\langle\sigma_z\rangle = \cos^2\tfrac{\pi}{8} - \sin^2\tfrac{\pi}{8} = \cos\tfrac{\pi}{4} = \tfrac{1}{\sqrt{2}}$$

2. $\langle\sigma_z^2\rangle = \langle\sigma_x^2\rangle = 1$, so $\Delta\sigma_z = \sqrt{1 - \frac{1}{2}} = \frac{1}{\sqrt{2}}$ and $\Delta\sigma_x = \sqrt{1 - 0} = 1$.
   $$\Delta\sigma_z \cdot \Delta\sigma_x = \tfrac{1}{\sqrt{2}} \cdot 1 = \tfrac{1}{\sqrt{2}} \geq |\langle\sigma_y\rangle| = \tfrac{1}{\sqrt{2}}$$ ✓ (equality holds; a minimum-uncertainty state).

3. From $\text{Tr}(\mathbf{AB}) = \text{Tr}(\mathbf{BA})$:
   $$\text{Tr}([\mathbf{A},\mathbf{B}]) = \text{Tr}(\mathbf{AB}) - \text{Tr}(\mathbf{BA}) = 0$$
   But $\text{Tr}(iℏ\mathbf{I}_n) = iℏ n \neq 0$. If $[\hat{X},\hat{P}] = iℏ\mathbf{I}_n$ held in finite dimensions, taking the trace of both sides would give $0 = iℏ n$, a contradiction. Hence the canonical commutation relation can hold only in an infinite-dimensional Hilbert space. $\square$

:::{admonition} §10.5 Summary
:class: tip
This section answered one of the deepest questions of quantum mechanics: why can certain pairs of physical quantities not be measured simultaneously with arbitrary precision—is this a technical limitation or a fundamental property of nature? With the Cauchy-Schwarz inequality at its core, the Robertson-Schrödinger inequality proves rigorously that for any Hermitian operators $\mathbf{A},\mathbf{B}$, the lower bound on the uncertainty product $\Delta\mathbf{A}\cdot\Delta\mathbf{B}$ is determined by the expectation value of the commutator and the covariance term in that state, and it may be zero; noncommutativity rules out a common complete eigenbasis. The Pauli matrices give a complete finite-dimensional demonstration—$\Delta\sigma_x\cdot\Delta\sigma_y\geq|\langle\sigma_z\rangle|$ can be verified directly by hand. The position-momentum relation $\Delta X\cdot\Delta P\geqℏ/2$ reveals a deeper boundary: the trace argument shows rigorously that the canonical commutation relation $[\hat{X},\hat{P}]=iℏ\mathbf{I}$ cannot hold in any finite-dimensional space, which forces us to use infinite-dimensional Hilbert spaces.

This boundary between finite and infinite dimensions marks the essential difference in mathematical structure between discrete degrees of freedom (such as spin) and continuous ones (such as motion in space). The uncertainty principle is not a regret that measuring instruments are not precise enough; it is a deep imprint that the noncommutative algebra of operators leaves on the laws of nature. The main thread of Chapter 10—from the language of probability, through quantum states and the theory of measurement, to the mathematical roots of uncertainty—is extended further in the next chapter by the Schmidt decomposition, which quantifies the even stranger phenomenon of quantum entanglement.
:::

`````python
import numpy as np

np.set_printoptions(precision=4, suppress=True, linewidth=100)

def print_header(title):
    print("=" * 60)
    print(f"  {title}")
    print("=" * 60)

# Pauli matrices
sx = np.array([[0, 1], [1, 0]], dtype=complex)
sy = np.array([[0, -1j], [1j, 0]], dtype=complex)
sz = np.array([[1, 0], [0, -1]], dtype=complex)

def expect(A, psi):
    """Expectation value ⟨ψ|A|ψ⟩ (necessarily real for a Hermitian operator)"""
    return np.real(psi.conj() @ A @ psi)

def delta(A, psi):
    """Standard deviation ΔA = sqrt(⟨A²⟩ − ⟨A⟩²)"""
    return np.sqrt(expect(A @ A, psi) - expect(A, psi) ** 2)

# ============================================================
print_header("Verifying the Robertson-Schrödinger Inequality: Δσx · Δσy ≥ |⟨σz⟩|")
# ============================================================

states = {
    "(a) σx eigenstate |↑x⟩": np.array([1, 1], dtype=complex) / np.sqrt(2),
    "(b) σz eigenstate |↑z⟩": np.array([1, 0], dtype=complex),
    "(c) tilted cos(π/8)|0⟩+sin(π/8)|1⟩":
        np.array([np.cos(np.pi/8), np.sin(np.pi/8)], dtype=complex),
}

print(f"\n{'State':36s}{'Δσx':>8s}{'Δσy':>8s}{'LHS':>9s}{'Bound|⟨σz⟩|':>12s}{'':>6s}")
print("-" * 79)
for name, psi in states.items():
    dx, dy = delta(sx, psi), delta(sy, psi)
    lhs = dx * dy
    lower = abs(expect(sz, psi))
    ok = "✔" if lhs >= lower - 1e-12 else "✘"
    print(f"{name:36s}{dx:8.4f}{dy:8.4f}{lhs:9.4f}{lower:12.4f}{ok:>4s}")

print("\nAll three examples satisfy the inequality, each with equality (minimum-uncertainty states).")
`````

## 10.6 Chapter Summary

### Review of the Theoretical Thread

This chapter took the mathematical language of discrete probability as its starting point (§10.1) and established the basic tools needed for the theory of quantum measurement: the normalization condition, the expectation value, and the variance. These seemingly elementary notions acquire deep physical counterparts in the quantum world—normalization corresponds to the conservation of probability, the expectation-value formula $\mathbb{E}[\mathsf{x}]=\sum x P(x)$ carries over directly to the quantum average of an observable $\langle\psi|\hat{A}|\psi\rangle$, and the standard deviation becomes the precise measure of uncertainty. The role of this section is to remind the reader that the strangeness of quantum mechanics lies not in any violation of probability theory but in the entirely new physical interpretation it gives to probability theory.

On this foundation, §10.2 gave the most central answer of quantum mechanics: a pure state is a normalized vector $|\psi\rangle$ in a Hilbert space, a "superposition state" is the natural result of expanding a vector in different bases, and the Born rule $P(i) = |\langle i|\psi\rangle|^2$ is the bridge between the geometric inner product and measurement probabilities. The description by pure states faces a fundamental limitation, however: many real physical systems—particles in thermal equilibrium, subsystems of entangled systems—cannot be characterized by a single state vector. §10.3 therefore introduced the density matrix $\boldsymbol{\rho}$, which places pure states (rank-one projections) in a unified framework of Hermitian, positive semidefinite matrices with trace one. The off-diagonal entries of the density matrix reveal quantum coherence, their disappearance is the mathematical embodiment of decoherence, and the purity $\text{Tr}(\boldsymbol{\rho}^2)$ provides an exact criterion distinguishing pure states from mixed states.

With the density matrix as a tool, §10.4 could describe completely the two levels of quantum measurement: the act of measurement (the off-diagonal entries vanish and coherence is lost) and the read-out of the outcome (the state collapses to a conditional pure state). The Pauli matrices $\sigma_x, \sigma_y, \sigma_z$, as the three observables of the spin-1/2 system, provide a $2\times2$ model that can be computed completely, and their noncommutation relations $[\sigma_j,\sigma_k]=2i\epsilon_{jkl}\sigma_l$ are not merely algebraic identities but the direct cause of the mutual interference of measurements. This preparation bears fruit in §10.5: with the Cauchy-Schwarz inequality at its core, the Robertson-Schrödinger inequality turns the noncommutativity of operators rigorously into a lower bound on the uncertainty product. The spin system demonstrates this structure perfectly in a finite-dimensional space, while the position-momentum uncertainty relation $\Delta X\cdot\Delta P\geqℏ/2$, through the trace argument, rigorously rules out any finite-dimensional realization, revealing the essential difference in mathematical structure between continuous and discrete degrees of freedom.

### Connections to Other Chapters

This chapter is deeply rooted in the theory of inner product spaces of Chapter 9. Hermitian matrices (observables), unitary matrices (quantum evolution), positive semidefinite matrices (density matrices), and the spectral theorem (the mathematical pillar of projective measurement)—these are exactly the tools that Chapter 9 built systematically, and in this chapter they receive their most vivid physical interpretation. In particular, the spectral decomposition $\hat{A}=\sum_j a_j\boldsymbol{\Pi}_j$ is a mathematical theorem in quantum mechanics, but connecting its eigenvalues and projections to measurement statistics also requires the Born rule and the measurement postulates; it shows that all measurement information about an observable is completely encoded in its eigenvalues and eigenprojections. The theorem of Chapter 9 that the eigenvalues of a Hermitian matrix are all real and its eigenvectors orthogonal reads physically as "measurement outcomes are necessarily real, and the post-measurement state is necessarily one of the orthogonal components"—a correspondence that depends on the postulate of finite-dimensional, rank-one ideal projective measurement adopted in this chapter.

The end point of this chapter leads naturally to the singular value decomposition (SVD) of Chapter 11. §10.3 already previewed the problem of the reduced density matrix of an entangled subsystem: when the composite system $|\Psi\rangle_{AB}$ is in an entangled state, the state of subsystem A must be described by a mixed state, and the standard tool for characterizing this mixed state is precisely the Schmidt decomposition $|\Psi\rangle_{AB}=\sum_k\lambda_k|u_k\rangle_A\otimes|v_k\rangle_B$—a special form of the SVD for composite quantum systems. The distribution of the Schmidt coefficients $\lambda_k$ quantifies the degree of entanglement, and the Schmidt number (the number of nonzero coefficients) determines whether entanglement is present. From this point of view, this chapter is the physical motivation for Chapter 11, and Chapter 11 is the algebraic answer to the questions this chapter leaves open.

### The Role of This Chapter in the Book

This chapter settles a central question that runs through the whole book: can the abstract algebraic machinery built in Chapters 8 and 9—eigenvalues, spectral decomposition, Hermitian matrices—describe the strangest physical phenomena of the real world? The answer is yes, with astonishing precision. Every basic rule of quantum mechanics—the discreteness of measurement outcomes, the superposition of states, the uncertainty principle, measurement collapse—is expressed in linear algebra, but the physical postulates and their range of validity must still be specified: observables are Hermitian matrices, measurements are projections, uncertainty is a necessary consequence of the noncommutativity of operators, and evolution is a unitary transformation. The quantum postulates are expressed in linear algebra; from these postulates together with linear algebra one can derive conclusions such as the measurement statistics and the uncertainty relations, but the physical postulates themselves are not theorems of linear algebra.

This chapter also opens a question, however: how should we describe nonclassical correlations among several quantum systems? An entangled state cannot be decomposed into a tensor product of states of its subsystems; this property of "the whole being greater than the sum of its parts" is the source of the power of quantum computing and quantum communication, and it is also the mathematical basis of the violation of Bell inequalities. Readers who enter Chapter 11 with this chapter's understanding of density matrices, measurement theory, and uncertainty will find that the singular value decomposition is not merely a technical tool of matrix analysis but the natural language for characterizing quantum entanglement. There, the abstract power of linear algebra meets the physical strangeness of the quantum world once again, and the reader is already equipped with every tool that is needed.
