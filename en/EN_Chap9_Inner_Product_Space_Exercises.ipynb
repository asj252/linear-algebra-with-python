### The Jacobi Method

The Jacobi method is one of the oldest numerical algorithms for finding the eigenvalues of a real symmetric matrix. It was proposed by Carl Gustav Jacob Jacobi in 1846, more than a century before the modern QR algorithm.

Its central idea comes from a simple observation: the eigenvalues of a diagonal matrix can be read off at a glance—they are the entries on the diagonal. So if a matrix can be "reduced" step by step to diagonal form by a sequence of **similarity transformations that leave the eigenvalues unchanged**, the problem is solved.

Recall the spectral theorem of Chapter 9: a real symmetric matrix $\mathbf{A}$ is always orthogonally diagonalizable, that is, there exists an orthogonal matrix $\mathbf{Q}$ such that

$$
\mathbf{Q}^\top \mathbf{A}\, \mathbf{Q} = \mathbf{\Lambda}
$$

The Jacobi method **approximates this $\mathbf{Q}$ constructively**—at each step it chooses a plane rotation (a Givens rotation) $\mathbf{G}_{ij}$ that eliminates one off-diagonal entry $a_{ij}$ exactly, while guaranteeing that the overall "off-diagonal energy"

$$
\text{off}(\mathbf{A})^2 = \sum_{i \neq j} a_{ij}^2
$$

decreases strictly monotonically. Although after one entry is eliminated, other entries in the same row and column may be disturbed and no longer be zero, we will see in the exercise below that each rotation reduces $\text{off}^2$ by exactly $2a_{ij}^2$; this "conservation of energy" viewpoint is the key to analyzing the convergence of the algorithm.


::::{exercise} The Jacobi Eigenvalue Method
:label: exer-jacobi-eigenvalue

Let the symmetric matrix be

$$
\mathbf{A} = \begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}
$$

**Part 1: A single rotation**

0. Write down the general form of the Givens rotation matrix $\mathbf{G}(\theta)$, and the formula that determines the rotation angle $\theta$ from the elimination condition.

1. Compute the angle $\theta$ needed to make the off-diagonal entries of $\mathbf{A}' = \mathbf{G}^\top \mathbf{A}\, \mathbf{G}$ zero.

2. Verify that the Frobenius norm is unchanged by the rotation: $\|\mathbf{A}'\|_F = \|\mathbf{A}\|_F$.

3. Verify the exact decrease formula for the off-diagonal norm:

$$
\text{off}(\mathbf{A}')^2 = \text{off}(\mathbf{A})^2 - 2a_{01}^2
$$

**Part 2: Geometric interpretation**

4. What is the geometric meaning of the columns of $\mathbf{G}$? How are they related to the eigenvectors of $\mathbf{A}$?

5. What quantities of $\mathbf{A}$ do the diagonal entries of $\mathbf{A}'$ correspond to?

**Part 3: Generalizing to $3\times 3$**

Consider the symmetric matrix

$$
\mathbf{B} = \begin{bmatrix} 4 & 2 & 0 \\ 2 & 3 & 1 \\ 0 & 1 & 2 \end{bmatrix}
$$

6. Following the classical Jacobi strategy (at each step choose the off-diagonal entry of largest absolute value), decide which position $(i, j)$ the first rotation should target, and compute the corresponding rotation angle $\theta$.

7. After performing one rotation, compute the new matrix $\mathbf{B}'$ and verify that $\text{off}(\mathbf{B}')^2 = \text{off}(\mathbf{B})^2 - 2b_{ij}^2$.

8. Explain why, after the rotation, the entry $b_{02}$ (and $b_{20}$) of $\mathbf{B}'$ that was previously zero may **no longer be zero**, while $\text{off}(\mathbf{B}')^2$ still decreases strictly.

:::{solution} exer-jacobi-eigenvalue
:class: dropdown

**Part 1**

**0.** The Givens rotation matrix is

$$
\mathbf{G}(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}
$$

The rotation angle is determined by the elimination condition: requiring the $(0,1)$ entry of $\mathbf{A}' = \mathbf{G}^\top \mathbf{A}\, \mathbf{G}$ to be zero, we derive

$$
\cot 2\theta = \frac{a_{00} - a_{11}}{2\, a_{01}} = \frac{3 - 3}{2 \cdot 1} = 0
\implies \theta = \frac{\pi}{4}
$$

**1.** Substituting $\theta = \pi/4$:

$$
\mathbf{G} = \begin{bmatrix} \frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \end{bmatrix}
$$

we compute

$$
\mathbf{A}' = \mathbf{G}^\top \mathbf{A}\, \mathbf{G} = \begin{bmatrix} 4 & 0 \\ 0 & 2 \end{bmatrix}
$$

**2.** Verify the Frobenius norm:

$$
\|\mathbf{A}\|_F^2 = 3^2 + 1^2 + 1^2 + 3^2 = 20
\qquad
\|\mathbf{A}'\|_F^2 = 4^2 + 0 + 0 + 2^2 = 20 \checkmark
$$

**3.** Verify the decrease formula:

$$
\text{off}(\mathbf{A})^2 = 1^2 + 1^2 = 2, \quad \text{off}(\mathbf{A}')^2 = 0
$$

$$
\text{off}(\mathbf{A}')^2 = 2 - 2 \times 1^2 = 0 \checkmark
$$

---

**Part 2**

**4.** The two columns of $\mathbf{G}$ are the basis directions after the rotation, and they are exactly the **eigenvectors** of $\mathbf{A}$ for the corresponding eigenvalues:

$$
\mathbf{v}_0 = \frac{1}{\sqrt{2}}\begin{bmatrix}1\\1\end{bmatrix}, \quad \mathbf{v}_1 = \frac{1}{\sqrt{2}}\begin{bmatrix}-1\\1\end{bmatrix}
$$

**5.** The diagonal entries $4$ and $2$ of $\mathbf{A}'$ are precisely the **eigenvalues** of $\mathbf{A}$.

---

**Part 3**

**6.** Compute the absolute values of the off-diagonal entries: $|b_{01}| = 2$, $|b_{12}| = 1$, $|b_{02}| = 0$. The maximum is at $(0,1)$, so the first rotation targets $(i,j) = (0,1)$:

$$
\cot 2\theta = \frac{b_{00} - b_{11}}{2\,b_{01}} = \frac{4-3}{2 \cdot 2} = \frac{1}{4}
$$

Solving gives $\theta \approx 0.6629$ rad ($\approx 37.98°$).

**7.** The rotation acts only on rows 0, 1 and columns 0, 1; computing gives

$$
\mathbf{B}' \approx \begin{bmatrix} 5.562 & 0 & 0.615 \\ 0 & 1.438 & 0.788 \\ 0.615 & 0.788 & 2 \end{bmatrix}
$$

Verify the decrease formula:

$$
\text{off}(\mathbf{B})^2 = 2(4 + 1 + 0) = 10
$$

$$
\text{off}(\mathbf{B}')^2 = 10 - 2 \times 2^2 = 2 \checkmark
$$

**8.** The rotation $\mathbf{G}_{01}$ acts only on the $\{0,1\}$ subspace. For each $k \neq 0,1$ (here $k=2$), the pair $(b_{0k},\, b_{1k})$ is rotated: its length is unchanged but its direction changes, so $b'_{02}$ changes from $0$ to $0.615 \neq 0$. However, since

$$
{b'}_{02}^2 + {b'}_{12}^2 = b_{02}^2 + b_{12}^2 = 0 + 1 = 1
$$

these entries merely **redistribute** the energy among different positions; the decrease in the overall off-diagonal norm is still exactly $2b_{01}^2$, a strict decrease.

$$\underline{\text{off}(\mathbf{B}')^2 = \text{off}(\mathbf{B})^2 - 2b_{01}^2 = 2}$$
