**Exercise: Global Phase Invariance and Physical Equivalence**

In quantum mechanics, a subtle but important idea is that **a global phase does not affect physical observables**. The following exercise will help you understand this property in depth.

::::{exercise} The Unobservability of a Global Phase
:label: exer-ch10-global-phase

Consider two states of a qubit:

$$|\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$$

$$|\psi'\rangle = e^{i\pi/3}\left(\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle\right) = \frac{e^{i\pi/3}}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle$$

**Part 1: Measurement Probabilities**

(a) Measuring in the computational basis $\{|0\rangle, |1\rangle\}$, compute:
   - the probabilities that $|\psi\rangle$ yields $|0\rangle$ and $|1\rangle$
   - the probabilities that $|\psi'\rangle$ yields $|0\rangle$ and $|1\rangle$
   - Are the measurement probabilities of the two states the same?

(b) Consider another measurement basis (the $x$ basis):
   $$|+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle), \quad |-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle)$$
   
   Compute the measurement probabilities of $|\psi\rangle$ and $|\psi'\rangle$ in this basis and check whether they are the same.

**Part 2: Expectation Values**

(c) For the Pauli operator $\hat{\sigma}_z = |0\rangle\langle 0| - |1\rangle\langle 1| = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$, compute:
   - $\langle \psi|\hat{\sigma}_z|\psi\rangle$
   - $\langle \psi'|\hat{\sigma}_z|\psi'\rangle$
   - and check whether the two are equal

(d) Repeat the computation above for the Pauli operator $\hat{\sigma}_x = |0\rangle\langle 1| + |1\rangle\langle 0| = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$.

**Part 3: Density Matrices**

(e) Compute the density matrices of $|\psi\rangle$ and $|\psi'\rangle$:
   $$\rho = |\psi\rangle\langle\psi|, \quad \rho' = |\psi'\rangle\langle\psi'|$$
   
   Write them in matrix form and verify that $\rho = \rho'$.

(f) Explain why equal density matrices mean that the two states are completely equivalent physically.

**Part 4: Observability of a Relative Phase (for Contrast)**

Now consider a third state, which differs from $|\psi\rangle$ only by a **relative phase**:
$$|\phi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle$$

Note: here only the coefficient of $|1\rangle$ carries a phase; this is not a global phase!

(g) Compute the measurement probabilities of $|\phi\rangle$ in the computational basis $\{|0\rangle, |1\rangle\}$. How do they compare with those of $|\psi\rangle$?

(h) Compute the measurement probabilities of $|\phi\rangle$ in the $x$ basis $\{|+\rangle, |-\rangle\}$. How do they compare with those of $|\psi\rangle$?

(i) Compute the density matrix $\rho_\phi = |\phi\rangle\langle\phi|$ of $|\phi\rangle$ and compare it with $\rho$. Pay particular attention to the **off-diagonal entries**.

(j) Summary: what is the essential difference between a global phase and a relative phase?

**Part 5: The Bloch Sphere Representation (Optional)**

(k) Using the Bloch sphere parametrization $|\psi(\theta,\phi)\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle$, write the values of $(\theta, \phi)$ corresponding to $|\psi\rangle$, $|\psi'\rangle$, and $|\phi\rangle$.

(l) Explain why $|\psi\rangle$ and $|\psi'\rangle$ are the same point on the Bloch sphere, while $|\phi\rangle$ is a different point.

**Part 1: Measurement Probabilities**

**(a) Measurement in the computational basis**

For $|\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$:
$$P_\psi(0) = \left|\langle 0|\psi\rangle\right|^2 = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2}$$
$$P_\psi(1) = \left|\langle 1|\psi\rangle\right|^2 = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2}$$

For $|\psi'\rangle = \frac{e^{i\pi/3}}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle$:
$$P_{\psi'}(0) = \left|\langle 0|\psi'\rangle\right|^2 = \left|\frac{e^{i\pi/3}}{\sqrt{2}}\right|^2 = |e^{i\pi/3}|^2 \cdot \frac{1}{2} = 1 \cdot \frac{1}{2} = \frac{1}{2}$$
$$P_{\psi'}(1) = \left|\langle 1|\psi'\rangle\right|^2 = \left|\frac{e^{i\pi/3}}{\sqrt{2}}\right|^2 = \frac{1}{2}$$

**Conclusion**: the measurement probabilities are exactly the same! The global phase factor $e^{i\pi/3}$ cancels when the squared modulus is taken.

**(b) Measurement in the $x$ basis**

First expand $|\psi\rangle$ in the $x$ basis:
$$|\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle = |+\rangle$$

Therefore:
$$P_\psi(+) = |\langle +|\psi\rangle|^2 = |1|^2 = 1$$
$$P_\psi(-) = |\langle -|\psi\rangle|^2 = |0|^2 = 0$$

For $|\psi'\rangle$:
$$|\psi'\rangle = e^{i\pi/3}\left(\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle\right) = e^{i\pi/3}|+\rangle$$

Therefore:
$$P_{\psi'}(+) = |\langle +|\psi'\rangle|^2 = |e^{i\pi/3}|^2 = 1$$
$$P_{\psi'}(-) = |\langle -|\psi'\rangle|^2 = 0$$

**Conclusion**: the measurement probabilities are the same in every basis.

---

**Part 2: Expectation Values**

**(c) The expectation value of $\hat{\sigma}_z$**

For $|\psi\rangle$:
$$\langle \psi|\hat{\sigma}_z|\psi\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$$
$$= \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{pmatrix} 1 \\ -1 \end{pmatrix} = \frac{1}{2}(1 - 1) = 0$$

For $|\psi'\rangle = e^{i\pi/3}|\psi\rangle$:
$$\langle \psi'|\hat{\sigma}_z|\psi'\rangle = \langle \psi|e^{-i\pi/3}\hat{\sigma}_z e^{i\pi/3}|\psi\rangle = e^{-i\pi/3}e^{i\pi/3}\langle \psi|\hat{\sigma}_z|\psi\rangle = 0$$

**Conclusion**: the expectation values are the same.

**(d) The expectation value of $\hat{\sigma}_x$**

For $|\psi\rangle$:
$$\langle \psi|\hat{\sigma}_x|\psi\rangle = \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$$
$$= \frac{1}{2}\begin{pmatrix} 1 & 1 \end{pmatrix}\begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{2}(1 + 1) = 1$$

For $|\psi'\rangle$:
$$\langle \psi'|\hat{\sigma}_x|\psi'\rangle = 1$$

(by the same reasoning)

---

**Part 3: Density Matrices**

**(e) Computing the density matrices**

For $|\psi\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$:
$$\rho = |\psi\rangle\langle\psi| = \frac{1}{2}\begin{pmatrix} 1 \\ 1 \end{pmatrix}\begin{pmatrix} 1 & 1 \end{pmatrix} = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$$

For $|\psi'\rangle = \frac{e^{i\pi/3}}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$:
$$\rho' = |\psi'\rangle\langle\psi'| = \frac{e^{i\pi/3}}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix} \cdot \frac{e^{-i\pi/3}}{\sqrt{2}}\begin{pmatrix} 1 & 1 \end{pmatrix}$$
$$= \frac{e^{i\pi/3}e^{-i\pi/3}}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix} = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix} = \rho$$

**Check**: $\rho = \rho'$ ✓

**(f) Explaining the physical equivalence**

The density matrix $\rho$ contains all the information needed to compute every observable:
- Any measurement probability: $P(n) = \langle n|\rho|n\rangle$
- Any expectation value: $\langle \hat{A} \rangle = \text{Tr}(\hat{A}\rho)$

If two states have the same density matrix ($\rho = \rho'$), they give the same results in **every possible experiment** and are therefore completely equivalent physically.

The global phase factor cancels automatically when the density matrix is formed:
$$\rho' = (e^{i\gamma}|\psi\rangle)(e^{-i\gamma}\langle\psi|) = e^{i\gamma}e^{-i\gamma}|\psi\rangle\langle\psi| = |\psi\rangle\langle\psi| = \rho$$

**Part 4: Observability of a Relative Phase**

**(g) Measurement in the computational basis**

For $|\phi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle$:
$$P_\phi(0) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2}$$
$$P_\phi(1) = \left|\frac{e^{i\pi/3}}{\sqrt{2}}\right|^2 = \frac{1}{2}$$

**Observation**: the measurement probabilities are the same as those of $|\psi\rangle$! The two states cannot be distinguished in the computational basis.

**(h) Measurement in the $x$ basis**

Expand $|\phi\rangle$ in the $x$ basis:
$$|\phi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle$$

Using $|0\rangle = \frac{1}{\sqrt{2}}(|+\rangle + |-\rangle)$ and $|1\rangle = \frac{1}{\sqrt{2}}(|+\rangle - |-\rangle)$:
$$|\phi\rangle = \frac{1}{2}(|+\rangle + |-\rangle) + \frac{e^{i\pi/3}}{2}(|+\rangle - |-\rangle)$$
$$= \frac{1 + e^{i\pi/3}}{2}|+\rangle + \frac{1 - e^{i\pi/3}}{2}|-\rangle$$

Compute $1 + e^{i\pi/3} = 1 + \cos(\pi/3) + i\sin(\pi/3) = 1 + 1/2 + i\sqrt{3}/2 = \frac{3 + i\sqrt{3}}{2}$:
$$|1 + e^{i\pi/3}|^2 = \left|\frac{3 + i\sqrt{3}}{2}\right|^2 = \frac{9 + 3}{4} = 3$$

Therefore:
$$P_\phi(+) = \frac{|1 + e^{i\pi/3}|^2}{4} = \frac{3}{4}$$
$$P_\phi(-) = \frac{|1 - e^{i\pi/3}|^2}{4} = \frac{1}{4}$$

**Observation**: these differ from those of $|\psi\rangle$! ($P_\psi(+) = 1$, $P_\phi(+) = 3/4$)

**Key conclusion**: a relative phase **can** be observed by choosing a suitable measurement basis.

**(i) Comparing the density matrices**

$$\rho_\phi = |\phi\rangle\langle\phi| = \frac{1}{2}\begin{pmatrix} 1 \\ e^{i\pi/3} \end{pmatrix}\begin{pmatrix} 1 & e^{-i\pi/3} \end{pmatrix}$$
$$= \frac{1}{2}\begin{pmatrix} 1 & e^{-i\pi/3} \\ e^{i\pi/3} & 1 \end{pmatrix}$$

whereas the earlier one is:
$$\rho = \frac{1}{2}\begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$$

**Key differences**:
- The off-diagonal entries of $\rho$: $\rho_{01} = \rho_{10} = 1/2$ (real)
- The off-diagonal entries of $\rho_\phi$: $(\rho_\phi)_{01} = e^{-i\pi/3}/2$, $(\rho_\phi)_{10} = e^{i\pi/3}/2$ (complex)

**The off-diagonal entries carry the relative phase information**; this is the mathematical signature of quantum coherence!

**(j) Summary of the comparison**

| Property | Global phase $e^{i\gamma}\|\psi\rangle$ | Relative phase $\alpha\|0\rangle + e^{i\phi}\beta\|1\rangle$ |
| :--- | :--- | :--- |
| **Physically observable** | No | Yes  |
| **Affects measurement probabilities** | No (in every basis) | Yes (in some bases) |
| **Density matrix** | Unchanged | Off-diagonal entries change |
| **Bloch sphere** | Same point | Different points |
| **Interference effects** | No effect | Decisive effect |


**Core distinction**:
- **Global phase**: all coefficients rotate together, and their relative relationships do not change
- **Relative phase**: the phase difference between coefficients, which determines the interference pattern

**Part 5: The Bloch Sphere Representation**

**(k) Parametrization**

The standard Bloch sphere form: $|\psi(\theta,\phi)\rangle = \cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle$

**For $|\psi\rangle$**:
$$|\psi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle = \cos\frac{\pi/2}{2}|0\rangle + e^{i \cdot 0}\sin\frac{\pi/2}{2}|1\rangle$$

Hence $\theta = \pi/2$, $\phi = 0$ (on the equator, in the positive $x$ direction).

**For $|\psi'\rangle$**:
$$|\psi'\rangle = e^{i\pi/3}\left(\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle\right)$$

By global phase invariance, factor out $e^{i\pi/3}$:
$$|\psi'\rangle \equiv \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$$

Hence $\theta = \pi/2$, $\phi = 0$ (**the same as $|\psi\rangle$!**).

**For $|\phi\rangle$**:
$$|\phi\rangle = \frac{1}{\sqrt{2}}|0\rangle + \frac{e^{i\pi/3}}{\sqrt{2}}|1\rangle = \cos\frac{\pi/2}{2}|0\rangle + e^{i\pi/3}\sin\frac{\pi/2}{2}|1\rangle$$

Hence $\theta = \pi/2$, $\phi = \pi/3$ (on the equator, but with a different azimuthal angle).

**(l) Explanation on the Bloch sphere**

- $|\psi\rangle$ and $|\psi'\rangle$: **the same point** $(\pi/2, 0)$
  - Reason: the global phase $e^{i\pi/3}$ can be "absorbed" and is not reflected in the Bloch sphere coordinates
  
- $|\phi\rangle$: **a different point** $(\pi/2, \pi/3)$
  - Reason: the relative phase $\phi = \pi/3$ affects the azimuthal angle on the Bloch sphere

**Geometric intuition**:
- Each point on the Bloch sphere represents a **physically inequivalent** quantum state
- Global phase invariance means that a phase "rotation" around the unit circle, $|\psi\rangle \mapsto e^{i\gamma}|\psi\rangle$, does not change the physical state
- Only the position on the sphere ($\theta, \phi$) has physical meaning

This is exactly why quotienting $\mathbb{C}^2$ (a 4-dimensional real space) by normalization (1 dimension) and the global phase (1 dimension) gives $S^2$ (a 2-dimensional sphere).

```{important}
:label: imp-global-vs-relative-phase
**Global phase vs. relative phase: the key points**

1. **Global phase**:
   $$|\psi'\rangle = e^{i\gamma}|\psi\rangle$$
   - All coefficients are multiplied by the same $e^{i\gamma}$
   - **Completely unobservable physically**
   - The density matrix is unchanged: $\rho' = \rho$
   - The same point on the Bloch sphere

2. **Relative phase**:
   $$|\psi\rangle = \alpha|0\rangle + e^{i\phi}\beta|1\rangle$$
   - The phase difference $\phi$ between the coefficients
   - **Physically observable** (in a suitable basis)
   - Affects the off-diagonal entries of the density matrix
   - Different points on the Bloch sphere

3. **Why is a global phase unobservable?**
   - All physical quantities are sesquilinear forms: $\langle \psi|\hat{A}|\psi\rangle$ or $|\langle n|\psi\rangle|^2$
   - The global phase cancels between the ket and the bra
   - Only the **relative phase** survives in the computation

4. **Mathematical essence**:
   - The space of quantum states is really the **projective Hilbert space** $\mathbb{P}\mathcal{H}$
   - The equivalence relation: $|\psi\rangle \sim e^{i\gamma}|\psi\rangle$ for all $\gamma \in \mathbb{R}$
   - The Bloch sphere is a geometric realization of this projective space

This idea is essential for understanding quantum interference, the Berry phase, phase gates in quantum computing, and more.
```


:::




::::

```{admonition} Exercise Hints
:class: tip

**Review of the key concepts**:

1. **Global phase**: all coefficients are multiplied by $e^{i\gamma}$
   - Form: $|\psi'\rangle = e^{i\gamma}|\psi\rangle$
   - Unobservable; it affects no physical quantity

2. **Relative phase**: the phase difference between coefficients
   - Form: the $\phi$ in $\alpha|0\rangle + e^{i\phi}\beta|1\rangle$
   - Observable; it affects interference effects

3. **Methods of verification**:
   - Probabilities: $P(n) = |\langle n|\psi\rangle|^2$
   - Expectation values: $\langle \hat{A} \rangle = \langle \psi|\hat{A}|\psi\rangle$
   - Density matrix: $\rho = |\psi\rangle\langle\psi|$

**Computational tips**:
- Inner products: $\langle 0|0\rangle = 1$, $\langle 0|1\rangle = 0$
- Modulus of a complex number: $|e^{i\theta}|^2 = 1$
- Density matrix: $\rho_{ij} = \psi_i\overline{\psi_j}$ (note the order: the row index $i$ takes the entry itself and the column index $j$ takes the conjugate, consistent with $\rho = |\psi\rangle\langle\psi|$)
```
