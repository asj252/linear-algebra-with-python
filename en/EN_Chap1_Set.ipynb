Sorting blocks by color, dividing food into what we like and what we do not—classification is one of the ways we make sense of the world. Everyday operations such as "these things go together" and "those go somewhere else" offer an intuitive starting point for approaching the language of sets.

Between everyday classification and rigorous set notation lies one more step: stating these familiar operations precisely, in definitions that anyone can check word by word.

Yet humans spent two thousand years computing before turning back to write these familiar operations down in rigorous symbols.

---

Why such a long detour? Because computation ran into a crisis.

The central task of ancient mathematics was to compute particular cases: how large this field is, how much tax is owed, how long this hypotenuse is. The *Nine Chapters on the Mathematical Art* works this way, the Babylonian clay tablets work this way, and so do the Egyptian papyri. Greek geometry took a step further—Euclid proved properties of **all** triangles, not of one particular triangle—but it still relied on geometric intuition, and wherever a boundary blurred it let the picture do the talking.

In the nineteenth century the boundaries of intuition began to collapse. In 1872 Weierstrass constructed a function that is continuous everywhere and differentiable nowhere. Geometric intuition says such an object should not exist at all, and yet it can be proved from the rigorous definitions that it is continuous everywhere and differentiable at no point. Pictures could no longer be trusted. Mathematicians realized that a feeling for individual objects was no longer enough: they had to say precisely **which class** of things was under discussion, and then reason rigorously about that entire class.

Dedekind conceived the idea of the Dedekind cut in 1858 and published it in 1872, using it to redefine the real numbers—not by producing a formula to compute with, but by defining each real number as a partition of the set of rational numbers. This showed another way to build a number system: characterizing numbers through sets and their properties, rather than only supplying a computational procedure. Cantor followed close behind. In the 1870s he built set theory, giving this way of describing objects a general language: **sets**, and the **mappings** between them.

This is the historical turning point at which mathematics moved from computing particular cases to discussing the properties of an entire class of objects.

---

In the middle of the twentieth century, a group of French mathematicians pushed that turn to its extreme.

In 1935 a number of young French mathematicians, writing under the pen name "Nicolas Bourbaki," began an enormously ambitious series of books, the *Éléments de mathématique*. The title pays homage to Euclid's *Elementa*, but it carries a deliberate provocation. In Euclid's day mathematics was a collection of many disciplines, so French normally writes the plural *mathématiques*; Bourbaki insisted on the singular *mathématique*—a declaration that mathematics is **one** subject, with a single foundation, and that the foundation is set theory. The name belongs to no real person but to a collective whose members included André Weil, Henri Cartan, and Jean Dieudonné, among the finest mathematicians of the day. Their goal was single-minded: to rewrite all of modern mathematics from the ground up in the language of set theory, leaving no place that appeals to intuition or leans on a picture.

Bourbaki's influence was profound and contested. Admirers say the group gave twentieth-century mathematics a unified foundation of language; critics say the books are too abstract, that they expel every trace of geometric intuition, and that reading them feels like reading statutes. Either way, by making formalization the highest standard of mathematics, Bourbaki changed the face of mathematical education and research for good.

The structural viewpoint that Bourbaki emphasized also offers a point of entry for thinking about children's cognition. The Swiss psychologist Jean Piaget observed that the classifying, ordering, and spatial operations that children gradually develop can be set side by side with the structural viewpoint of modern mathematics.

This comparison gives the chapter a point of departure: sets and mappings, abstract as they look, can be approached through the everyday activities of classifying, ordering, and matching. It suggests that abstract language can connect with everyday cognition; it does not mean that the full concept of a set, or its formal definition, is innate.

---

The task of this chapter is to build the foundation of that formal language.

Set theory is not an isolated chapter in this book; it is the bedrock language of the whole book. "A vector space is a set equipped with two operations satisfying eight axioms"—this sentence speaks about **all** vector spaces, not about one concrete example. Once a theorem has been proved from it, that theorem applies at once to $\mathbb{R}^n$, to polynomial spaces, to function spaces, and to matrix spaces, with no need to redo the work for each example. This is something computation cannot do, and it is what the language of sets does most naturally.

## 1.1 Foundations of Set Theory

We begin with the most basic notions, laying a solid mathematical foundation for the central theories of linear algebra: vector spaces, linear transformations, and everything built on them.

### 1.1.1 Definition and Representation of Sets

:::{prf:definition} Set
:label: def-set-detailed
A set is a whole made up of certain distinct objects. Each object in a set is called an element, or a member, of that set.

**Key properties.**

- **Definiteness.** For any object whatsoever, either it belongs to the set or it does not.

- **Distinctness.** The elements of a set are pairwise different.

- **Unorderedness.** The elements of a set have no fixed order of arrangement.

**Notation.** We write $a \in A$ to say that the element $a$ belongs to the set $A$, and $a ∉ A$ to say that $a$ does not belong to $A$.
:::

A set may be presented in either of the following ways:

1. **Roster notation**: list all of the elements directly, as in $A = \{1, 2, 3, 4, 5\}$
2. **Set-builder notation**: describe the set by a property or logical condition that its elements satisfy, as in $B = \{x | x \text{ is a positive even number less than } 10\}$

```{note} Logical and Set-Theoretic Symbols Used in This Book
:label: note-logic-symbols
So that a first encounter with a formal definition does not bring you to a halt, we fix here a few symbols that run through the entire book:

- $\{x \mid P(x)\}$: the set of all $x$ for which the statement $P(x)$ holds (the vertical bar $\mid$ is read "such that").
- $\forall$: "for all"; $\exists$: "there exists". For example, $\forall x \in A,\ \exists y \in B$ is read "for every $x$ in $A$ there exists some $y$ in $B$".
- $\Rightarrow$: "implies" (if ... then ...); $\Leftrightarrow$: "if and only if" (implication in both directions).
- $|S|$: the **cardinality** of the set $S$, that is, its number of elements (for a finite set, simply a count).
```

:::{prf:example} Number Systems
:label: ex-special-set
- **The natural numbers** $\mathbb{N}$: $\{0, 1, 2, 3, ...\}$ or $\{1, 2, 3, ...\}$ (depending on the convention adopted)
- **The integers** $\mathbb{Z}$: $\{..., -2, -1, 0, 1, 2, ...\}$
- **The finite field** $\mathbb{F}_p$: $\{0,1,2,...,p-1\}$, where $p$ is a prime number and both addition and multiplication are carried out "modulo $p$" (for instance, in $\mathbb{F}_5$ we have $3+4=2$); the rigorous definition of a field appears in a later chapter.
- **The rational numbers** $\mathbb{Q}$: $\{p/q | p, q \in \mathbb{Z}, q \neq 0\}$
- **The real numbers** $\mathbb{R}$: all rational numbers together with all irrational numbers
- **The complex numbers** $\mathbb{C}$: $\{a + bi | a, b \in \mathbb{R}, i^2 = -1\}$
:::

### 1.1.2 Relations and Operations Between Sets

:::{prf:definition} Subset and Inclusion
:label: def-subset
If every element of a set $A$ is also an element of a set $B$, then $A$ is called a **subset** of $B$, or equivalently $B$ contains $A$, written $A \subseteq B$.
:::

:::{prf:definition} Equality of Sets
:label: def-set-equality
Two sets $A$ and $B$ are **equal** (written $A = B$) when they have exactly the same elements: for every object $x$, $x \in A$ if and only if $x \in B$.
:::

The most common way to prove that two sets are equal is **double inclusion**: prove $A \subseteq B$ and $B \subseteq A$ separately, and once both hold we have $A = B$. To prove a single inclusion $A \subseteq B$, the standard template is one sentence long—**take an arbitrary $x \in A$ and derive $x \in B$**. The example below demonstrates that template in its most basic setting.



::::{exercise} Mutually Contained Sets
:label: exer-subset-equal
Suppose $A \subseteq B$ and $B \subseteq A$. Prove that $A = B$.

:::{solution} exer-subset-equal
:class: dropdown
1. By {prf:ref}`def-set-equality`, proving $A = B$ requires showing that for every object $x$, $x \in A$ if and only if $x \in B$.
2. ($\Rightarrow$) Take an arbitrary $x \in A$ (this is the fixed opening move for proving an inclusion). Since $A \subseteq B$, the element $x$ belongs to $A$ and therefore also to $B$, that is, $x \in B$.
3. ($\Leftarrow$) Conversely, take an arbitrary $x \in B$. Since $B \subseteq A$, the same reasoning gives $x \in A$.
4. Combining the two directions yields $x \in A \Leftrightarrow x \in B$, so $A = B$ by {prf:ref}`def-set-equality`.

$\square$
:::
::::



:::{prf:example} Sets Defined by Their Relation to Others
:label: ex-set-terminology
- **The universal set** $U$: within a particular discussion, the set containing every element under consideration
- **A singleton**: a set containing exactly one element, such as $\{a\}$
- **The empty set** $\emptyset$, also written $\{\}$, contains no elements at all, and it is a subset of every set.

Why is the empty set a subset of every set? The claim conflicts with intuition (if it contains nothing, how can it be "contained in" anything?), but reading {prf:ref}`def-subset` word for word dissolves the difficulty: to refute $\emptyset \subseteq A$ we would have to find an element of $\emptyset$ that does not belong to $A$; since $\emptyset$ has no elements to find, no refutation is possible, and the statement stands. Reasoning of this kind—"true because no counterexample can be found"—is called **vacuous truth**, and it is our first demonstration of the formal spirit of reasoning strictly from the definition.
:::



:::{prf:definition} Power Set
:label: def-powerset
The set of all subsets of a set $A$ is called the **power set** of $A$, written $P(A)$.
:::

:::{prf:example} Listing a Power Set
:label: ex-powerset-enum
Take $A = \{1, 2\}$. The subsets of $A$ are the empty set $\emptyset$, $\{1\}$, $\{2\}$, and $\{1,2\}$—four in all—so
$$P(\{1,2\}) = \{\emptyset,\ \{1\},\ \{2\},\ \{1,2\}\}.$$
Why exactly $4 = 2^2$ of them? Because constructing a subset amounts to making, for **each element** of $A$, an independent two-way decision: include it or leave it out. Element $1$ is in or out (2 ways), element $2$ is in or out (2 ways), and multiplying gives $2 \times 2 = 4$ combinations. This point of view also supplies the intuition behind the formula $2^n$ in the next exercise.
:::

```{exercise} The Power Set of a Finite Set
:label: exer-powerset-size
If a finite set $A$ has $n$ elements, then its power set $P(A)$ has $2^n$ elements.

*Hint.* Follow the "in or out for each element" idea of {prf:ref}`ex-powerset-enum` and multiply $n$ independent two-way choices to obtain $2^n$; alternatively, use mathematical induction on $n$ (adding one new element splits each existing subset into two, one containing the new element and one not).
```

:::{prf:definition} Cartesian Product
:label: def-cartesian-product
The Cartesian product (direct product) $A \times B$ of sets $A$ and $B$ is the set of all ordered pairs $(a, b)$ with $a \in A$ and $b \in B$.

**Intuition.** Picture the Cartesian product as a grid whose columns are indexed by the elements of $A$ and whose rows are indexed by the elements of $B$; each intersection in the grid is one ordered pair. For example, if $A = \{1, 2\}$ and $B = \{x, y\}$, then $A \times B = \{(1,x), (1,y), (2,x), (2,y)\}$.

**Generalization to the $n$-fold product.** The two-set definition above generalizes step by step. Given $n$ sets $A_0, A_1, \ldots, A_{n-1}$, their direct product
$$A_0 \times A_1 \times \cdots \times A_{n-1} = \{(a_0, a_1, \ldots, a_{n-1}) \mid a_i \in A_i,\ i = 0, 1, \ldots, n-1\}$$
is the set of all ordered $n$-tuples. In particular, taking every $A_i$ to be the set of real numbers $\mathbb{R}$ gives the $n$-dimensional coordinate space
$$\mathbb{R}^n = \underbrace{\mathbb{R} \times \cdots \times \mathbb{R}}_{n \text{ copies}}.$$
This is exactly the concrete support for the statement in Chapter 2 that a vector space "is first of all a set."
:::

```{note} An Anecdote About the Cartesian Coordinate System
:label: note-cartesian-space
Several versions circulate of the story of how the philosopher and mathematician Descartes invented the Cartesian coordinate system. In the best known one, Descartes lay ill in bed one morning, staring at the ceiling. He noticed a fly crawling across it and began to wonder how he might describe the fly's exact position to someone else. He realized that the position could be pinned down by measuring the fly's distances from two perpendicular walls of the room. That insight led him to develop a coordinate system in which any point can be located by two numbers, its x- and y-coordinates.
```

:::{prf:definition} Set Operations
:label: def-set-operations
* **Union**: $A \cup B = \{x \mid x \in A \text{ or } x \in B\}$
* **Intersection**: $A \cap B = \{x \mid x \in A \text{ and } x \in B\}$
* **Complement**: $A^\complement = U\backslash A = \{x \in U \mid x ∉ A\}$, where $U$ is the universal set
* **Set difference**: $A \backslash B = \{x \in A \mid x ∉ B\}$
:::

:label: ex-set-venn
Set operations can be displayed directly by means of a **Venn diagram**. A union appears as the whole of the two circles, and an intersection as the region where they overlap.

The intersection must be subtracted once because it has been counted once inside $|A|$ and once inside $|B|$.
:::

## 1.2. Foundations of Mappings

Linear algebra begins with *multivariable linear functions*, that is, with linear mappings. Elementary as this class of functions is, it lays the groundwork for understanding far more intricate mathematical ideas. As you go further, you will find that other kinds of functions also have their place in linear algebra.

### 1.2.1. Definition and Basic Notions of a Mapping

:::{prf:definition} Mapping
:label: def-mapping
Let $X$ and $Y$ be nonempty sets. If there is a rule $f$ assigning to each $x \in X$ a uniquely determined $y \in Y$, then $f$ is called a **mapping** (or a **function**) from $X$ to $Y$, written $f: X \rightarrow Y$.

This book adopts the convention that the domain and codomain of a mapping are nonempty sets. This is a convention about the scope of the book, not a theorem about mappings in general; a definition that allows the empty set would also be acceptable, but this book does not discuss empty mappings.

Mappings go by other names as well, such as transformations and morphisms; this book draws no strict distinction among them.
:::

A function may be written as $f: X \rightarrow Y$, $y = f(x)$, where $x \in X$ and $y \in Y$.

```{attention} *The Defining Feature of a Mapping*
:label: att-mapping-key
The defining feature of a mapping is this: **each element $x \in X$ has one and only one corresponding element $y \in Y$**.
```

:::{prf:definition} Domain, Codomain, Image, Image of a Set, Range
:label: def-domain-range
* **Domain**: the set $X$ in a mapping $f: X \rightarrow Y$
* **Codomain**: the set $Y$ in a mapping $f: X \rightarrow Y$
* **Image of an element**: for $x \in X$, the element $f(x) \in Y$ is called the image of $x$.
* **Image of a set**: for a subset $A \subseteq X$, the image of $A$ under $f$ is $f[A] = \{f(x) | x \in A\}$ (also written $f(A)$)
* **Range** (that is, the image of $f$): $f[X] = \operatorname{Im}f = \{y \in Y | \exists x \in X, y = f(x)\}$, the set of the images of all elements of $X$. **The range is in general a subset of the codomain**, $f[X] \subseteq Y$, and the two need not coincide—this is exactly the distinction emphasized in {ref}`cau-range-codomain` below.
:::

```{caution} *The Difference Between the Image and the Codomain*
:label: cau-range-codomain
Take care to distinguish the **image of a set** from the **codomain**: the image is the set of elements of $Y$ that elements of $X$ are actually mapped to, whereas the codomain is the range of values available in principle. The image is a subset of the codomain, $f[X] \subseteq Y$.
```

:::{prf:example} A Function
:label: ex-function-domain
The function $f(x) = \frac{1}{x-2}$ has domain $\{x \in \mathbb{R} | x \neq 2\}$; its codomain may be taken to be $\mathbb{R}$, and its range is $\{y \in \mathbb{R} | y \neq 0\}$.
:::

:::{prf:example} Common Types of Single-Variable Functions
:label: ex-common-function
1. **Constant function**: $f(x) = c$ (with $c$ a constant)
2. **Identity function**: $f(x) = x$
3. **Polynomial function**: $f(x) = a_nx^n + a_{n-1}x^{n-1} + \cdots + a_1x + a_0$
4. **Exponential function**: $f(x) = a^x$ (with $a > 0, a \neq 1$)
5. **Logarithmic function**: $f(x) = \log_a x$ (with $a > 0, a \neq 1$)
:::


:::{prf:definition} Equality of Mappings
:label: def-mapping-equality
Let $f: X \rightarrow Y$ and $g: X \rightarrow Y$ be mappings. If $f(x) = g(x)$ for every $x \in X$, then the two mappings are said to be **equal**, written $f = g$.
:::

:::{prf:example} Deciding Whether Two Mappings Are Equal
:label: ex-mapping-equality
Consider the following three mappings:
- $f_0: \mathbb{R} \rightarrow \mathbb{R}$, $f_0(x) = x^2$
- $f_1: \mathbb{R} \rightarrow \mathbb{R}$, $f_1(x) = (-x)^2$
- $f_2: [0,+\infty) \rightarrow \mathbb{R}$, $f_2(x) = x^2$

**Analysis.**
- $f_0 = f_1$, because they have the same domain and the same codomain, and for every $x \in \mathbb{R}$ we have $f_0(x) = x^2 = (-x)^2 = f_1(x)$
- $f_0 \neq f_2$: although the rule of correspondence is the same (squaring in both cases), the domains differ, since $f_0$ may be applied to negative numbers while $f_2$ may be applied only to nonnegative ones
:::

:::{prf:definition} Composite Mapping
:label: def-composite-mapping
Let $f: X \rightarrow Y$ and $g: Y \rightarrow Z$ be mappings. The **composite mapping** $g \circ f: X \rightarrow Z$ is defined by

$$(g \circ f)(x) = g(f(x)), \quad \forall x \in X$$
:::

Forming a composite mapping may be understood as "**apply $f$ first, then apply $g$**."

:::{prf:property} Basic Properties of Composition
:label: pro-composite-properties
Let $f: X \rightarrow Y$, $g: Y \rightarrow Z$, and $h: Z \rightarrow W$ be three mappings. Then:

1. **Associativity**: $(h \circ g) \circ f = h \circ (g \circ f)$
2. **Action of the identity mapping**: $f \circ \text{id}_X = f$ and $\text{id}_Y \circ f = f$, where $\text{id}_X(x) = x$ is the identity mapping on $X$
3. **Commutativity fails in general**: in general, $g\circ f$ being defined does not imply that $f\circ g$ is defined; even when $f,g:X\to X$, the two need not be equal.
:::

:::{prf:example} Associativity of Composition
:label: ex-composite-associativity
Let:
- $f: \mathbb{R} \rightarrow \mathbb{R}$, $f(x) = x + 1$
- $g: \mathbb{R} \rightarrow \mathbb{R}$, $g(x) = 2x$
- $h: \mathbb{R} \rightarrow \mathbb{R}$, $h(x) = x^2$

Verifying associativity:
- $((h \circ g) \circ f)(x) = (h \circ g)(x + 1) = h(2(x + 1)) = h(2x + 2) = (2x + 2)^2$
- $(h \circ (g \circ f))(x) = h((g \circ f)(x)) = h(2(x + 1)) = h(2x + 2) = (2x + 2)^2$

Hence $(h \circ g) \circ f = h \circ (g \circ f)$.

With the same $f$ and $g$ as in this example, we can also compare $g(f(x))=2x+2$ with $f(g(x))=2x+1$; hence $g\circ f\neq f\circ g$.
:::

```{tip} ◇*Composition and Matrices*
:label: tip-composite-notation
In linear algebra, composition of mappings shows up constantly in **matrix multiplication**. If matrices $A$ and $B$ represent linear mappings $T_A$ and $T_B$ respectively, then the matrix product $BA$ corresponds to the composite mapping $T_B \circ T_A$.
```

:::{prf:definition} Restriction of a Mapping
:label: def-restricted-mapping
Let $f: X \to Y$ be a mapping and let $\emptyset \neq A \subseteq X$ be a nonempty subset of $X$. The **restriction** of $f$ to $A$, written $f|_A$, is defined by

$$f|_A: A \to Y$$

$$f|_A(x) = f(x), \quad \forall x \in A$$

In other words, $f|_A$ is the original mapping $f$ with its domain cut down to the subset $A$.
:::

### 1.2.2 The Preimage

:::{prf:definition} The Notion of a Preimage
:label: def-preimage
Let $f: X \rightarrow Y$ be a mapping and let $B \subseteq Y$. The **preimage** of $B$ is

$$f^{-1}[B] = \{x \in X | f(x) \in B\}$$
:::

```{caution} *The Preimage Is Not the Inverse Mapping*
:label: cau-preimage
**Note.** The symbol $f^{-1}[B]$ denotes a **preimage**; the $f^{-1}$ appearing in it is *not* an inverse mapping. The notion of a preimage remains meaningful even when $f$ is not invertible. We use square brackets $[\,]$ rather than parentheses $(\,)$ to avoid confusion with function notation.
```

:::{prf:example} Preimages
:label: ex-image-preimage
For the function $f(x) = x^2$ (from $\mathbb{R}$ to $\mathbb{R}$):
* the image of the set $A = [-1, 1]$ is $f[A] = [0, 1]$
* the preimage of the set $B = [1, 4]$ is $f^{-1}[B] = [-2, -1] \cup [1, 2]$
:::

:::{prf:property} Properties of Images and Preimages

For a mapping $f: X \rightarrow Y$ and subsets $A_0, A_1 \subseteq X$, $B_0, B_1 \subseteq Y$:

1. $A_0 \subseteq A_1 \Rightarrow f[A_0] \subseteq f[A_1]$
2. $B_0 \subseteq B_1 \Rightarrow f^{-1}[B_0] \subseteq f^{-1}[B_1]$
3. $f[A_0 \cup A_1] = f[A_0] \cup f[A_1]$
4. $f[A_0 \cap A_1] \subseteq f[A_0] \cap f[A_1]$ (note: equality need not hold)
5. $f^{-1}[B_0 \cup B_1] = f^{-1}[B_0] \cup f^{-1}[B_1]$
6. $f^{-1}[B_0 \cap B_1] = f^{-1}[B_0] \cap f^{-1}[B_1]$
:::

### 1.2.3 Types of Mappings

:::{prf:definition} Injective Mapping
:label: def-injection
A mapping $f: X \rightarrow Y$ is called **injective** (also **one-to-one**, or **1 to 1**) if for any $x_0, x_1 \in X$ with $x_0 \neq x_1$ we have $f(x_0) \neq f(x_1)$.

**Equivalent formulation.** $\forall x_0, x_1 \in X, f(x_0) = f(x_1) \Rightarrow x_0 = x_1$

**Preimage characterization.** For every $y \in Y$, the preimage $f^{-1}[\{y\}]$ contains at most one element, that is,

$$|f^{-1}[\{y\}]| \leq 1, \quad \forall y \in Y$$
:::

:::{prf:example} An Injective Mapping
:label: ex-injection
The function $f(x) = 2^x + 3$ is injective, because distinct values of $x$ yield distinct values of $f(x)$.

The function $g(x) = x^2$ (defined on the real numbers), by contrast, is not injective, since $g(-1) = g(1) = 1$: distinct inputs can produce the same output. From the preimage point of view, $g^{-1}[\{1\}] = \{-1, 1\}$ contains two elements, violating the preimage characterization of injectivity.
:::


:::{exercise} Restriction of an Injective Mapping
:label: exer-restricted-mapping-injection
Let $f: X \to Y$ be a mapping and let $\emptyset \neq A \subseteq X$ be a nonempty subset of $X$.
If $f$ is injective, prove that $f|_A$ is injective as well.
:::


```{hint} *Testing a Function for Injectivity*
:label: hint-injection
Ways to test a function for injectivity:
1. **Geometric method**: every horizontal line meets the graph of the function at most once
2. **Preimage method**: check whether there is some $y$ with $|f^{-1}[\{y\}]| > 1$
```

:::{prf:definition} Surjective Mapping
:label: def-surjection
A mapping $f: X \rightarrow Y$ is called **surjective** (also said to be **onto**) if for every $y \in Y$ there is at least one $x \in X$ with $f(x) = y$.

**Equivalent formulation.** $\forall y \in Y, \exists x \in X, f(x) = y$

**Preimage characterization.** For every $y \in Y$, the preimage $f^{-1}[\{y\}]$ is nonempty, that is,

$$|f^{-1}[\{y\}]| \geq 1, \quad \forall y \in Y$$
:::

:::{prf:example} A Surjective Mapping
:label: ex-surjection
The function $f(x) = x^3$ (from $\mathbb{R}$ to $\mathbb{R}$) is surjective, because for any $y \in \mathbb{R}$ the equation $x^3 = y$ always has the solution $x = \sqrt[3]{y}$. From the preimage point of view, every $y \in \mathbb{R}$ has preimage $f^{-1}[\{y\}] = \{\sqrt[3]{y}\} \neq \emptyset$.

The function $g(x) = x^2$ (from $\mathbb{R}$ to $\mathbb{R}$), by contrast, is not surjective, since no $x \in \mathbb{R}$ satisfies $x^2 = -1$. From the preimage point of view, $g^{-1}[\{-1\}] = \emptyset$, violating the preimage characterization of surjectivity.
:::

```{hint} *Testing for Surjectivity*
:label: hint-surjection
Ways to test for surjectivity:
1. **Geometric method**: the graph of the function covers the whole codomain (every point on the vertical axis has a corresponding point on the graph)
2. **Preimage method**: check whether there is some $y \in Y$ with $f^{-1}[\{y\}] = \emptyset$
```

:::{prf:definition} Bijective Mapping
:label: def-bijection
A mapping $f: X \rightarrow Y$ that is both injective and surjective is called **bijective**; it is also called an **invertible mapping**, or one-to-one and onto (1-1 & onto).

**Preimage characterization.** For every $y \in Y$, the preimage $f^{-1}[\{y\}]$ contains exactly one element, that is,

$$|f^{-1}[\{y\}]| = 1, \quad \forall y \in Y$$
:::

Characteristic features of a bijection:
* every element of the domain is mapped to a unique element of the range
* every element of the range has exactly one preimage

```{important} *Bijections*
:label: imp-bijection
A bijection is the essential tool for setting up a **one-to-one correspondence** between two sets, and it is the basis for defining **equinumerosity** (having the same number of elements). A mapping $f$ has an inverse mapping if and only if $f$ is bijective. A function that is not bijective has no inverse function.
```

:::{prf:definition} Inverse Mapping
:label: def-inverse

Let $f: X \rightarrow Y$ be a bijection. Then there exists a unique mapping $f^{-1}: Y \rightarrow X$, called the inverse mapping of $f$, satisfying:

for all $x \in X$ and $y \in Y$,

$$f(x) = y \Leftrightarrow f^{-1}(y) = x$$

Basic properties:

$f^{-1}(f(x)) = x$ holds for every $x \in X$

$f(f^{-1}(y)) = y$ holds for every $y \in Y$

**An important distinction.** The inverse mapping $f^{-1}$ and the preimage notation $f^{-1}[\cdot]$ are different notions.
:::

:::{prf:example} A Bijection and Its Inverse Function
:label: ex-bijection-inverse
The function $f(x) = 2x + 3$ (from $\mathbb{R}$ to $\mathbb{R}$) is bijective:

* it is injective: if $2x_0 + 3 = 2x_1 + 3$, then $x_0 = x_1$
* it is surjective: for any $y \in \mathbb{R}$ there exists $x = f^{-1}(y) = \frac{y-3}{2} \in \mathbb{R}$ with $f(x) = y$
* from the preimage point of view: for any $y \in \mathbb{R}$, the preimage $f^{-1}[\{y\}] = \{\frac{y-3}{2}\}$ contains exactly one element

It is easy to check that $f^{-1}(f(x)) = f^{-1}(2x+3) = \frac{(2x+3)-3}{2} = x$ and
$f(f^{-1}(y)) = f(\frac{y-3}{2}) = 2\cdot\frac{y-3}{2}+3 = y-3+3 = y$
:::

#### A Unified Theory of Preimages and Types of Mappings

We have now understood the types of mappings from two angles, the traditional definitions and the preimage characterizations. In fact these two formulations reveal a deeper unity in the theory of mappings:

:::{prf:property} Characterization of Mapping Types by Preimages
:label: pro-preimage-characterization

Let $f: X \rightarrow Y$ be a mapping. Then:

1. **Test for injectivity**: $f$ is injective $\Leftrightarrow$ $|f^{-1}[\{y\}]| \leq 1$ for all $y \in Y$
2. **Test for surjectivity**: $f$ is surjective $\Leftrightarrow$ $f^{-1}[\{y\}] \neq \emptyset$ for all $y \in Y$
3. **Test for bijectivity**: $f$ is bijective $\Leftrightarrow$ $|f^{-1}[\{y\}]| = 1$ for all $y \in Y$
:::

```{note} Geometric Intuition for the Distribution of Preimages
:label: note-preimage-distribution

**How the distribution of preimages relates to the properties of a mapping:**

1. **The image is too concentrated (some elements have no preimage)** → **not surjective**
   - The image is concentrated in part of the codomain, and some elements of the codomain are never "hit"
   - There exists $y \in Y$ with $f^{-1}[\{y\}] = \emptyset$

2. **Images overlap** → **not injective**
   - Several elements of the domain are mapped to the same element of the range
   - There exists $y \in Y$ with $|f^{-1}[\{y\}]| > 1$

3. **The preimages are perfectly distributed** → **bijective**
   - The perfect balance of a one-to-one correspondence
   - $|f^{-1}[\{y\}]| = 1$ for all $y \in Y$


```

````python
# This code block is not shown in the printed book; only the figure is displayed

import numpy as np
import plotly.graph_objects as go

np.set_printoptions(precision=4, suppress=True, linewidth=100)

# --- Color palette: NYU Official (Accent Mix) ---
C_BG   = "#F8F8F8"   # canvas background
C_GRID = "#D6D6D6"   # gridlines (used by plotly)
C_V1   = "#57068C"   # NYU Violet (spare)
C_V2   = "#006385"   # domain nodes (Bleecker Deep Blue)
C_T1   = "#2AD2C9"   # mapping arrows (Mercer Teal)
C_WARN = "#FF5D47"   # codomain nodes (Hudson Orange)
C_AUX  = "#AB82C5"   # dropdown background (Mayfair Light Violet)
C_AREA = "rgba(87, 6, 140, 0.08)"   # set-boundary fill (transparent NYU Violet)

# --- Visual hierarchy helpers ---
def print_header(title):
    print("=" * 60)
    print(f"  {title}")
    print("=" * 60)

def print_step(step, desc):
    print(f"\n▶ Step {step}: {desc}")
    print("-" * 40)

def compare_print(label, actual, expected_desc):
    print(f"\n[{label}]")
    print(f"  Computed value: {actual}")
    print(f"  Expected value: {expected_desc}")

# ============================================================
print_header("Example 1.2 | Interactive Figure of Mapping Types")
# ============================================================

# --- Step 0: define the data for each type of mapping ---
print_step(0, "Define five types of mapping (domain X, codomain Y, rule f)")

FUNCTION_TYPES = {
    "Injective": {
        "domain":   [1, 2, 3, 4],
        "codomain": ['a', 'b', 'c', 'd', 'e'],
        "mapping":  {1: 'a', 2: 'b', 3: 'c', 4: 'd'},
        "desc": "One-to-one: distinct x are sent to distinct f(x), and some codomain elements are never hit",
    },
    "Surjective": {
        "domain":   [1, 2, 3, 4, 5],
        "codomain": ['a', 'b', 'c', 'd'],
        "mapping":  {1: 'a', 2: 'b', 3: 'c', 4: 'd', 5: 'a'},
        "desc": "Onto: every codomain element has a preimage; many-to-one is allowed",
    },
    "Bijective": {
        "domain":   [1, 2, 3, 4],
        "codomain": ['a', 'b', 'c', 'd'],
        "mapping":  {1: 'a', 2: 'b', 3: 'c', 4: 'd'},
        "desc": "One-to-one correspondence: injective and surjective, so the inverse mapping f⁻¹ exists",
    },
    "General mapping": {
        "domain":   [1, 2, 3, 4],
        "codomain": ['a', 'b', 'c', 'd', 'e'],
        "mapping":  {1: 'a', 2: 'a', 3: 'b', 4: 'c'},
        "desc": "Many-to-one and not onto: images repeat, and some codomain elements have no preimage",
    },
    "Constant mapping": {
        "domain":   [1, 2, 3, 4],
        "codomain": ['a', 'b', 'c'],
        "mapping":  {1: 'b', 2: 'b', 3: 'b', 4: 'b'},
        "desc": "Many-to-one in the extreme: every element is sent to the same image",
    },
}

# --- Step 1: classify the mapping by the size of the preimages f⁻¹[{y}] ---
print_step(1, "Classify the mapping by the size of the preimages f⁻¹[{y}]")

def analyze_mapping(domain, codomain, mapping):
    """
    Characterize the mapping by the size of the preimages f⁻¹[{y}]:
      injective   ⟺  |f⁻¹[{y}]| ≤ 1 for all y ∈ Y
      surjective  ⟺  f⁻¹[{y}] ≠ ∅ for all y ∈ Y
      bijective   ⟺  |f⁻¹[{y}]| = 1 for all y ∈ Y
    """
    preimage = {y: [] for y in codomain}
    for x, y in mapping.items():
        preimage[y].append(x)

    sizes         = [len(preimage[y]) for y in codomain]
    is_injective  = all(s <= 1 for s in sizes)
    is_surjective = all(s >= 1 for s in sizes)

    return {
        "is_injective":  is_injective,
        "is_surjective": is_surjective,
        "is_bijective":  is_injective and is_surjective,
        "preimage":      preimage,
        "empty_preimage": [y for y in codomain if len(preimage[y]) == 0],
        "multi_preimage": [y for y in codomain if len(preimage[y]) > 1],
    }

for name, fdata in FUNCTION_TYPES.items():
    result = analyze_mapping(fdata["domain"], fdata["codomain"], fdata["mapping"])
    print(f"\n  {name}")
    print(f"    {fdata['desc']}")
    compare_print(
        "Mapping properties",
        f"injective={result['is_injective']}, surjective={result['is_surjective']}, bijective={result['is_bijective']}",
        "Determined by the size of the preimages (see property pro-preimage-characterization)"
    )
    if result["empty_preimage"]:
        print(f"    elements with no preimage (surjectivity fails): {result['empty_preimage']}")
    if result["multi_preimage"]:
        print(f"    elements hit more than once (injectivity fails): {result['multi_preimage']}")

# --- Step 2: build the interactive Plotly figure ---
print_step(2, "Build an interactive figure that switches between mapping types (Plotly)")

def _add_mapping_traces(fig, fdata, visible):
    """Add the domain, the codomain and the mapping arrows for one type of mapping."""
    domain_elems   = fdata["domain"]
    codomain_elems = fdata["codomain"]
    mapping        = fdata["mapping"]
    desc           = fdata["desc"]

    dy = np.linspace(1, len(domain_elems),   len(domain_elems))
    cy = np.linspace(1, len(codomain_elems), len(codomain_elems))

    # domain nodes (C_V2 Bleecker Deep Blue)
    fig.add_trace(go.Scatter(
        x=[1] * len(domain_elems), y=dy,
        mode='markers+text',
        marker=dict(size=30, color=C_V2, line=dict(width=2, color=C_V1)),
        text=[str(e) for e in domain_elems],
        textposition="middle left",
        textfont=dict(size=16, color='white', family='Arial Black'),
        name='Domain X',
        hovertemplate="x = %{text}<br>" + desc + "<extra></extra>",
        visible=visible,
    ))

    # codomain nodes (C_WARN Hudson Orange)
    fig.add_trace(go.Scatter(
        x=[5] * len(codomain_elems), y=cy,
        mode='markers+text',
        marker=dict(size=30, color=C_WARN, line=dict(width=2, color="#C0392B")),
        text=[str(e) for e in codomain_elems],
        textposition="middle right",
        textfont=dict(size=16, color='white', family='Arial Black'),
        name='Codomain Y',
        hovertemplate="y = %{text}<br>" + desc + "<extra></extra>",
        visible=visible,
    ))

    # mapping arrows (C_T1 Mercer Teal)
    for x_elem, y_elem in mapping.items():
        xi = domain_elems.index(x_elem)
        yi = codomain_elems.index(y_elem)
        sx, sy = 1.25, dy[xi]
        ex, ey = 4.75, cy[yi]
        fig.add_trace(go.Scatter(
            x=[sx, ex], y=[sy, ey],
            mode='lines',
            line=dict(color=C_T1, width=3),
            hovertemplate=f"f({x_elem}) = {y_elem}<extra></extra>",
            showlegend=False, visible=visible,
        ))

def build_interactive_figure():
    fig = go.Figure()
    names        = list(FUNCTION_TYPES.keys())
    trace_counts = []

    for i, (name, fdata) in enumerate(FUNCTION_TYPES.items()):
        n_before = len(fig.data)
        _add_mapping_traces(fig, fdata, visible=(i == 0))
        trace_counts.append(len(fig.data) - n_before)

    # dropdown menu
    buttons = []
    for i, name in enumerate(names):
        vis   = [False] * len(fig.data)
        start = sum(trace_counts[:i])
        for j in range(trace_counts[i]):
            vis[start + j] = True
        buttons.append(dict(
            label=name,
            method="update",
            args=[{"visible": vis}, {"title": f"Type of mapping: {name}"}],
        ))

    # set boundaries (C_V2 domain, C_WARN codomain)
    for x0, x1, color in [(0.3, 1.7, C_V2), (4.3, 5.7, C_WARN)]:
        fig.add_shape(type="rect", x0=x0, y0=0.5, x1=x1, y1=5.5,
                      line=dict(color=color, width=3),
                      fillcolor=C_AREA)

    fig.add_annotation(x=1, y=6, text="Domain X", showarrow=False,
                       font=dict(size=18, color=C_V2, family="Arial Black"))
    fig.add_annotation(x=5, y=6, text="Codomain Y", showarrow=False,
                       font=dict(size=18, color=C_WARN, family="Arial Black"))

    fig.update_layout(
        title="Type of mapping: Injective",
        updatemenus=[dict(
            buttons=buttons, direction="down",
            showactive=True, x=0.02, xanchor="left", y=0.98, yanchor="top",
            bgcolor=C_AUX, bordercolor=C_V1, borderwidth=2,
        )],
        xaxis=dict(range=[0, 6], showgrid=False, showticklabels=False, zeroline=False),
        yaxis=dict(range=[0, 7], showgrid=False, showticklabels=False, zeroline=False),
        showlegend=False, width=800, height=600,
        plot_bgcolor=C_BG, paper_bgcolor=C_BG,
        margin=dict(t=100, b=50, l=50, r=50),
    )
    return fig

# --- Step 3: show the interactive figure ---
print_step(3, "Show the interactive figure (use the dropdown at the top left to switch types)")
fig = build_interactive_figure()
fig.show()

print("\n✓ Done. The dropdown switches between: Injective, Surjective, Bijective, General mapping, Constant mapping.")
````

:class: tip
This section answered a question that runs through the whole book: how should the "correspondence" between two sets be defined precisely, and what structurally different types does such a correspondence come in?

The core constraint on a mapping—each element of the domain has one and only one image—looks like a technical detail, but it rules out the ambiguous situation of "one input with several outputs," and so makes "computing" a well-defined operation. The associativity of composition $g \circ f$ is the set-theoretic root of the associativity of matrix multiplication: in §3, composition of linear mappings corresponds exactly to multiplication of matrices. The importance of the preimage $f^{-1}[B]$ lies in its being meaningful for every mapping, with no requirement that $f$ be invertible—which lets us describe the solution structure of linear systems uniformly in §6 by saying "the solution set is a certain preimage," and lets us define eigenspaces as particular preimages in §8.

The threefold classification of mappings—injective, surjective, bijective—receives a unified characterization through the size of preimage sets (at most one, nonempty, exactly one), revealing the essential difference between a mapping that "compresses information," one that "fails to cover the whole codomain," and one that "establishes a one-to-one correspondence." These three notions reappear in algebraic guise in §4: injectivity corresponds to a kernel containing only the zero vector, surjectivity to an image equal to the codomain, and bijectivity to invertibility of the linear mapping—and there the language of sets and the language of algebra converge into one.

## 1.3 Chapter Summary

### Review of the Theoretical Thread

This chapter set out from a historical turning point: the nineteenth-century crisis in mathematics forced mathematicians to give up their reliance on geometric intuition and to seek a language capable of describing "an entire class of objects" precisely. Weierstrass's pathological function and Dedekind's cuts pointed jointly to the same answer—set theory.

§1.1 established the vocabulary of that language. The three properties of a set (definiteness, distinctness, unorderedness) are not fussy stipulations but the minimum required for "a class of objects" to be referred to unambiguously. The subset relation introduces the logic of whole and part; the power set gathers "all possible substructures" into a mathematical object in its own right. The Cartesian product $A \times B$ is the most crucial construction of all: it makes the pairing relation between two sets concrete as a set of ordered pairs, laying the structural groundwork for describing "correspondence" in the next step. Union, intersection, complement, and set difference constitute the algebra of sets, allowing us to manipulate and combine these objects—and it is exactly this capacity for manipulation that makes every later notion of "space" amenable to rigorous treatment.

§1.2 introduced mappings on top of the language of sets, answering the question of what structural connections can be established between two sets. Distinguishing the domain, codomain, image, and range of a mapping removes the confusion between "which values a function is allowed to take" and "which values it actually takes"; the associativity of composition then reveals the algebraic structure of the operation of composing mappings. The introduction of the preimage is especially profound: $f^{-1}[B]$ is defined without requiring $f$ to be invertible, so that the question "given an output, what were the inputs?" can be posed in complete generality. Finally, taking the size of preimage sets as a unifying viewpoint, the three classes of mappings—injective ($|f^{-1}[\{y\}]| \leq 1$), surjective ($f^{-1}[\{y\}] \neq \emptyset$), and bijective ($|f^{-1}[\{y\}]| = 1$)—acquire an internally consistent characterization rather than three unrelated definitions.

### Connections to Other Chapters

This is the first chapter of the book, and it supplies the common linguistic foundation for everything that follows. The most direct bridges appear in §2 on vectors and §3 on linear transformations: a vector space is first of all a set, equipped with two operations satisfying particular axioms; a linear transformation is first of all a mapping between sets, together with the requirement that it preserve linear structure. Without the precise definition of "set" given in §1.1, the sentence "a vector space is a set" would be a metaphor rather than a mathematical statement.

The classification of mappings from §1.2 reappears in algebraic guise in §4: a linear mapping is injective exactly when its kernel contains only the zero vector, surjective exactly when its image equals the codomain, and bijective exactly when it is invertible—and the essence of the rank-nullity theorem is precisely an algebraic characterization of these three classes of mappings in finite-dimensional spaces. The notion of a preimage then takes bodily form in §6 as the structure of the solutions of a linear system: the solution set of the equation $\mathbf{A}\mathbf{x} = \mathbf{b}$ is exactly the preimage of $\{\mathbf{b}\}$ under the mapping $\mathbf{x} \mapsto \mathbf{A}\mathbf{x}$. In §8, the definition of an eigenspace—the set of all vectors satisfying $\mathbf{A}\mathbf{v} = \lambda\mathbf{v}$—is likewise a special case of a preimage. The influence of the language of sets runs through the entire book.

### The Role of This Chapter in the Book

The problem this chapter solves is that of supplying the whole book with an unambiguous way of speaking. It introduces no content belonging to "linear algebra" as such, and yet it is what makes every later statement of linear algebra possible. "A vector space is a set equipped with two operations satisfying eight axioms"—once every word of that sentence has a precise meaning, the theorems proved from it apply simultaneously to $\mathbb{R}^n$, to polynomial spaces, to function spaces, and to matrix spaces, with no need to redo the derivation for each kind of object. This kind of once-and-for-all general argument is exactly where "discussing an entire class of objects" outstrips "computing individual examples."

This chapter also opens a deeper question that runs through the whole book. At the level of sets, injectivity, surjectivity, and bijectivity characterize the preservation and the loss of information; at the level of linear algebra, the same question will be made concrete through algebraic quantities such as rank, kernel, and invertibility. A reader who carries a set-theoretic eye into the following chapters will recognize familiar structures inside every new definition: a subspace is a subset of a vector space, linear dependence is a constraint on linear-combination relations among sets, and an eigenspace is an algebraic instance of a preimage. Formal language is not an ornament on mathematics; it is the infrastructure that allows abstract reasoning to be transmitted precisely.
