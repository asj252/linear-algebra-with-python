The word "vector" goes back to a name Hamilton gave in 1846 while organizing his theory of quaternions. He borrowed a Latin root meaning "carrier" to refer to the imaginary components that represent a geometric displacement in three-dimensional space. Although Hamilton gave the thing its name, he held to the end of his life that three-dimensional vectors could never exist apart from the quaternions that contained them. History, however, took exactly the opposite road: this geometric component, once attached to the quaternion system, went on to become the basic language of all of linear algebra.

The geometric intuition behind vectors, however, is two thousand years older than the name.

Around the fourth to third century BCE, the Aristotelian *Mechanical Problems* already observed that when an object takes part in two straight-line motions in different directions at once, its actual path is the diagonal of the parallelogram formed by the directions and distances of those two motions. The geometric intuition is correct, but it is in essence a kinematic composition based on the theory of proportions: it is not restricted to right angles, and it does not yet have the dynamical notion of "force." In the first century CE, Hero of Alexandria used the construction of a parallelogram more explicitly to analyze compound motion and the decomposition of collisions—but this was still purely the language of geometric line segments, with no algebraic structure at all.

The real turning point came in 1586. The Dutch mathematician Simon Stevin used a thought experiment with a closed chain of beads hung over a double inclined plane to turn "force" from a vague experience of pushing into a line-segment length that could be quantified and operated on geometrically—the first time a physical quantity and a geometric operation were bound together rigorously. Half a century later, Galileo decomposed projectile motion into two independent components, uniform horizontal motion and accelerated vertical motion, which is in essence an orthogonal decomposition of vectors; he simply did not yet have the term. In 1687, in Corollary I of the *Principia*, Newton formally established the parallelogram law of forces as a foundational corollary of the entire system of classical mechanics—a universal, axiomatized construction, though its derivations still relied purely on the language of Euclidean geometry rather than on algebraic symbols.

From Aristotle to Newton, what accumulated over two thousand years was geometric and physical intuition, not a general algebraic tool.

The algebraic tool arrived from an entirely different direction. In 1843, Hamilton carved the quaternion multiplication formula into Brougham Bridge in Dublin (that is Chapter 3's story). Seeking an algebraic system for three-dimensional space in which division is possible, he wrote a quaternion as a real scalar plus a combination of three imaginary directions. In 1846 he named the part with the three imaginary directions the **vector**—from the Latin *vehere*; the word means "carrier" or "bearer," since a vector "carries" a point in space from the origin to its destination. Hamilton coined the term, but what he cared about most was always the quaternion as a whole, complete under multiplication and division; in his eyes the vector could never be separated out on its own.

Forty years later, this component, bound up within the quaternion system, staged a revolt.

In the 1880s the American physicist Josiah Willard Gibbs and the British engineer Oliver Heaviside independently took the quaternion apart: they kept vector addition, "brutally" split quaternion multiplication, once a seamless whole, into two independent operations—the scalar part, with its sign reversed, became the dot product (the inner product), and the vector part became the cross product—and then threw away entirely the quaternion framework that had held them together. Their reason was thoroughly practical: for a physicist, being able to solve problems simply matters more than preserving the closure of an algebraic structure. Both of the symbols "$\cdot$" and "$\times$" were introduced by Gibbs at this time.

This set off a fierce controversy, the "Great Vector Debate." Hamilton's loyalists, led by Peter Tait, denounced it as a desecration of the master's legacy: they charged that the cross product obtained by taking things apart was a defective pseudovector (its sign behaves abnormally under a reflection of coordinates), and they condemned Gibbs's system as a "monstrosity" that destroyed the completeness of the algebra. Gibbs and Heaviside retorted that physicists need handy geometric tools, not the obscure and cumbersome baggage of quaternions carried for the sake of mathematicians' algebraic purity.

The argument ran for nearly twenty years. In 1901 Gibbs's student E. B. Wilson compiled his lecture notes and published them as *Vector Analysis*, formally fixing the standard form of vectors found in every physics and engineering textbook today—addition, scalar multiplication, dot product, cross product, together with the symbol system we now know so well. Hamilton's quaternions did not disappear; they are active today in computer graphics and robotics, where they describe three-dimensional rotations precisely and without any risk of gimbal lock. But the winner of that argument over a common language for physics was, in the end, Gibbs.

The geometric intuition behind vectors has a two-thousand-year history; the word "vector" has less than two hundred years; and the dot product and the cross product were settled as independent algebraic operations even later than matrix multiplication (1858). The tools we take for granted on the blackboard today are, in essence, the spoils left behind by a great collision of mathematical and physical thought.

## 2.1 Introduction to Vectors

In the geometric formulation, a **vector** has both a magnitude (a length) and a direction; in the algebraic formulation, a vector is usually a list of numbers arranged in order. Vectors can represent displacement, velocity, and force in physics, and they can equally represent a color (an RGB color, say). In the sections that follow we introduce the basic notions and operations of vectors a few at a time.

:::{prf:definition} Two-Dimensional Vectors
:label: 2d-vector-def

$$
\mathbf{v} = \overrightarrow{\mathbf{ov}} = \vec{v} =  | \mathbf{v} \rangle  = \begin{bmatrix} x \\ y \end{bmatrix}
$$

This denotes the arrow pointing from the origin $\mathbf{o}=(0,0)$ to the point $(x,y)$. We shall routinely treat the point $(x,y)$ as equivalent to the vector $\mathbf{v}$. From Chapter 3 onward we shall see that $\mathbf{v}$ is by default a column vector. The line above lists several equivalent ways of denoting a vector; the last is the notation of quantum mechanics, which is perfectly usable for real vectors as well.
:::

:::{prf:definition} Three-Dimensional Vectors
:label: 3d-vector-def

$$
\mathbf{v} = \begin{bmatrix} x \\ y \\ z \end{bmatrix}
$$

This denotes the arrow pointing from the origin $(0,0,0)$ to the point $(x,y,z)$. We shall routinely treat the point $(x,y,z)$ as equivalent to the vector $\mathbf{v}$.
:::

Vectors of higher dimension follow the same pattern. In general, we define the space in which $n$-dimensional real vectors live as the $n$-fold Cartesian product of the set of real numbers.

:::{prf:definition} The $n$-Dimensional Real Vector Space $\mathbb{R}^n$
:label: def-rn-space

$$\mathbb{R}^n = \underbrace{\mathbb{R}\times\mathbb{R}\times\cdots\times\mathbb{R}}_{n \text{ copies}} = \{(x_0, x_1, \dots, x_{n-1}) \mid x_i \in \mathbb{R}\}$$

That is, the set of all ordered tuples of $n$ real numbers, each element $\mathbf{v} = (x_0, x_1, \dots, x_{n-1})$ of which is an $n$-dimensional vector. This definition carries over the Cartesian product of {prf:ref}`def-cartesian-product` from Chapter 1, and it makes good on the preview of higher-dimensional coordinate spaces given in {ref}`exer-cartesian-space`. Everything said about $\mathbb{R}^n$ in the rest of the book (the standard basis, the inner product, the exercises in $\mathbb{R}^4$, and so on) rests on this definition.
:::

:::{prf:example} Vectors in Practice
:label: ex-vector-applications
  - **Geometric vectors:** arrows in a figure indicating a direction and a length.
  - **Force vectors in physics:** the magnitude and direction of a force.
  - **Color vectors:** an RGB color may be written as $(r,\, g,\, b)$.
:::

## 2.2 Linear Operations on Vectors

The most basic operations between vectors are **addition** and **scalar multiplication**. These operations constitute the foundation of linear algebra.

:::{prf:definition} Vector Addition
:label: vector-addition

For two vectors

$$
\mathbf{u} = \begin{bmatrix} u_0 \\ u_1 \end{bmatrix}, \quad \mathbf{v} = \begin{bmatrix} v_0 \\ v_1 \end{bmatrix}
$$

their sum is defined as

$$
\mathbf{u} + \mathbf{v} = \begin{bmatrix} u_0+v_0 \\ u_1+v_1 \end{bmatrix}
$$

The three-dimensional and higher-dimensional cases are analogous.
:::

```{note} Geometric Interpretation
:label: note-vector-addition-geometry
**Geometric interpretation.** The parallelogram law: place the two vectors so that their initial points coincide; the diagonal of the parallelogram they form is the vector sum. This displays the commutativity of vector addition directly.
```

:::{prf:example} Computing a Vector Sum
:label: vector-add-example

If

$$
\mathbf{a} = \begin{bmatrix} 2 \\ 3 \end{bmatrix}, \quad \mathbf{b} = \begin{bmatrix} 4 \\ -1 \end{bmatrix}
$$

then

$$
\mathbf{a} + \mathbf{b} = \begin{bmatrix} 2+4 \\ 3+(-1) \end{bmatrix} = \begin{bmatrix} 6 \\ 2 \end{bmatrix}
$$
:::

:::{prf:definition} Scalar Multiplication of Vectors
:label: def-scalar-multiplication

Given a vector

$$
\mathbf{u} = \begin{bmatrix} u_0 \\ u_1 \end{bmatrix} \quad \text{and a scalar} \; \lambda
$$

the result of scalar multiplication is

$$
\lambda\,\mathbf{u} = \begin{bmatrix} \lambda u_0 \\ \lambda u_1 \end{bmatrix}
$$
:::

```{note}
:label: note-scalar-multiplication-geometry
**Geometric interpretation.**
- When $\lambda > 0$: the direction is unchanged and the length is scaled proportionally
- When $\lambda < 0$: the direction is reversed
- When $\lambda = 0$: the result is the zero vector
```

:::{prf:example} Computing a Scalar Multiple
:label: ex-scalar-multiplication
If $\lambda = -3$ and

$$
\mathbf{u} = \begin{bmatrix} 2 \\ 5 \end{bmatrix}
$$

then

$$
-3\,\mathbf{u} = \begin{bmatrix} -6 \\ -15 \end{bmatrix}
$$
:::

````python
# This code block is not shown in the printed book; only the figure is displayed
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True, linewidth=100)

# --- Visual hierarchy helpers ---
def print_header(title):
    print("=" * 60)
    print(f"  {title}")
    print("=" * 60)

def print_step(step, desc):
    print(f"\n▶ Step {step}: {desc}")
    print("-" * 40)

def compare_print(label, actual, expected_desc):
    print(f"\n[{label}]")
    print(f"  Computed value: {actual}")
    print(f"  Expected value: {expected_desc}")

# Fonts: the English edition needs no CJK font, so nothing is downloaded when _lang == 'en';
# the Chinese editions use this same block to fetch Noto Sans TC/SC where no CJK font is installed
import os, urllib.request
import matplotlib.font_manager as fm
_lang = 'en'
_cjk = ['Microsoft JhengHei', 'PingFang TC', 'Noto Sans CJK TC', 'Noto Sans TC']
_have = {f.name for f in fm.fontManager.ttflist}
if _lang != 'en' and not _have & set(_cjk):
    _font = os.path.join(os.path.expanduser('~'), '.cache', 'fonts', 'NotoSansTC.ttf')
    try:
        if not os.path.exists(_font):
            os.makedirs(os.path.dirname(_font), exist_ok=True)
            urllib.request.urlretrieve('https://github.com/google/fonts/raw/main/ofl/notosanstc/NotoSansTC%5Bwght%5D.ttf', _font + '.part')
            os.replace(_font + '.part', _font)
        fm.fontManager.addfont(_font)
        _have.add('Noto Sans TC')
    except OSError as err:
        print('Could not download the CJK font; Chinese text in figures may not display:', err)
plt.rcParams['font.family'] = [f for f in _cjk if f in _have] + ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# --- Color palette (Accent Mix) ---
C_BG   = "#F8F8F8"
C_GRID = "#D6D6D6"
C_AXIS = "#000000"
C_V1   = "#57068C"   # 2v, scaled up (violet)
C_V2   = "#006385"   # v, the original vector (deep blue)
C_T1   = "#2AD2C9"   # 0.5v, scaled down (teal)
C_WARN = "#FF5D47"   # −v, reversed (orange)
C_AUX  = "#AB82C5"   # dashed reference vector (light violet)

# ============================================================
print_header("Example 2.2 | Geometric Picture of Scalar Multiplication")
# ============================================================

# --- Step 0: define the original vector and the list of scalars ---
v       = np.array([2, 3])
scalars = [0.5,  1.0,  2.0,   -1.0]
colors  = [C_T1, C_V2, C_V1,   C_WARN]
labels  = ['0.5v (scaled down)', 'v (original)', '2v (scaled up)', '−v (reversed)']

print_step(0, "Define the vector v and the scalars")
print(f"  original vector v = {v}")

# --- Step 1: verify ‖λv‖ = |λ|·‖v‖ ---
print_step(1, "Verify the length formula for scalar multiplication: ‖λv‖ = |λ|·‖v‖")
norm_v = np.linalg.norm(v)
for lam in scalars:
    lv = lam * v
    compare_print(
        f"λ = {lam:+.1f}",
        f"‖λv‖ = {np.linalg.norm(lv):.4f}",
        f"|λ|·‖v‖ = {abs(lam) * norm_v:.4f}"
    )

# --- Step 2: set up a 2×2 grid of subplots ---
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes_flat = axes.flatten()

# set the axis limits of each subplot first, then compute data_range
for ax in axes_flat:
    ax.set_xlim(-5, 5)
    ax.set_ylim(-7, 7)

ranges     = [a.get_xlim()[1] - a.get_xlim()[0] for a in axes_flat]
data_range = sum(ranges) / len(ranges)   # xlim(-5,5) → 10
kw = dict(angles='xy', scale_units='xy', scale=1, units='xy',
          width=data_range * 0.012)

# --- Step 3: draw the scalar multiples, one per subplot ---
for i, (lam, color, label) in enumerate(zip(scalars, colors, labels)):
    ax      = axes_flat[i]
    scaled_v = lam * v

    ax.set_facecolor(C_BG)
    ax.grid(True, color=C_GRID, linewidth=0.8, alpha=0.7)
    ax.set_aspect('equal', adjustable='box')
    ax.set_box_aspect(1)
    # reference vector v (shown in the auxiliary color when λ≠1)
    if lam != 1.0:
        ax.quiver(0, 0, v[0], v[1], color=C_AUX, alpha=0.5, **kw,
                  label=r'$\mathbf{v}$ (reference)')

    # λv, the principal vector
    ax.quiver(0, 0, scaled_v[0], scaled_v[1], color=color, **kw,
              label=rf'${lam}\mathbf{{v}}$')

    # label at the tip of the vector
    off_x = 0.3 if scaled_v[0] >= 0 else -0.8
    off_y = 0.3 if scaled_v[1] >= 0 else -1.0
    ax.text(scaled_v[0] + off_x, scaled_v[1] + off_y,
            f'{lam}v', fontsize=13, fontweight='bold', color=color,
            bbox=dict(boxstyle="round,pad=0.2", facecolor='white', alpha=0.8))

    # length information
    length = np.linalg.norm(scaled_v)
    ax.text(0.02, 0.98, f'‖λv‖ = {length:.2f}',
            transform=ax.transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

    ax.plot(0, 0, 'o', color=C_AXIS, markersize=7)
    ax.set_xlabel("$x_0$", fontsize=12, color=C_AXIS)
    ax.set_ylabel("$x_1$", fontsize=12, color=C_AXIS)
    ax.set_title(label, fontweight='bold', color=color, fontsize=14)
    ax.legend(fontsize=11, loc='lower right')

fig.suptitle('Geometric Picture of Scalar Multiplication', fontsize=18, fontweight='bold',
             color=C_AXIS, y=0.995)
plt.tight_layout()
plt.show()
````

### 2.2.1 Linear Combinations

:::{prf:definition} Linear Combination
:label: def-linear-combination

A new vector obtained by *taking a weighted sum of several vectors* is called their **linear combination**. (Note that the vectors entering a linear combination must live in the same space, that is, they must have the same dimension.) For example,

$$
\lambda_0\mathbf{u}+\lambda_1\mathbf{v}
$$
is a linear combination of $\mathbf{u}$ and $\mathbf{v}$, where $\lambda_0$ and $\lambda_1$ are scalars.
:::

:::{prf:example} Mixing RGB Colors
:label: ex-rgb-color-mixing
Weighting red and green in various proportions and mixing them produces yellow. This is an intuitive instance of a linear combination.
:::

```{note} A Preview of Two Related Notions: Span and Linear Independence
:label: note-span-independence-preview
Linear combinations lead naturally to two notions that run through the entire book. We give the intuition here and leave the rigorous definitions to Chapter 4.

Given a set of vectors, the collection of **all** of their linear combinations is called the **span** of that set. For example, given two non-parallel vectors $\mathbf{u}$ and $\mathbf{v}$ in the plane, their linear combinations $\lambda_0\mathbf{u}+\lambda_1\mathbf{v}$ range over the whole plane, and we say that $\{\mathbf{u},\mathbf{v}\}$ spans $\mathbb{R}^2$.

If no vector in a set can be written as a linear combination of the others (that is, if they are mutually non-redundant), the set is called **linearly independent**; otherwise it is **linearly dependent**. Two vectors in the plane are linearly independent exactly when, geometrically, they are not collinear. These two terms are used with this intuitive meaning in the remainder of the chapter and in the exercises; the rigorous definitions appear in Chapter 4.
```

Linear combinations can be applied to define line segments in vector form.

:::{prf:definition} Parametric Representation of a Line Segment
:label: def-line-segment
The **line segment** $\overrightarrow{\mathbf{ab}}$ joining two points $\mathbf{a}$ and $\mathbf{b}$ can be expressed in parametric form. The vector of any point $\mathbf{p}$ on the segment can be written as

$$
\mathbf{p} = (1-t)\mathbf{a} + t\mathbf{b}
$$

where $t \in [0,1]$ is the parameter:
- when $t = 0$, $\mathbf{p} = \mathbf{a}$ (the initial point)
- when $t = 1$, $\mathbf{p} = \mathbf{b}$ (the terminal point)
- when $t = 0.5$, $\mathbf{p}$ is the *midpoint* of the segment $\mathbf{ab}$

This is an important application of the **linear combination**, in which the weights $(1-t)$ and $t$ sum to 1.
:::

### 2.2.2 The Standard Basis

In $\mathbb{R}^n$, the **standard basis** is a particular set of unit vectors that are mutually orthogonal and can represent any vector in the space.

:::{prf:definition} Standard Basis
:label: def-standard-basis

In the $n$-dimensional Euclidean space $\mathbb{R}^n$, the **standard basis** is the set of $n$ unit vectors

$$\{\mathbf{e}_0, \mathbf{e}_1, \ldots, \mathbf{e}_{n-1}\}$$

where the $i$-th standard basis vector $\mathbf{e}_i$ is defined by

$$\mathbf{e}_i = \begin{bmatrix} 0 \\ \vdots \\ 0 \\ 1 \\ 0 \\ \vdots \\ 0 \end{bmatrix} \leftarrow \text{position } i$$

That is, its entry in position $i$ equals 1 ($i=0,\dots,n-1$) and all of its other entries equal 0.
:::

:::{prf:example} The Standard Basis in Two and Three Dimensions
:label: ex-standard-basis-2d-3d

**The two-dimensional standard basis:**

$$\mathbf{e}_0 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad \mathbf{e}_1 = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$

**The three-dimensional standard basis:**

$$\mathbf{e}_0 = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}, \quad \mathbf{e}_1 = \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}, \quad \mathbf{e}_2 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}$$

In three-dimensional geometry these are usually called the unit vectors in the **$x$ direction**, the **$y$ direction**, and the **$z$ direction** respectively.
:::

```{important}
:label: imp-standard-basis-properties
The properties below use the **inner product** $\mathbf{e}_i \cdot \mathbf{e}_j$ and the **length** $\|\mathbf{e}_i\|$, which are not formally defined until §2.3. For now you may read them with the geometric intuition of "mutually perpendicular" and "of length 1," and return to them once §2.3 has supplied the rigorous definitions.

Important properties of the standard basis:

1. **Orthogonality**: any two distinct standard basis vectors are perpendicular, that is, $\mathbf{e}_i \cdot \mathbf{e}_j = 0$ when $i \neq j$
2. **Unit length**: each standard basis vector has length 1, that is, $\|\mathbf{e}_i\| = 1$
3. **Completeness**: every vector can be expressed as a linear combination of the standard basis vectors
4. **Uniqueness**: the representation of each vector in the standard basis is unique
```

:::{prf:property} Representation of a Vector in the Standard Basis
:label: pro-vector-standard-representation

Every vector $\mathbf{v} \in \mathbb{R}^n$ can be expressed uniquely as a linear combination of the standard basis vectors:

$$\mathbf{v} = \begin{bmatrix} v_0 \\ v_1 \\ \vdots \\ v_{n-1} \end{bmatrix} = v_0\mathbf{e}_0 + v_1\mathbf{e}_1 + \cdots + v_{n-1}\mathbf{e}_{n-1} = \sum_{i=0}^{n-1} v_i\mathbf{e}_i$$

Here $v_i$ is called the **component**, or the **coordinate**, of the vector $\mathbf{v}$ in the direction of the $i$-th standard basis vector.
:::

```{note}
:label: note-coordinate-interpretation
**Geometric interpretation.** Each component $v_i$ of a vector gives the "signed component" of that vector along the corresponding coordinate axis. For example, the three-dimensional vector $\mathbf{v} = (3, -2, 5)$ means: starting from the origin, move 3 units in the positive $x$ direction, 2 units in the negative $y$ direction, and 5 units in the positive $z$ direction.
```

:class: tip
This section answered the question of how vectors may be combined with one another, and brought out the geometric meaning and algebraic structure of those combinations. Vector addition rests intuitively on the parallelogram law, while scalar multiplication governs the scaling and the reversal of a vector; together they are called the linear operations, they satisfy commutativity, associativity, and distributivity, and they constitute the concrete prototype of the eight axioms of an abstract vector space in Chapter 4.

Built on top of them, the linear combination $\sum_i \lambda_i \mathbf{v}_i$ provides a systematic way of generating new vectors: varying the coefficients $\lambda_i$ lets us roam through the space. The parametric representation of a line segment, $\mathbf{p}=(1-t)\mathbf{a}+t\mathbf{b}$, is the simplest linear combination, and the centroid of a triangle is the special case of equal weights; both foreshadow the emerging notion of a span. The introduction of the standard basis $\{\mathbf{e}_0, \dots, \mathbf{e}_{n-1}\}$ is more crucial still: it guarantees that every vector in $\mathbb{R}^n$ has a unique representation in components, $\mathbf{v}=\sum_{i=0}^{n-1}v_i\mathbf{e}_i$, tying "decomposition" to "uniqueness." The next section equips this algebraic structure with a metric—the inner product—so that we may speak of lengths, angles, and projections.

## 2.3 The Inner Product and Projection

:::{prf:definition} Inner Product of Vectors
:label: dot-product

For two-dimensional real vectors

$$
\mathbf{u} = \begin{bmatrix} u_0 \\ u_1 \end{bmatrix},\quad \mathbf{v} = \begin{bmatrix} v_0 \\ v_1 \end{bmatrix}
$$

the inner product is

$$
\mathbf{u} \cdot \mathbf{v} = u_0v_0+u_1v_1
$$

The three-dimensional and higher-dimensional real cases are analogous.
:::

:::{prf:example} Computing an Inner Product
:label: ex-dot-product
If

$$
\mathbf{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix},\quad \mathbf{v} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}
$$

then

$$
\mathbf{u}\cdot\mathbf{v}= \begin{bmatrix} 1 & 2 \end{bmatrix}  \begin{bmatrix} 3 \\ 4 \end{bmatrix} =1\times 3+2\times 4=11
$$
:::

```{tip}
:label: tip-dot-product
Mathematics uses several different symbols for the inner product, some of them involving the (conjugate) transpose of a matrix. We shall discuss the inner product on complex vector spaces in Chapter 9; here we restrict attention to real vector spaces and list some equivalent notations for the inner product.

$$\mathbf{u} \cdot \mathbf{v} = \mathbf{u}^{\top} \mathbf{v} =\langle \mathbf{u} , \mathbf{v} \rangle  = \langle \mathbf{u} || \mathbf{v} \rangle = \langle \mathbf{u} | \mathbf{v} \rangle $$

The last of these is the quantum-mechanical notation for the inner product of vectors. It is not common in mathematics, but it is rather clearer when writing by hand.
```

```{tip}
:label: tip-dot-product-physics
The inner product has an important application in physics: the inner product of force and displacement represents **work**, that is, the effective component of the force along the direction of the displacement multiplied by the magnitude of the displacement.
```

:::{prf:definition} The $l_2$ Norm, or Euclidean Norm
:label: l2-norm
For an $n$-dimensional real vector $\mathbf{v} = \begin{bmatrix} v_0 \\ v_1 \\ \vdots \\ v_{n-1} \end{bmatrix}$, the $l_2$ norm is defined as the square root of the inner product of the vector with itself:

$$
\|\mathbf{v}\|_2 = \sqrt{\mathbf{v} \cdot \mathbf{v}}= \sqrt{\langle \mathbf{v} | \mathbf{v} \rangle} = \sqrt{\sum_{i=0}^{n-1} v_i^2}
$$

This norm is also called the Euclidean norm; it represents the length of the vector in Euclidean space. Throughout this book the length of a vector is by default the $l_2$ norm, so that $\|\mathbf{v}\|=  \|\mathbf{v}\|_2 $. In Chapters 9 and 11 we shall further consider norms of matrices, as well as the genuine $l_1$ norm (the sum of the absolute values of the components, $\sum_{i=0}^{n-1}|v_i|$) and the $l_\infty$ norm (the largest of the absolute values of the components).
:::

:::{prf:example} Computing an $l_2$ Norm
:label: ex-l2-norm
If

$$
\mathbf{v} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}
$$
then

$$
\|\mathbf{v}\|_2 = \sqrt{\mathbf{v} \cdot \mathbf{v}} = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5
$$

For the three-dimensional vector

$$
\mathbf{w} = \begin{bmatrix} 1 \\ 2 \\ 2 \end{bmatrix}
$$
we have

$$
\|\mathbf{w}\|_2 = \sqrt{1^2 + 2^2 + 2^2} = \sqrt{1 + 4 + 4} = \sqrt{9} = 3
$$
:::

:::{prf:definition} Normalization of a Vector
:label: vector-normalization
For a nonzero vector $\mathbf{v}$, its normalized vector (also called its unit vector) is defined by

$$
\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|_2}
$$

A normalized vector has the following properties:
1. $\|\hat{\mathbf{v}}\|_2 = 1$ (its length is 1)
2. $\hat{\mathbf{v}}$ points in the same direction as $\mathbf{v}$
:::

:::{prf:example} Computing a Normalized Vector
:label: ex-vector-normalization
If

$$
\mathbf{v} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}
$$
first compute its $l_2$ norm:

$$
\|\mathbf{v}\|_2 = \sqrt{3^2 + 4^2} = 5
$$
The normalized vector is then

$$
\hat{\mathbf{v}} = \frac{1}{5}\begin{bmatrix} 3 \\ 4 \end{bmatrix} = \begin{bmatrix} 0.6 \\ 0.8 \end{bmatrix}
$$

Check: $\|\hat{\mathbf{v}}\|_2 = \sqrt{0.6^2 + 0.8^2} = \sqrt{0.36 + 0.64} = 1$
:::

```{note} The Cosine Formula, the Cauchy-Schwarz Inequality, and the Triangle Inequality
:label: note-inner-product-geometry
In the case of $\mathbb{R}^2$, plane geometry gives us the cosine formula $\mathbf{u}\cdot\mathbf{v}=\|\mathbf{u}\|\|\mathbf{v}\|\cos\theta$, from which it is easy to derive that the $l_2$ norm satisfies the Cauchy-Schwarz inequality and the triangle inequality. For a general real Cartesian coordinate system we shall instead start from the Cauchy-Schwarz inequality, show that the cosine formula is **well-defined**, and then derive the triangle inequality. In Chapter 9 we shall revisit this: for a general Euclidean space, two vectors in a high-dimensional space can be rotated into $\mathbb{R}^2$ with their angle preserved, so the cosine formula in high dimensions may be understood as the very same formula as in two dimensions.
```

:::{prf:theorem} The Cauchy-Schwarz Inequality
:label: thm-cauchy-schwarz

$$
|\mathbf{u}\cdot\mathbf{v}|= |\langle \mathbf{u} |\mathbf{v} \rangle |\le\|\mathbf{u}\| \|\mathbf{v}\|
$$
:::

The Cauchy-Schwarz inequality admits several different proofs. We give one of the simpler ones here.

:::{prf:proof} The Cauchy-Schwarz Inequality
1. If $\mathbf{u}=\mathbf{0}$, the conclusion holds trivially; assume from now on that $\mathbf{u}\neq\mathbf{0}$. In $\mathbb{R}^n$ we have vectors $\mathbf{u}=(u_0,u_1,\dots,u_{n-1})$ and $\mathbf{v}=(v_0,v_1,\dots,v_{n-1})$.
2. Consider the quadratic polynomial in $x$

$$(u_0 x + v_0)^2+(u_1 x + v_1)^2+ \dots + (u_{n-1} x + v_{n-1})^2= \left(\sum_{i=0}^{n-1} u_i^2\right)x^2 +2\left(\sum_{i=0}^{n-1} u_i v_i\right)x + \sum_{i=0}^{n-1} v_i^2 \geq 0$$

3. Since it has at most one real root, its discriminant is less than or equal to 0, so that

$$4\left(\sum_{i=0}^{n-1} u_i v_i\right)^2 - 4 \left(\sum_{i=0}^{n-1} u_i^2\right)\left(\sum_{i=0}^{n-1} v_i^2\right)\leq 0$$

Rearranging gives $|\mathbf{u}\cdot\mathbf{v}|\le\|\mathbf{u}\|\|\mathbf{v}\|$. $\square$
:::

:::{prf:definition} The Angle Between Vectors in Space
:label: def-vector-angle
Consider two nonzero real vectors $\mathbf{u}$ and $\mathbf{v}$ issuing from the origin in space.
By the Cauchy-Schwarz inequality $|\langle \mathbf{u} |\mathbf{v} \rangle |\le\|\mathbf{u}\| \|\mathbf{v}\|$, the quotient $\dfrac{\langle \mathbf{u} |\mathbf{v} \rangle}{\|\mathbf{u}\| \|\mathbf{v}\|}$ necessarily lies in $[-1,1]$, so we may define the cosine formula

$$\cos \theta = \frac{\langle \mathbf{u} |\mathbf{v} \rangle}{\|\mathbf{u}\| \|\mathbf{v}\|}$$

and call $\theta$ (with $0\leq\theta \leq \pi$) the angle between the two vectors. The sign of the inner product reflects the type of angle directly: when $\langle \mathbf{u} |\mathbf{v} \rangle > 0$ we have $0\leq\theta<\pi/2$ ($\theta=0$ meaning the same direction), and when $\langle \mathbf{u} |\mathbf{v} \rangle < 0$ we have $\pi/2<\theta\leq\pi$ ($\theta=\pi$ meaning opposite directions). In particular, when $\cos \theta =0$ (that is, when the inner product vanishes) we write $\mathbf{u} \perp \mathbf{v}$ and say that the two vectors are perpendicular.
:::

:::{prf:example} The Angle Between Two Lines in Space
:label: ex-lines-angle
Consider two lines in space,

$$
\begin{array}{l}
\mathbf{x} =  \mathbf{x}_0 + t\mathbf{a} \\
\mathbf{x} =  \mathbf{x}_1 + t\mathbf{b} \\
\end{array}
$$
They may well fail to intersect, but we may translate them so that they do, without changing their directions. We then call the angle between the two lines $\theta$ (with $0\leq\theta \leq \frac{\pi}{2}$), where

$$\cos \theta= \frac{|\langle \mathbf{a} | \mathbf{b} \rangle|}{\| \mathbf{a} \|\| \mathbf{b} \|}$$
:::

:::{prf:remark} The Difference Between the Angle of Vectors and the Angle of Lines
:label: rem-vector-vs-line-angle
The angle between vectors in {prf:ref}`def-vector-angle` keeps the sign of the inner product and ranges over $[0,\pi]$, so it distinguishes acute from obtuse angles—precisely the intuition needed later for correlation coefficients and for deciding which half-space a point lies in. A line, by contrast, has no direction of its own: the direction vectors $\mathbf{a}$ and $-\mathbf{a}$ describe the same line. So that the angle does not change with the orientation chosen for the direction vector, the angle between lines takes the **absolute value** of the inner product and is restricted to $[0,\frac{\pi}{2}]$, reporting only the smaller of the angles between the two lines.
:::

:::{prf:definition} Projection of a Vector
:label: vector-projection

For $\mathbf{u}\neq\mathbf{0}$, the orthogonal projection of a vector $\mathbf{v}$ onto a vector $\mathbf{u}$ is

$$
\text{Proj}_{\mathbf{u}}(\mathbf{v})=\frac{ \langle \mathbf{u} | \mathbf{v} \rangle}{\langle \mathbf{u} | \mathbf{u} \rangle}| \mathbf{u} \rangle
$$
:::

:::{prf:example} The Distance from a Point to a Line in Space
:label: ex-point-line
Consider a point $\mathbf{x}_0$ in space and a line $l$,

$$
\mathbf{x} =  \mathbf{x}_1 + t\mathbf{a}
$$

Dropping a perpendicular from the point $\mathbf{x}_0$ to the line $l$ gives a foot of the perpendicular at a point $\mathbf{p}$, so that $\mathbf{x}_0$, $\mathbf{x}_1$, and $\mathbf{p}$ form a right triangle.

Moreover,

$$ \mathbf{p} =  \mathbf{x}_1 + \frac{\langle \mathbf{a} | \mathbf{x_0-x_1} \rangle}{\langle \mathbf{a} | \mathbf{a} \rangle}\mathbf{a} $$

and

$$ \| \mathbf{x}_0 - \mathbf{x}_1 \|^2 = \| \mathbf{x}_0 - \mathbf{p}\|^2+ \| \mathbf{p}-\mathbf{x}_1 \|^2 =  \| \mathbf{x}_0 - \mathbf{p}\|^2+  \frac{\langle \mathbf{a} | \mathbf{x_0-x_1} \rangle^2}{\langle \mathbf{a} | \mathbf{a} \rangle}$$

Note that $ \| \mathbf{x}_0 - \mathbf{p}\| $ is the shortest distance from the point $\mathbf{x}_0$ to the line $l$.
:::

:::{prf:example} The Inner Product and Planes
:label: ex-hypersurface
The inner product also gives a convenient way to define the equation of a plane. Consider a unit normal vector $\mathbf{n}$ and the set $\{ \mathbf{x} | \langle \mathbf{n} , \mathbf{x} \rangle= d_0  \}$. One sees readily that this set constitutes a plane $S_0$ at distance $|d_0|$ from the origin, and that the points of this plane all have the same signed component $d_0$ along the direction $\mathbf{n}$.

Consider a point $\mathbf{p}$ off the plane, and suppose $\langle \mathbf{n} , \mathbf{p} \rangle= d_1$. Then $d_1$ may be used to define another plane $S_1$: $\{ \mathbf{x} | \langle \mathbf{n} , \mathbf{x} \rangle= d_1  \}$. Since $\mathbf{n}$ is a unit vector, the signed component of any point $\mathbf{x}$ along the direction $\mathbf{n}$ is exactly $\langle \mathbf{n} , \mathbf{x} \rangle$, so the coordinates of the points of $S_0$ and $S_1$ along the direction $\mathbf{n}$ are fixed at $d_0$ and $d_1$ respectively. The two planes have the same normal vector $\mathbf{n}$ and are therefore parallel, and their shortest distance is the difference of those coordinates along the direction $\mathbf{n}$, namely $|d_0-d_1|$; the distance from $\mathbf{p}\in S_1$ to the plane $S_0$ is measured along $\mathbf{n}$ in the same way and is therefore also $|d_0-d_1|$.
:::

:class: tip
This section answered the question of how distance and direction are to be measured in a vector space. The inner product $\langle \mathbf{u} | \mathbf{v} \rangle = \sum_{i=0}^{n-1} u_i v_i$ is a purely algebraic quantity, yet it conceals a wealth of geometric information: the $l_2$ norm it induces, $\|\mathbf{v}\|_2 = \sqrt{\langle\mathbf{v}|\mathbf{v}\rangle}$, measures the length of a vector, while the Cauchy-Schwarz inequality $|\langle\mathbf{u}|\mathbf{v}\rangle| \le \|\mathbf{u}\|\|\mathbf{v}\|$ guarantees, when both vectors are nonzero, the legitimacy of the definition of angle—it is exactly this inequality that keeps the value of $\cos\theta = \langle\mathbf{u}|\mathbf{v}\rangle/(\|\mathbf{u}\|\|\mathbf{v}\|)$ inside $[-1, 1]$, so that "angle" remains meaningful as a geometric notion in any dimension.

These notions will be generalized to complex inner product spaces in Chapter 9, while the exercise on the polarization identity at the end of this section previews the possibility of recovering the inner product from the norm. ## 2.4 The Cross Product

This chapter discusses the standard three-dimensional cross product. It has wide application in physics, engineering, and computer graphics, and it provides a natural tool for describing rotations, torques, and normal vectors.

:::{prf:definition} The Cross Product of Vectors
:label: cross-product

For three-dimensional vectors

$$
\mathbf{u}=\begin{bmatrix} u_0 \\ u_1 \\ u_2 \end{bmatrix},\quad \mathbf{v}=\begin{bmatrix} v_0 \\ v_1 \\ v_2 \end{bmatrix}
$$

the cross product is

$$
\mathbf{u}\times\mathbf{v}=\begin{bmatrix} u_1v_2-u_2v_1 \\ u_2v_0-u_0v_2 \\ u_0v_1-u_1v_0 \end{bmatrix}
$$

One verifies readily that $(\mathbf{u}\times\mathbf{v}) \perp \mathbf{u}$ and $(\mathbf{u}\times\mathbf{v}) \perp \mathbf{v}$
:::

```{warning} The Dimensional Restriction on the Cross Product
:label: warn-cross-product-dimension
This chapter discusses the standard three-dimensional cross product; its component formula does not carry over directly to arbitrary dimensions. Seven dimensions admit a generalization that requires additional algebraic structure, and in general dimensions one usually turns to tools such as the wedge product instead. The exercises at the end of Chapter 3 contain further practice on the cross product.

**Handling two-dimensional vectors.** Two-dimensional vectors $(u_0,u_1)$ and $(v_0,v_1)$ may be regarded as the three-dimensional vectors $(u_0,u_1,0)$ and $(v_0,v_1,0)$, after which the cross product may be computed. The result always has the form $(0,0,u_0v_1-u_1v_0)$, that is, it points purely along the $z$ axis.

For this reason the "cross product" in two dimensions is sometimes defined as the scalar $u_0v_1-u_1v_0$, but this is in substance the $z$ component of the three-dimensional cross product.
```

:::{prf:example} The Cross Product of Basis Vectors
:label: ex-basic-cross-product
If

$$
\mathbf{u}=\begin{bmatrix}1\\0\\0\end{bmatrix},\quad \mathbf{v}=\begin{bmatrix}0\\1\\0\end{bmatrix}
$$

then

$$
\mathbf{u}\times\mathbf{v}=\begin{bmatrix}0\\0\\1\end{bmatrix}
$$

so the result is perpendicular to the $xy$ plane and points along the positive $z$ axis.
:::

```{note} The Geometric Meaning of the Cross Product
:label: note-cross-product-geometry
**Geometric properties.**
- **Perpendicularity**: the cross product $\mathbf{u}\times\mathbf{v}$ is perpendicular to the plane containing $\mathbf{u}$ and $\mathbf{v}$
- **Orientation**: determined by the **right-hand rule**—curl the fingers of the right hand from $\mathbf{u}$ toward $\mathbf{v}$, and the thumb points in the direction of the cross product
- **Magnitude**: the length equals the area of the parallelogram formed by the two vectors

$$
\|\mathbf{u}\times\mathbf{v}\|=\|\mathbf{u}\|\|\mathbf{v}\||\sin\theta|
$$
where $\theta$ is the angle between the two nonzero real vectors. If either vector is zero, the component definition shows that the cross product is zero, and no angle needs to be defined.
```

:::{prf:theorem} The Geometric Formula for the Magnitude of a Cross Product
:label: thm-cross-product-magnitude

Let $\mathbf u,\mathbf v\in\mathbb R^3$ both be nonzero, and let $\theta$ be the angle between them.

$$
\|\mathbf{u}\times\mathbf{v}\|=\|\mathbf{u}\|\|\mathbf{v}\| |\sin\theta|
$$
:::

:::{prf:proof} The Formula for the Magnitude of a Cross Product
1. We must prove that $\|\mathbf{u}\times\mathbf{v}\|^2 = \|\mathbf{u}\|^2 \|\mathbf{v}\|^2 \sin^2\theta$
2. Use the trigonometric identity $\sin^2\theta = 1 - \cos^2\theta$
3. By the geometric meaning of the inner product, $\mathbf{u}\cdot\mathbf{v} = \|\mathbf{u}\|\|\mathbf{v}\|\cos\theta$
4. Therefore $\|\mathbf{u}\|^2 \|\mathbf{v}\|^2 \sin^2\theta = \|\mathbf{u}\|^2 \|\mathbf{v}\|^2 - (\mathbf{u}\cdot\mathbf{v})^2$
5. $\underline{\text{Substituting the definitions of the cross product and the inner product into the identity above and carrying out the algebra verifies it}}$

$\square$
:::


````python
# Here we use symbolic computation to help verify the cross product formula
import sympy as sp
from sympy.vector import CoordSys3D
u0,u1,u2,v0,v1,v2 = sp.symbols("u0,u1,u2,v0,v1,v2",real=True)

N = CoordSys3D('N')
u = u0 * N.i + u1 * N.j + u2 * N.k
v = v0 * N.i + v1 * N.j + v2 * N.k
print("inner product of u and v :",u.dot(v))
print("cross product of u and v :",u.cross(v))

assert sp.simplify(u.cross(v).dot(u.cross(v)) +u.dot(v) * u.dot(v)- u.dot(u) * v.dot(v) ) ==0 ,"The cross product formula is wrong!"
print("The cross product formula is correct!")
````

:::{prf:example} Extending a Line in Three-Dimensional Space to a Perpendicular Plane
:label: line-to-perpendicular-plane

Suppose we have a line $l$ and a plane $S_0$ in three-dimensional space:

**The line** $l$:

$$\mathbf{x} = \mathbf{p} + t\mathbf{d} \quad (-\infty < t < \infty)$$
where $\mathbf{p}$ is a point on the line and $\mathbf{d}$ is its direction vector.

**The plane** $S_0$:

$$\mathbf{x} = \mathbf{q} + s\mathbf{a} + u\mathbf{b} \quad (-\infty < s,u < \infty)$$
whose normal vector is $\mathbf{n_0} = \mathbf{a} \times \mathbf{b}$ (after normalization).

If $\mathbf{d}$ is not parallel to $\mathbf{n_0}$, that is, if $\mathbf{n_0}\times\mathbf{d}\neq\mathbf{0}$, then there is a unique plane $S_1$ that contains $l$ and is perpendicular to $S_0$.


Choose the normal vector of the plane $S_1$ to be

$$\mathbf{n_1} = \mathbf{n_0} \times \mathbf{d}$$

We see that $\mathbf{n_1} \perp \mathbf{n_0}$, so the two planes are perpendicular (which means that the projection of either plane onto the other degenerates to a line, or equivalently that each of them contains a line passing perpendicularly through the other), and that $\mathbf{n_1}$ and $\mathbf{d}$ are linearly independent.

Set $\mathbf{e}= \mathbf{n_1} \times \mathbf{d}$

**The plane** $S_1$:

$$\mathbf{x} = \mathbf{p} + t\mathbf{d} + v\mathbf{e} \quad (-\infty < t,v < \infty)$$
:::

:::{prf:example} Extending Skew Lines to Parallel Planes
:label: skew-lines-parallel-planes

For two skew lines $l_0$ and $l_1$ in three-dimensional space:

**The line** $l_0$: $\mathbf{x} = \mathbf{p_0} + s\mathbf{d_0}$

**The line** $l_1$: $\mathbf{x} = \mathbf{p_1} + t\mathbf{d_1}$

When $\mathbf{d_0} \times \mathbf{d_1} \neq \mathbf{0}$, the two lines may be extended separately to parallel planes:

**The plane** $S_0$ (containing the line $l_0$):

$$\mathbf{x} = \mathbf{p_0} + s\mathbf{d_0} + t\mathbf{d_1} \quad (-\infty < s,t < \infty)$$

**The plane** $S_1$ (containing the line $l_1$):

$$\mathbf{x} = \mathbf{p_1} + s\mathbf{d_0} + t\mathbf{d_1} \quad (-\infty < s,t < \infty)$$

**Proof of parallelism.**
Both planes have normal vector $\mathbf{n} = \mathbf{d_0} \times \mathbf{d_1}$, so $S_0 \parallel S_1$.

**Computing the distance.**
The distance between the two parallel planes is the projected length of the vector $(\mathbf{p_1} - \mathbf{p_0})$ along the normal vector $\mathbf{n}$:

$$d = \frac{|(\mathbf{p_1} - \mathbf{p_0}) \cdot \mathbf{n}|}{\|\mathbf{n}\|} = \frac{|(\mathbf{p_1} - \mathbf{p_0}) \cdot (\mathbf{d_0} \times \mathbf{d_1})|}{\|\mathbf{d_0} \times \mathbf{d_1}\|}$$

This distance is exactly the shortest distance between the two skew lines.
:::

:::{admonition} §2.4 Summary
:class: tip
This section answered the question of whether, besides the inner product, three-dimensional space carries any other natural multiplicative structure. The cross product $\mathbf{u} \times \mathbf{v}$ yields a new vector perpendicular to both $\mathbf{u}$ and $\mathbf{v}$, whose magnitude $\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\|\|\mathbf{v}\||\sin\theta|$ equals the area of the parallelogram the two vectors span, and whose orientation is fixed by the right-hand rule. This geometric reading makes the cross product a natural tool for computing normal vectors, torques, and areas.

The cross product does, however, carry one fundamental restriction: the cross product used in this chapter is the standard three-dimensional one; its component formula does not carry over directly to arbitrary dimensions, and seven dimensions admit a generalization that requires additional structure. This restriction is not a technical defect but a deep feature of three-dimensional geometry—behind it lie the subtle structures of quaternion algebra and Lie groups, a topic touched on further in the historical account opening Chapter 3. This section also displayed the power of the cross product in geometric problems: taking the cross product of the direction vectors as a common normal vector reduces the shortest distance between two skew lines to the distance between two parallel planes, unifying several geometric propositions elegantly. The next section puts all of the vector operations of this chapter into practice through Python code.
:::

## 2.5 Vector Operations in Python

The NumPy library in Python provides powerful facilities for vector operations. Here are some basic examples:



````python
import numpy as np

np.set_printoptions(precision=4, suppress=True, linewidth=100)

# ── Helper functions ────────────────────────────────────────
def print_header(title):
    print("=" * 60)
    print(f"  {title}")
    print("=" * 60)

def print_step(step, desc):
    print(f"\n▶ Step {step}: {desc}")
    print("-" * 40)

def compare_print(label, actual, expected_desc):
    print(f"  {label}")
    print(f"    computed = {actual}")
    print(f"    expected = {expected_desc}")

# ============================================================
print_header("Example 2.5.1 | Basic Vector Operations")
# ============================================================

# --- Step 0 ---
print_step(0, "Create the vectors 𝐚, 𝐛 ∈ ℝ³")
a = np.array([2, 3, 1], dtype=float)
b = np.array([4, -1, 2], dtype=float)
print(f"  𝐚 = {a}")
print(f"  𝐛 = {b}")

# --- Step 1 ---
print_step(1, "Vector addition and scalar multiplication")
compare_print("𝐚 + 𝐛", a + b, "entrywise addition")
compare_print("3𝐚",    3 * a,  "entrywise multiplication by the scalar λ = 3")
compare_print("0.3𝐚 + 0.7𝐛 (segment parameter t = 0.7)",
              0.3*a + 0.7*b,
              "(1−t)𝐚 + t𝐛, t ∈ [0,1]")

# --- Step 2 ---
print_step(2, "Parametric representation of a segment: 𝐩(t) = (1−t)𝐚 + t𝐛")
def line_segment(a, b, t):
    """Return the point on the segment 𝐚𝐛 with parameter t"""
    return (1 - t) * a + t * b

p0 = np.array([1, 2], dtype=float)
p1 = np.array([5, 4], dtype=float)
midpoint = line_segment(p0, p1, 0.5)
compare_print("midpoint 𝐩(0.5)", midpoint, "(𝐩₀ + 𝐩₁) / 2")
````

````python
# ============================================================
print_header("Example 2.5.2 | Inner Product, Norm and the Cauchy-Schwarz Inequality")
# ============================================================

# --- Step 0 ---
print_step(0, "Create the vectors 𝐮, 𝐯 ∈ ℝ³")
u = np.array([1, 2, 3], dtype=float)
v = np.array([4, -1, 2], dtype=float)
print(f"  𝐮 = {u}")
print(f"  𝐯 = {v}")

# --- Step 1 ---
print_step(1, "Three equivalent ways to compute the inner product ⟨𝐮|𝐯⟩ = Σ uᵢvᵢ")
dot1 = np.dot(u, v)
dot2 = u @ v
dot3 = np.inner(u, v)
print(f"  np.dot(u, v)   = {dot1}")
print(f"  u @ v          = {dot2}")
print(f"  np.inner(u, v) = {dot3}")
print(f"  all three agree: {dot1 == dot2 == dot3}")

# --- Step 2 ---
print_step(2, "The L₂ norm and normalization")
def normalize(v):
    """Normalize a vector to the unit vector v̂ = v / ‖v‖"""
    norm = np.linalg.norm(v)
    return v / norm if norm != 0 else v

norm_v = np.linalg.norm(v)
v_hat  = normalize(v)
compare_print("‖𝐯‖₂",       norm_v, "√(Σ vᵢ²)")
compare_print("v̂ = 𝐯/‖𝐯‖₂", v_hat,  "‖v̂‖ should be 1.0")
print(f"  check ‖v̂‖ = {np.linalg.norm(v_hat):.4f}")

# --- Step 3 ---
print_step(3, "Angle between vectors: cos θ = ⟨𝐮|𝐯⟩ / (‖𝐮‖‖𝐯‖)")
def vector_angle(u, v, degrees=True):
    """Compute the angle between two vectors (in degrees by default)"""
    cos_theta = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
    cos_theta = np.clip(cos_theta, -1, 1)          # numerical safeguard
    theta = np.arccos(cos_theta)
    return np.degrees(theta) if degrees else theta

angle = vector_angle(u, v)
print(f"  θ(𝐮, 𝐯) = {angle:.4f}°")

# --- Step 4 ---
print_step(4, "Cauchy-Schwarz inequality |⟨𝐮|𝐯⟩| ≤ ‖𝐮‖‖𝐯‖")
lhs = abs(np.dot(u, v))
rhs = np.linalg.norm(u) * np.linalg.norm(v)
compare_print("|⟨𝐮|𝐯⟩|",    lhs, "left-hand side")
compare_print("‖𝐮‖ · ‖𝐯‖", rhs, "right-hand side (upper bound)")
print(f"  inequality holds: {lhs <= rhs + 1e-10}")

# --- Step 5 ---
print_step(5, "Orthogonal projection Proj_𝐮(𝐯) = (⟨𝐯|𝐮⟩ / ⟨𝐮|𝐮⟩) 𝐮")
def vector_projection(v, u):
    """Compute the orthogonal projection of 𝐯 onto the direction of 𝐮"""
    return (np.dot(v, u) / np.dot(u, u)) * u

proj = vector_projection(v, u)
compare_print("Proj_𝐮(𝐯)", proj, "the projected component; the residual should be orthogonal to 𝐮")
residual = v - proj
print(f"  check orthogonality ⟨𝐮 | 𝐯 − Proj_𝐮(𝐯)⟩ = {np.dot(u, residual):.1e}")
````

````python
# ============================================================
print_header("Example 2.5.3 | The Cross Product and Its Geometric Applications")
# ============================================================

# --- Step 0 ---
print_step(0, "The three-dimensional cross product 𝐚 × 𝐛")
a3d = np.array([1, 2, 3], dtype=float)
b3d = np.array([4, -1, 2], dtype=float)
cross = np.cross(a3d, b3d)
compare_print("𝐚 × 𝐛", cross, "expansion as a determinant: see the definition cross-product")

# --- Step 1 ---
print_step(1, "Verify perpendicularity of the cross product: (𝐚×𝐛)·𝐚 = 0, (𝐚×𝐛)·𝐛 = 0")
print(f"  (𝐚×𝐛)·𝐚 = {np.dot(cross, a3d):.1e}  (should be 0)")
print(f"  (𝐚×𝐛)·𝐛 = {np.dot(cross, b3d):.1e}  (should be 0)")

# --- Step 2 ---
print_step(2, "Area of the parallelogram = ‖𝐚 × 𝐛‖ = ‖𝐚‖‖𝐛‖|sin θ|")
area = np.linalg.norm(cross)
compare_print("‖𝐚 × 𝐛‖", f"{area:.4f}", "area of the parallelogram")

# --- Step 3 ---
print_step(3, "Application: the distance from a point to a line")
def point_to_line_distance(point, line_point, line_dir):
    """Compute the distance from a point in space to a line (using the projection residual)"""
    v    = point - line_point
    proj = vector_projection(v, line_dir)
    return np.linalg.norm(v - proj)

p      = np.array([5, 3, 1], dtype=float)
line_p = np.array([1, 1, 1], dtype=float)
line_d = np.array([1, 1, 0], dtype=float)
dist = point_to_line_distance(p, line_p, line_d)
compare_print("distance from point to line", f"{dist:.4f}", "‖𝐯 − Proj_𝐝(𝐯)‖")

# --- Step 4 ---
print_step(4, "Application: the unit normal vector of a plane (three points determine a plane)")
def plane_normal(q0, q1, q2):
    """Given three points on a plane, compute the unit normal vector 𝐧̂"""
    v0 = q1 - q0
    v1 = q2 - q0
    return normalize(np.cross(v0, v1))

q0 = np.array([1, 0, 0], dtype=float)
q1 = np.array([0, 1, 0], dtype=float)
q2 = np.array([0, 0, 1], dtype=float)
n_hat = plane_normal(q0, q1, q2)
compare_print("𝐧̂ (unit normal vector)", n_hat, "‖𝐧̂‖ should be 1.0")
print(f"  check ‖𝐧̂‖ = {np.linalg.norm(n_hat):.4f}")
````

````python
# ============================================================
print_header("Example 2.5.4 | Gram-Schmidt Orthogonalization (a first look)")
# ============================================================

# --- Step 0 ---
print_step(0, "Given two linearly independent vectors 𝐯₀, 𝐯₁")
v0 = np.array([1.0, 1.0, 0.0])
v1 = np.array([1.0, 0.0, 1.0])
print(f"  𝐯₀ = {v0}")
print(f"  𝐯₁ = {v1}")

# --- Step 1 ---
print_step(1, "Gram-Schmidt: 𝐮₀ = v̂₀; 𝐮₁ = 𝐯₁ − Proj_𝐮₀(𝐯₁), then normalize")
def gram_schmidt(v0, v1):
    """Apply Gram-Schmidt orthogonalization to two vectors, returning the orthonormal pair (𝐮₀, 𝐮₁)"""
    u0 = normalize(v0)
    u1 = normalize(v1 - vector_projection(v1, u0))
    return u0, u1

u0, u1 = gram_schmidt(v0, v1)
print(f"  𝐮₀ = {u0}")
print(f"  𝐮₁ = {u1}")

# --- Step 2 ---
print_step(2, "Verify orthogonality and unit norm")
compare_print("⟨𝐮₀|𝐮₁⟩", f"{np.dot(u0, u1):.1e}", "should be 0 (orthogonal)")
compare_print("‖𝐮₀‖",    f"{np.linalg.norm(u0):.4f}", "should be 1.0 (unit norm)")
compare_print("‖𝐮₁‖",    f"{np.linalg.norm(u1):.4f}", "should be 1.0 (unit norm)")
````

NumPy takes `np.array` as its basic container and provides functions such as `np.dot`, `np.cross`, and `np.linalg.norm` that correspond directly to the definitions of the inner product, the cross product, and the norm, achieving very nearly a line-by-line translation between mathematical notation and code. It is worth noting that a one-dimensional NumPy array does not by default distinguish rows from columns, so shapes require particular care once matrix operations are involved; we shall confront this squarely when we take up matrix multiplication in Chapter 3.

The code also includes a short demonstration of Gram-Schmidt orthogonalization: it uses the projection formula to remove, step by step, the correlation between directions, producing an orthogonal set of vectors. The code is offered here only as a first look; the geometric mechanism behind it—that subtracting the projection removes the correlated component—and the full theory are left to the orthogonalization of Chapter 9 and the singular value decomposition (SVD) of Chapter 11. Through hands-on implementation, abstract geometric intuition and concrete numerical computation come together in this section, preparing the reader to enter the world of matrices.

## 2.6 Chapter Summary

### Review of the Theoretical Thread

This chapter set out from a question that looks like an everyday one: "what is a vector?" The answer given in §2.1 is twofold—a vector is both a geometric arrow and an ordered list of numbers, and once coordinates are introduced the two readings are completely equivalent. This dual identity is not decoration; it is the basic tension running through the whole book: geometric intuition tells us "why," and algebraic structure tells us "how to compute."

§2.2 equipped the set of vectors with two operations, addition and scalar multiplication, and immediately displayed their geometric content: addition is the diagonal of a parallelogram, and scalar multiplication is a stretch along the original direction. More importantly, these two operations gave rise to the notion of a linear combination: given several vectors, taking a weighted sum with arbitrary coefficients lets us "reach" a space spanned jointly by those vectors. The unique-decomposition theorem for the standard basis is the summit of this section: $\mathbf{v} = \sum_{i=0}^{n-1} v_i \mathbf{e}_i$ is not merely a computational formula but the rigorous definition of the notion of a "coordinate."

Addition and scalar multiplication, however, tell us only how to "move," not how to "measure." §2.3 introduced the inner product, hiding all of the information about length and angle inside the algebraic quantity $\langle\mathbf{u}|\mathbf{v}\rangle = \sum_i u_i v_i$. The Cauchy-Schwarz inequality $|\langle\mathbf{u}|\mathbf{v}\rangle| \le \|\mathbf{u}\|\|\mathbf{v}\|$ is the core of that section: it guarantees the legitimacy of the definition of angle in any dimension and makes orthogonal projection a precise geometric operation rather than merely an intuitive impression. §2.4 turned to the cross product, peculiar to three-dimensional space, adding through $\mathbf{u} \times \mathbf{v}$ a mechanism for generating normal vectors; computing the shortest distance between skew lines displayed the surprising economy of the cross product in geometric problems. §2.5 then put all of the above into practice with Python and NumPy, including a first look at a code fragment for Gram-Schmidt orthogonalization as a preview of the theory of orthogonality to come.

### Connections to Other Chapters

There is a clear bridge between this chapter and the language of sets from Chapter 1: set theory supplied precise language for "a set of elements" and for "functions," and this chapter added **operations** on top of that language—vector addition and scalar multiplication—promoting the set $\mathbb{R}^n$ to a "vector space" carrying algebraic structure. That promotion will be formalized completely in the abstract linear spaces of Chapter 4: the eight computational rules observed in §2.2 (commutativity, associativity, distributivity, and the rest) are precisely the prototype of the vector space axioms, and readers will discover there that function spaces and polynomial spaces satisfy the very same axioms, so that the notion of a vector is enormously generalized.

Looking ahead, this chapter supplies almost all of the conceptual material for the matrices and linear transformations of Chapter 3: the change of basis presupposes the unique representation in the standard basis, the geometric meaning of matrix multiplication requires vector addition and scalar multiplication working together, and the introduction of quaternions in the historical opening of Chapter 3 will echo the three-dimensional restriction on the cross product. Further ahead, the real inner product of §2.3 is the finite-dimensional prototype of the inner product spaces of Chapter 9, where the complex inner product and the polarization identity are generalized; and the Gram-Schmidt code fragment first met in §2.5 foreshadows the orthogonalization techniques required by the singular value decomposition of Chapter 11.

### The Role of This Chapter in the Book

This chapter solved a basic and unavoidable problem: before we may speak of matrices, transformations, and decompositions, we must first be clear about what a vector is, what may be done to it, and how it is to be measured. Addition and scalar multiplication establish the linear structure, the inner product establishes the metric structure, and together they constitute the complete framework of Euclidean space. Carrying that framework, a reader who meets the matrices of Chapter 3 is not merely manipulating blocks of numbers but tracking how the coordinates of a vector transform under different bases; and a reader who meets the symmetric matrices of Chapter 9 will see at once that the orthogonality of eigenvectors originates in the geometric meaning of the inner product.

The ubiquity of the vector toolkit in nature, engineering, and information science also gives this chapter a significance beyond the classroom. Force and displacement in physics, spectral components in signal processing, embedding vectors in machine learning—these fields each developed a need for "vectors" independently, in their own contexts, and yet all converged on one and the same mathematical language. Historically, the birth of that language was not smooth: the twenty-year "quaternion war" between Hamilton's quaternion school and Gibbs's vector school produced as its spoils exactly the addition, inner product, and cross product that a linear algebra course now takes for granted. A reader who carries this history forward may come to hold each definition in a little more regard, and to feel a little more impulse to ask why each definition is as it is.
